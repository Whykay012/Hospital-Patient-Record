```python
# ============================================================
# STAGE 3 — DATA CLEANING & PREPARATION
# STEP 3.1 — CREATE CLEANING WORKING COPIES
# ============================================================

# Create independent working copies of the raw datasets
patients_clean = patients.copy()
encounters_clean = encounters.copy()
procedures_clean = procedures.copy()
payers_clean = payers.copy()
organizations_clean = organizations.copy()

# Store cleaned working tables in a dictionary
clean_tables = {
    "patients": patients_clean,
    "encounters": encounters_clean,
    "procedures": procedures_clean,
    "payers": payers_clean,
    "organizations": organizations_clean
}

print("Cleaning working copies created successfully.")
```

```python
clean_tables = {
    "patients": patients_clean,
    "encounters": encounters_clean,
    "procedures": procedures_clean,
    "payers": payers_clean,
    "organizations": organizations_clean
}
```

```python
# ============================================================
# STEP 3.2.1 — CONVERT PATIENT DATE COLUMNS
# ============================================================

patients_clean["BIRTHDATE"] = pd.to_datetime(
    patients_clean["BIRTHDATE"],
    errors="coerce"
)

patients_clean["DEATHDATE"] = pd.to_datetime(
    patients_clean["DEATHDATE"],
    errors="coerce"
)

print("Patient date conversion completed.")
```

```python
# ============================================================
# STEP 3.2.2 — CONVERT ENCOUNTER DATETIME COLUMNS
# ============================================================

encounters_clean["START"] = pd.to_datetime(
    encounters_clean["START"],
    errors="coerce",
    utc=True
)

encounters_clean["STOP"] = pd.to_datetime(
    encounters_clean["STOP"],
    errors="coerce",
    utc=True
)

print("Encounter datetime conversion completed.")
```

```python
# ============================================================
# STEP 3.2.3 — CONVERT PROCEDURE DATETIME COLUMNS
# ============================================================

procedures_clean["START"] = pd.to_datetime(
    procedures_clean["START"],
    errors="coerce",
    utc=True
)

procedures_clean["STOP"] = pd.to_datetime(
    procedures_clean["STOP"],
    errors="coerce",
    utc=True
)

print("Procedure datetime conversion completed.")
```

```python
# ============================================================
# STEP 3.2.4 — VERIFY DATE TYPES
# ============================================================

print("PATIENTS")
print("BIRTHDATE:", patients_clean["BIRTHDATE"].dtype)
print("DEATHDATE:", patients_clean["DEATHDATE"].dtype)

print("\nENCOUNTERS")
print("START:", encounters_clean["START"].dtype)
print("STOP:", encounters_clean["STOP"].dtype)

print("\nPROCEDURES")
print("START:", procedures_clean["START"].dtype)
print("STOP:", procedures_clean["STOP"].dtype)
```

```python
# ============================================================
# STEP 3.2.5 — CHECK FOR INVALID DATE CONVERSIONS
# ============================================================

date_columns = {
    "patients": ["BIRTHDATE", "DEATHDATE"],
    "encounters": ["START", "STOP"],
    "procedures": ["START", "STOP"]
}

for table_name, columns in date_columns.items():

    print(f"\n===== {table_name.upper()} =====")

    df = clean_tables[table_name]

    for column in columns:
        print(
            f"{column}: "
            f"{df[column].isna().sum()} missing"
        )
```

```python
# ============================================================
# STEP 3.2.6 — BEFORE vs AFTER DATE VALIDATION
# ============================================================

date_comparison = {
    "patients.BIRTHDATE": (
        patients["BIRTHDATE"].isna().sum(),
        patients_clean["BIRTHDATE"].isna().sum()
    ),

    "patients.DEATHDATE": (
        patients["DEATHDATE"].isna().sum(),
        patients_clean["DEATHDATE"].isna().sum()
    ),

    "encounters.START": (
        encounters["START"].isna().sum(),
        encounters_clean["START"].isna().sum()
    ),

    "encounters.STOP": (
        encounters["STOP"].isna().sum(),
        encounters_clean["STOP"].isna().sum()
    ),

    "procedures.START": (
        procedures["START"].isna().sum(),
        procedures_clean["START"].isna().sum()
    ),

    "procedures.STOP": (
        procedures["STOP"].isna().sum(),
        procedures_clean["STOP"].isna().sum()
    )
}

for column, (before, after) in date_comparison.items():

    print(
        f"{column}: "
        f"Before = {before}, "
        f"After = {after}"
    )
```

```python
# ============================================================
# STEP 3.3.4 — CHECK ENCOUNTERS AGAINST PATIENT BIRTHDATES
# ============================================================

encounter_patient_dates = encounters_clean[
    ["Id", "PATIENT", "START"]
].merge(
    patients_clean[
        ["Id", "BIRTHDATE"]
    ],
    left_on="PATIENT",
    right_on="Id",
    how="left",
    suffixes=("_ENCOUNTER", "_PATIENT")
)

invalid_birth_encounters = encounter_patient_dates[
    encounter_patient_dates["START"].dt.date <
    encounter_patient_dates["BIRTHDATE"].dt.date
]

print(
    "Encounters occurring before patient birth:",
    len(invalid_birth_encounters)
)
```

```python
# ============================================================
# STEP 3.4.11 — ENCOUNTER DURATION DISTRIBUTION
# ============================================================

duration_ranges = pd.cut(
    encounters_clean["ENCOUNTER_DURATION_MIN"],
    bins=[
        0,
        30,
        60,
        120,
        360,
        1440,
        10080,
        float("inf")
    ],
    labels=[
        "0–30 minutes",
        "31–60 minutes",
        "61–120 minutes",
        "121–360 minutes",
        "6–24 hours",
        "1–7 days",
        "Over 7 days"
    ],
    include_lowest=True
)

duration_distribution = (
    duration_ranges
    .value_counts()
    .sort_index()
    .to_frame("Encounter_Count")
)

duration_distribution["Percentage"] = (
    duration_distribution["Encounter_Count"]
    / len(encounters_clean)
    * 100
)

duration_distribution
```
