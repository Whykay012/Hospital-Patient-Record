Test the Direct Patient/Encounter Consistency

```python
procedure_check = procedures.merge(
    encounters[["Id", "PATIENT"]],
    left_on="ENCOUNTER",
    right_on="Id",
    how="left",
    suffixes=("_procedure", "_encounter")
)

mismatches = procedure_check[
    procedure_check["PATIENT_procedure"] !=
    procedure_check["PATIENT_encounter"]
]

print("Patient/encounter mismatches:", len(mismatches))
```

```python
# ============================================================
# STAGE 3 — DATA CLEANING & PREPARATION
# STEP 3.1 — CREATE CLEANING WORKING COPIES
# ============================================================

# Create independent working copies of the raw datasets
patients_clean = patients.copy()
encounters_clean = encounters.copy()
procedures_clean = procedures.copy()
payers_clean = payers.copy()
organizations_clean = organizations.copy()

# Store cleaned working tables in a dictionary
clean_tables = {
    "patients": patients_clean,
    "encounters": encounters_clean,
    "procedures": procedures_clean,
    "payers": payers_clean,
    "organizations": organizations_clean
}

print("Cleaning working copies created successfully.")
```

```python
# ============================================================================
# STEP 3.13.1 — Missing-Value Audit
# ============================================================================

datasets = {
    "patients_clean": patients_clean,
    "encounters_clean": encounters_clean,
    "procedures_clean": procedures_clean,
    "payers_clean": payers_clean,
    "organizations_clean": organizations_clean
}

missing_value_audit = []

for name, data in datasets.items():

    total_rows = len(data)

    for column in data.columns:

        missing_count = data[column].isna().sum()

        missing_percentage = (
            missing_count / total_rows * 100
            if total_rows > 0 else 0
        )

        missing_value_audit.append({
            "Dataset": name,
            "Column": column,
            "Total_Rows": total_rows,
            "Missing_Count": missing_count,
            "Missing_Percentage": missing_percentage
        })

missing_value_audit = pd.DataFrame(missing_value_audit)

# Show only columns containing missing values
missing_value_audit = (
    missing_value_audit[
        missing_value_audit["Missing_Count"] > 0
    ]
    .sort_values(
        ["Dataset", "Missing_Count"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

missing_value_audit
```

```python
# Summary of missing values by dataset

missing_summary = (
    missing_value_audit
    .groupby("Dataset", as_index=False)
    .agg(
        Columns_With_Missing_Values=("Column", "count"),
        Total_Missing_Values=("Missing_Count", "sum")
    )
)

missing_summary
```

```python
# Complete missing-value audit

print("Missing-value audit completed.")
print(f"Datasets reviewed: {len(datasets)}")
print(f"Columns containing missing values: {len(missing_value_audit)}")
print(
    f"Total missing values identified: "
    f"{missing_value_audit['Missing_Count'].sum():,}"
)
```

```python
# ============================================================================
# STEP 3.13.3 — Missing-Value Treatment
# ============================================================================

# Create explicit copies before treatment
patients_clean = patients_clean.copy()
encounters_clean = encounters_clean.copy()
procedures_clean = procedures_clean.copy()
payers_clean = payers_clean.copy()
organizations_clean = organizations_clean.copy()


# ---------------------------------------------------------------------------
# CATEGORICAL / DESCRIPTIVE FIELDS
# ---------------------------------------------------------------------------

# Patient marital status
patients_clean["MARITAL"] = patients_clean["MARITAL"].fillna("Unknown")


# Payer descriptive fields
payer_missing_columns = [
    "ADDRESS",
    "CITY",
    "STATE_HEADQUARTERED",
    "ZIP",
    "PHONE"
]

for column in payer_missing_columns:
    payers_clean[column] = payers_clean[column].fillna("Unknown")


print("Missing-value treatment applied.")
```

```python
# ============================================================================
# STEP 3.13.4 — Post-Treatment Validation
# ============================================================================

datasets_after_treatment = {
    "patients_clean": patients_clean,
    "encounters_clean": encounters_clean,
    "procedures_clean": procedures_clean,
    "payers_clean": payers_clean,
    "organizations_clean": organizations_clean
}

post_treatment_missing = []

for name, data in datasets_after_treatment.items():

    for column in data.columns:

        missing_count = data[column].isna().sum()

        if missing_count > 0:
            post_treatment_missing.append({
                "Dataset": name,
                "Column": column,
                "Missing_Count": missing_count,
                "Missing_Percentage": round(
                    missing_count / len(data) * 100, 2
                )
            })

post_treatment_missing = (
    pd.DataFrame(post_treatment_missing)
    .sort_values(["Dataset", "Missing_Count"], ascending=[True, False])
    .reset_index(drop=True)
)

post_treatment_missing
```

```python
# ============================================================================
# STEP 3.13.5 — Row-Count Integrity Check
# ============================================================================

expected_row_counts = {
    "patients_clean": 974,
    "encounters_clean": 27891,
    "procedures_clean": 47701,
    "payers_clean": 10,
    "organizations_clean": 1
}

row_count_check = []

for name, expected_count in expected_row_counts.items():

    actual_count = len(datasets_after_treatment[name])

    row_count_check.append({
        "Dataset": name,
        "Expected_Rows": expected_count,
        "Actual_Rows": actual_count,
        "Status": "PASS" if actual_count == expected_count else "FAIL"
    })

row_count_check = pd.DataFrame(row_count_check)

row_count_check
```

```python
# ============================================================================
# STEP 3.13.6 — Missing-Value Treatment Log
# ============================================================================

missing_treatment_log = pd.DataFrame([
    ["encounters_clean", "REASONCODE", 70.06, "Keep missing",
     "Reason information is unavailable for many encounters; no imputation performed."],

    ["encounters_clean", "REASONDESCRIPTION", 70.06, "Keep missing",
     "Reason information is unavailable for many encounters; no imputation performed."],

    ["patients_clean", "SUFFIX", 97.84, "Keep missing",
     "Optional name field; missingness does not affect analysis."],

    ["patients_clean", "DEATHDATE", 84.19, "Keep missing",
     "Missing death date is analytically meaningful and must not be fabricated."],

    ["patients_clean", "AGE_AT_DEATH", 84.19, "Keep missing",
     "Only applicable to patients with a recorded death date."],

    ["patients_clean", "AGE_GROUP", 84.19, "Keep missing",
     "Derived death-related age grouping; no value assigned without death information."],

    ["patients_clean", "AGE_GROUP_AT_DEATH", 84.19, "Keep missing",
     "Only applicable to patients with a recorded death date."],

    ["patients_clean", "MAIDEN", 60.37, "Keep missing",
     "Optional demographic/name field."],

    ["patients_clean", "ZIP", 14.58, "Keep missing",
     "Geographic value unavailable; no value fabricated."],

    ["patients_clean", "MARITAL", 0.10, "Replace with Unknown",
     "Categorical field with one missing observation."],

    ["payers_clean", "ADDRESS", 10.00, "Replace with Unknown",
     "Descriptive payer field; unavailable value retained as Unknown."],

    ["payers_clean", "CITY", 10.00, "Replace with Unknown",
     "Descriptive payer field; unavailable value retained as Unknown."],

    ["payers_clean", "STATE_HEADQUARTERED", 10.00, "Replace with Unknown",
     "Descriptive payer field; unavailable value retained as Unknown."],

    ["payers_clean", "ZIP", 10.00, "Replace with Unknown",
     "Descriptive payer field; unavailable value retained as Unknown."],

    ["payers_clean", "PHONE", 10.00, "Replace with Unknown",
     "Descriptive payer field; unavailable value retained as Unknown."],

    ["procedures_clean", "REASONCODE", 77.45, "Keep missing",
     "Procedure reason is unavailable for many records; no clinical value fabricated."],

    ["procedures_clean", "REASONDESCRIPTION", 77.45, "Keep missing",
     "Procedure reason is unavailable for many records; no clinical value fabricated."]
], columns=[
    "Dataset",
    "Column",
    "Original_Missing_Percentage",
    "Treatment",
    "Justification"
])

missing_treatment_log
```

```python
# ============================================================
# STEP 3.13 — STANDARDIZE DATE/TIME COLUMNS
# ============================================================

# Patient dates
patients_clean["BIRTHDATE"] = pd.to_datetime(
    patients_clean["BIRTHDATE"],
    errors="coerce"
)

patients_clean["DEATHDATE"] = pd.to_datetime(
    patients_clean["DEATHDATE"],
    errors="coerce"
)

# Encounter dates
encounters_clean["START"] = pd.to_datetime(
    encounters_clean["START"],
    errors="coerce"
)

encounters_clean["STOP"] = pd.to_datetime(
    encounters_clean["STOP"],
    errors="coerce"
)

# Procedure dates
procedures_clean["START"] = pd.to_datetime(
    procedures_clean["START"],
    errors="coerce"
)

procedures_clean["STOP"] = pd.to_datetime(
    procedures_clean["STOP"],
    errors="coerce"
)

print("Step 3.13 — Date/time columns converted successfully.")
```

```python
# ============================================================
# STEP 3.14 — ANALYTICAL DATE FEATURES
# ============================================================

# ------------------------------------------------------------
# 1. PATIENT DATE FEATURES
# ------------------------------------------------------------

patients_clean["Birth_Year"] = patients_clean["BIRTHDATE"].dt.year
patients_clean["Birth_Month"] = patients_clean["BIRTHDATE"].dt.month
patients_clean["Birth_Month_Name"] = patients_clean["BIRTHDATE"].dt.month_name()
patients_clean["Birth_Quarter"] = patients_clean["BIRTHDATE"].dt.quarter

patients_clean["Death_Year"] = patients_clean["DEATHDATE"].dt.year
patients_clean["Death_Month"] = patients_clean["DEATHDATE"].dt.month
patients_clean["Death_Month_Name"] = patients_clean["DEATHDATE"].dt.month_name()
patients_clean["Death_Quarter"] = patients_clean["DEATHDATE"].dt.quarter


# ------------------------------------------------------------
# 2. ENCOUNTER DATE FEATURES
# ------------------------------------------------------------

encounters_clean["Encounter_Year"] = encounters_clean["START"].dt.year
encounters_clean["Encounter_Month"] = encounters_clean["START"].dt.month
encounters_clean["Encounter_Month_Name"] = encounters_clean["START"].dt.month_name()
encounters_clean["Encounter_Quarter"] = encounters_clean["START"].dt.quarter

encounters_clean["Encounter_Year_Month"] = (
    encounters_clean["START"].dt.to_period("M").astype(str)
)

encounters_clean["Encounter_Day"] = encounters_clean["START"].dt.day
encounters_clean["Encounter_Day_Name"] = encounters_clean["START"].dt.day_name()

encounters_clean["Encounter_Week"] = (
    encounters_clean["START"].dt.isocalendar().week.astype("Int64")
)


# ------------------------------------------------------------
# 3. PROCEDURE DATE FEATURES
# ------------------------------------------------------------

procedures_clean["Procedure_Year"] = procedures_clean["START"].dt.year
procedures_clean["Procedure_Month"] = procedures_clean["START"].dt.month
procedures_clean["Procedure_Month_Name"] = procedures_clean["START"].dt.month_name()
procedures_clean["Procedure_Quarter"] = procedures_clean["START"].dt.quarter

procedures_clean["Procedure_Year_Month"] = (
    procedures_clean["START"].dt.to_period("M").astype(str)
)

procedures_clean["Procedure_Day"] = procedures_clean["START"].dt.day
procedures_clean["Procedure_Day_Name"] = procedures_clean["START"].dt.day_name()

procedures_clean["Procedure_Week"] = (
    procedures_clean["START"].dt.isocalendar().week.astype("Int64")
)

print("Step 3.14 — Analytical date features created successfully.")
```

```python
# ============================================================
# STEP 3.14 — ANALYTICAL DATE FEATURES
# ============================================================

# ------------------------------------------------------------
# 1. PATIENT DATE FEATURES
# ------------------------------------------------------------

patients_clean["Birth_Year"] = patients_clean["BIRTHDATE"].dt.year

patients_clean["Birth_Month"] = patients_clean["BIRTHDATE"].dt.month

patients_clean["Birth_Month_Name"] = (
    patients_clean["BIRTHDATE"].dt.month_name()
)

patients_clean["Birth_Quarter"] = (
    patients_clean["BIRTHDATE"].dt.quarter
)

# Death-related features
patients_clean["Death_Year"] = patients_clean["DEATHDATE"].dt.year

patients_clean["Death_Month"] = patients_clean["DEATHDATE"].dt.month

patients_clean["Death_Month_Name"] = (
    patients_clean["DEATHDATE"].dt.month_name()
)

patients_clean["Death_Quarter"] = (
    patients_clean["DEATHDATE"].dt.quarter
)


# ------------------------------------------------------------
# 2. ENCOUNTER DATE FEATURES
# ------------------------------------------------------------

encounters_clean["Encounter_Year"] = encounters_clean["START"].dt.year

encounters_clean["Encounter_Month"] = encounters_clean["START"].dt.month

encounters_clean["Encounter_Month_Name"] = (
    encounters_clean["START"].dt.month_name()
)

encounters_clean["Encounter_Quarter"] = (
    encounters_clean["START"].dt.quarter
)

encounters_clean["Encounter_Year_Month"] = (
    encounters_clean["START"].dt.to_period("M").astype(str)
)

encounters_clean["Encounter_Day"] = (
    encounters_clean["START"].dt.day
)

encounters_clean["Encounter_Day_Name"] = (
    encounters_clean["START"].dt.day_name()
)

encounters_clean["Encounter_Week"] = (
    encounters_clean["START"].dt.isocalendar().week.astype("Int64")
)


# ------------------------------------------------------------
# 3. PROCEDURE DATE FEATURES
# ------------------------------------------------------------

procedures_clean["Procedure_Year"] = procedures_clean["START"].dt.year

procedures_clean["Procedure_Month"] = procedures_clean["START"].dt.month

procedures_clean["Procedure_Month_Name"] = (
    procedures_clean["START"].dt.month_name()
)

procedures_clean["Procedure_Quarter"] = (
    procedures_clean["START"].dt.quarter
)

procedures_clean["Procedure_Year_Month"] = (
    procedures_clean["START"].dt.to_period("M").astype(str)
)

procedures_clean["Procedure_Day"] = (
    procedures_clean["START"].dt.day
)

procedures_clean["Procedure_Day_Name"] = (
    procedures_clean["START"].dt.day_name()
)

procedures_clean["Procedure_Week"] = (
    procedures_clean["START"].dt.isocalendar().week.astype("Int64")
)


print("Step 3.14 — Analytical date features created successfully.")
```

```python
# =====================================================
# — PROCEDURE DURATION
# ============================================================

# Calculate procedure duration in minutes
procedures_clean["PROCEDURE_DURATION_MIN"] = (
    procedures_clean["STOP"] - procedures_clean["START"]
).dt.total_seconds() / 60


# Calculate procedure duration in hours
procedures_clean["PROCEDURE_DURATION_HOURS"] = (
    procedures_clean["PROCEDURE_DURATION_MIN"] / 60
)


# Calculate procedure duration in days
procedures_clean["PROCEDURE_DURATION_DAYS"] = (
    procedures_clean["PROCEDURE_DURATION_HOURS"] / 24
)


# Create a duration validation flag
procedures_clean["PROCEDURE_DURATION_FLAG"] = "Valid"

procedures_clean.loc[
    procedures_clean["PROCEDURE_DURATION_MIN"] < 0,
    "PROCEDURE_DURATION_FLAG"
] = "Negative Duration"

procedures_clean.loc[
    procedures_clean["PROCEDURE_DURATION_MIN"].isna(),
    "PROCEDURE_DURATION_FLAG"
] = "Missing Duration"


# Display results
print("Step 3.5 — Procedure duration created successfully.")

display(
    procedures_clean[
        [
            "START",
            "STOP",
            "PROCEDURE_DURATION_MIN",
            "PROCEDURE_DURATION_HOURS",
            "PROCEDURE_DURATION_DAYS",
            "PROCEDURE_DURATION_FLAG"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 3.5 — ENCOUNTER DURATION FEATURE ENGINEERING
# ============================================================

print("=" * 70)
print("STEP 3.5 — ENCOUNTER DURATION FEATURE ENGINEERING")
print("=" * 70)


# ------------------------------------------------------------
# 1. ENSURE START AND STOP ARE DATETIME
# ------------------------------------------------------------

encounters_clean["START"] = pd.to_datetime(
    encounters_clean["START"],
    errors="coerce"
)

encounters_clean["STOP"] = pd.to_datetime(
    encounters_clean["STOP"],
    errors="coerce"
)


# ------------------------------------------------------------
# 2. CALCULATE ENCOUNTER DURATION
# ------------------------------------------------------------

encounters_clean["ENCOUNTER_DURATION_MIN"] = (
    encounters_clean["STOP"] - encounters_clean["START"]
).dt.total_seconds() / 60


encounters_clean["ENCOUNTER_DURATION_HOURS"] = (
    encounters_clean["ENCOUNTER_DURATION_MIN"] / 60
)


encounters_clean["DURATION_DAYS"] = (
    encounters_clean["ENCOUNTER_DURATION_HOURS"] / 24
)


# ------------------------------------------------------------
# 3. CREATE DURATION QUALITY FLAG
# ------------------------------------------------------------

encounters_clean["DURATION_FLAG"] = "Normal"


# Missing duration
encounters_clean.loc[
    encounters_clean["ENCOUNTER_DURATION_HOURS"].isna(),
    "DURATION_FLAG"
] = "Missing"


# Negative duration
encounters_clean.loc[
    encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0,
    "DURATION_FLAG"
] = "Negative"


# Zero duration
encounters_clean.loc[
    encounters_clean["ENCOUNTER_DURATION_HOURS"] == 0,
    "DURATION_FLAG"
] = "Zero"


# Duration greater than 7 days
encounters_clean.loc[
    encounters_clean["DURATION_DAYS"] > 7,
    "DURATION_FLAG"
] = "Over 7 Days"


# ------------------------------------------------------------
# 4. VALIDATION OUTPUT
# ------------------------------------------------------------

print("\nEncounter duration columns created:")

print([
    col
    for col in encounters_clean.columns
    if "DURATION" in col.upper()
])


print("\nDuration flag distribution:")

print(
    encounters_clean["DURATION_FLAG"]
    .value_counts(dropna=False)
)


print("\nStep 3.5 — Encounter duration features created successfully.")
```

```python
# ============================================================
# STEP 3.6 — DURATION VALIDATION
# ============================================================

print("=" * 70)
print("STEP 3.6 — DURATION VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. ENCOUNTER DURATION VALIDATION
# ------------------------------------------------------------

encounter_duration_validation = pd.DataFrame({
    "Validation": [
        "Total encounters",
        "Missing duration",
        "Negative duration",
        "Zero duration",
        "Duration > 7 days"
    ],
    "Count": [
        len(encounters_clean),

        encounters_clean[
            "ENCOUNTER_DURATION_HOURS"
        ].isna().sum(),

        (
            encounters_clean[
                "ENCOUNTER_DURATION_HOURS"
            ] < 0
        ).sum(),

        (
            encounters_clean[
                "ENCOUNTER_DURATION_HOURS"
            ] == 0
        ).sum(),

        (
            encounters_clean[
                "DURATION_DAYS"
            ] > 7
        ).sum()
    ]
})

print("\nENCOUNTER DURATION VALIDATION")
display(encounter_duration_validation)


# ------------------------------------------------------------
# 2. PROCEDURE DURATION VALIDATION
# ------------------------------------------------------------

procedure_duration_validation = pd.DataFrame({
    "Validation": [
        "Total procedures",
        "Missing duration",
        "Negative duration",
        "Zero duration",
        "Duration > 7 days"
    ],
    "Count": [
        len(procedures_clean),

        procedures_clean[
            "PROCEDURE_DURATION_HOURS"
        ].isna().sum(),

        (
            procedures_clean[
                "PROCEDURE_DURATION_HOURS"
            ] < 0
        ).sum(),

        (
            procedures_clean[
                "PROCEDURE_DURATION_HOURS"
            ] == 0
        ).sum(),

        (
            procedures_clean[
                "PROCEDURE_DURATION_DAYS"
            ] > 7
        ).sum()
    ]
})

print("\nPROCEDURE DURATION VALIDATION")
display(procedure_duration_validation)


# ------------------------------------------------------------
# 3. IDENTIFY ENCOUNTER DURATION ANOMALIES
# ------------------------------------------------------------

encounter_duration_anomalies = encounters_clean[
    (
        (encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0) |
        (encounters_clean["DURATION_DAYS"] > 7)
    )
].copy()

print("\nENCOUNTER DURATION ANOMALIES")
print(
    "Number of anomalous encounters:",
    len(encounter_duration_anomalies)
)

if len(encounter_duration_anomalies) > 0:

    display(
        encounter_duration_anomalies[
            [
                "Id",
                "PATIENT",
                "START",
                "STOP",
                "ENCOUNTER_DURATION_HOURS",
                "DURATION_DAYS",
                "DURATION_FLAG"
            ]
        ].sort_values(
            "DURATION_DAYS",
            ascending=False
        ).head(20)
    )


# ------------------------------------------------------------
# 4. IDENTIFY PROCEDURE DURATION ANOMALIES
# ------------------------------------------------------------

procedure_duration_anomalies = procedures_clean[
    (
        (procedures_clean["PROCEDURE_DURATION_HOURS"] < 0) |
        (procedures_clean["PROCEDURE_DURATION_DAYS"] > 7)
    )
].copy()

print("\nPROCEDURE DURATION ANOMALIES")
print(
    "Number of anomalous procedures:",
    len(procedure_duration_anomalies)
)

if len(procedure_duration_anomalies) > 0:

    display(
        procedure_duration_anomalies[
            [
                "PATIENT",
                "ENCOUNTER",
                "START",
                "STOP",
                "PROCEDURE_DURATION_HOURS",
                "PROCEDURE_DURATION_DAYS",
                "PROCEDURE_DURATION_FLAG"
            ]
        ].sort_values(
            "PROCEDURE_DURATION_DAYS",
            ascending=False
        ).head(20)
    )


# ------------------------------------------------------------
# 5. OVERALL VALIDATION STATUS
# ------------------------------------------------------------

encounter_negative = (
    encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0
).sum()

encounter_missing = (
    encounters_clean["ENCOUNTER_DURATION_HOURS"].isna()
).sum()

procedure_negative = (
    procedures_clean["PROCEDURE_DURATION_HOURS"] < 0
).sum()

procedure_missing = (
    procedures_clean["PROCEDURE_DURATION_HOURS"].isna()
).sum()

encounter_long = (
    encounters_clean["DURATION_DAYS"] > 7
).sum()

procedure_long = (
    procedures_clean["PROCEDURE_DURATION_DAYS"] > 7
).sum()


print("\n" + "=" * 70)
print("DURATION VALIDATION SUMMARY")
print("=" * 70)

print(
    "Encounter missing duration:",
    encounter_missing
)

print(
    "Encounter negative duration:",
    encounter_negative
)

print(
    "Encounter >7 days:",
    encounter_long
)

print(
    "Procedure missing duration:",
    procedure_missing
)

print(
    "Procedure negative duration:",
    procedure_negative
)

print(
    "Procedure >7 days:",
    procedure_long
)
```

```python
# ============================================================
# STEP 3.15 — FINAL DATA-QUALITY REPORT
# ============================================================

print("=" * 70)
print("STEP 3.15 — FINAL DATA-QUALITY REPORT")
print("=" * 70)


# ------------------------------------------------------------
# 1. DATASET ROW COUNTS
# ------------------------------------------------------------

dataset_counts = pd.DataFrame({
    "Dataset": [
        "patients_clean",
        "encounters_clean",
        "procedures_clean",
        "payers_clean",
        "organizations_clean"
    ],
    "Rows": [
        len(patients_clean),
        len(encounters_clean),
        len(procedures_clean),
        len(payers_clean),
        len(organizations_clean)
    ]
})

print("\nDATASET ROW COUNTS")
display(dataset_counts)


# ------------------------------------------------------------
# 2. MISSING VALUE SUMMARY
# ------------------------------------------------------------

missing_summary = []

for name, df in {
    "patients_clean": patients_clean,
    "encounters_clean": encounters_clean,
    "procedures_clean": procedures_clean,
    "payers_clean": payers_clean,
    "organizations_clean": organizations_clean
}.items():

    missing_count = int(df.isna().sum().sum())
    columns_with_missing = int((df.isna().sum() > 0).sum())

    missing_summary.append({
        "Dataset": name,
        "Columns_With_Missing": columns_with_missing,
        "Total_Missing": missing_count
    })

missing_summary = pd.DataFrame(missing_summary)

print("\nMISSING VALUE SUMMARY")
display(missing_summary)


# ------------------------------------------------------------
# 3. DUPLICATE SUMMARY
# ------------------------------------------------------------

duplicate_summary = pd.DataFrame({
    "Dataset": [
        "patients_clean",
        "encounters_clean",
        "procedures_clean",
        "payers_clean",
        "organizations_clean"
    ],
    "Duplicate_Rows": [
        patients_clean.duplicated().sum(),
        encounters_clean.duplicated().sum(),
        procedures_clean.duplicated().sum(),
        payers_clean.duplicated().sum(),
        organizations_clean.duplicated().sum()
    ]
})

print("\nDUPLICATE SUMMARY")
display(duplicate_summary)


# ------------------------------------------------------------
# 4. ENCOUNTER DATE VALIDATION
# ------------------------------------------------------------

encounter_date_validation = pd.DataFrame({
    "Validation": [
        "Missing START",
        "Missing STOP",
        "STOP before START"
    ],
    "Count": [
        encounters_clean["START"].isna().sum(),
        encounters_clean["STOP"].isna().sum(),
        (encounters_clean["STOP"] < encounters_clean["START"]).sum()
    ]
})

print("\nENCOUNTER DATE VALIDATION")
display(encounter_date_validation)


# ------------------------------------------------------------
# 5. PROCEDURE DATE VALIDATION
# ------------------------------------------------------------

procedure_date_validation = pd.DataFrame({
    "Validation": [
        "Missing START",
        "Missing STOP",
        "STOP before START"
    ],
    "Count": [
        procedures_clean["START"].isna().sum(),
        procedures_clean["STOP"].isna().sum(),
        (procedures_clean["STOP"] < procedures_clean["START"]).sum()
    ]
})

print("\nPROCEDURE DATE VALIDATION")
display(procedure_date_validation)


# ------------------------------------------------------------
# 6. ENCOUNTER DURATION VALIDATION
# ------------------------------------------------------------

encounter_duration_validation = pd.DataFrame({
    "Validation": [
        "Total encounters",
        "Missing duration",
        "Negative duration",
        "Zero duration",
        "Duration > 7 days"
    ],
    "Count": [
        len(encounters_clean),
        encounters_clean["ENCOUNTER_DURATION_HOURS"].isna().sum(),
        (encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0).sum(),
        (encounters_clean["ENCOUNTER_DURATION_HOURS"] == 0).sum(),
        (encounters_clean["ENCOUNTER_DURATION_HOURS"] > 168).sum()
    ]
})

print("\nENCOUNTER DURATION VALIDATION")
display(encounter_duration_validation)


# ------------------------------------------------------------
# 7. PROCEDURE DURATION VALIDATION
# ------------------------------------------------------------

procedure_duration_validation = pd.DataFrame({
    "Validation": [
        "Total procedures",
        "Missing duration",
        "Negative duration",
        "Zero duration",
        "Duration > 7 days"
    ],
    "Count": [
        len(procedures_clean),
        procedures_clean["PROCEDURE_DURATION_HOURS"].isna().sum(),
        (procedures_clean["PROCEDURE_DURATION_HOURS"] < 0).sum(),
        (procedures_clean["PROCEDURE_DURATION_HOURS"] == 0).sum(),
        (procedures_clean["PROCEDURE_DURATION_HOURS"] > 168).sum()
    ]
})

print("\nPROCEDURE DURATION VALIDATION")
display(procedure_duration_validation)


# ------------------------------------------------------------
# 8. OVERALL QUALITY CHECKS
# ------------------------------------------------------------

quality_checks = pd.DataFrame({
    "Quality_Check": [
        "Expected patient rows",
        "Expected encounter rows",
        "Expected procedure rows",
        "Expected payer rows",
        "Expected organization rows",
        "Encounter negative duration",
        "Procedure negative duration",
        "Encounter missing duration",
        "Procedure missing duration",
        "Encounter duration anomalies",
        "Procedure duration anomalies"
    ],

    "Count": [
        len(patients_clean),
        len(encounters_clean),
        len(procedures_clean),
        len(payers_clean),
        len(organizations_clean),

        (encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0).sum(),
        (procedures_clean["PROCEDURE_DURATION_HOURS"] < 0).sum(),

        encounters_clean["ENCOUNTER_DURATION_HOURS"].isna().sum(),
        procedures_clean["PROCEDURE_DURATION_HOURS"].isna().sum(),

        (encounters_clean["ENCOUNTER_DURATION_HOURS"] > 168).sum(),
        (procedures_clean["PROCEDURE_DURATION_HOURS"] > 168).sum()
    ]
})

print("\nOVERALL QUALITY CHECKS")
display(quality_checks)


# ------------------------------------------------------------
# 9. FINAL STATUS
# ------------------------------------------------------------

critical_errors = (
    (encounters_clean["ENCOUNTER_DURATION_HOURS"] < 0).sum()
    +
    (procedures_clean["PROCEDURE_DURATION_HOURS"] < 0).sum()
    +
    encounters_clean["ENCOUNTER_DURATION_HOURS"].isna().sum()
    +
    procedures_clean["PROCEDURE_DURATION_HOURS"].isna().sum()
)

duration_anomalies = (
    (encounters_clean["ENCOUNTER_DURATION_HOURS"] > 168).sum()
    +
    (procedures_clean["PROCEDURE_DURATION_HOURS"] > 168).sum()
)


if critical_errors == 0:
    final_status = "PASS — No critical duration/date errors detected."
else:
    final_status = "WARNING — Critical data-quality issues remain."


print("\n" + "=" * 70)
print("FINAL DATA-QUALITY STATUS")
print("=" * 70)

print(final_status)
print(f"Duration anomalies requiring analytical awareness: {duration_anomalies}")

print("\nStage 3.15 completed.")
```

```python
# ============================================================
# STAGE 4.1 — STANDARDIZE DATE TIMEZONE BEFORE AGE CALCULATION
# ============================================================

# Make patient birthdates timezone-aware (UTC)
patients_age = patients_clean.copy()

patients_age["BIRTHDATE"] = pd.to_datetime(
    patients_age["BIRTHDATE"],
    errors="coerce",
    utc=True
)

# Make encounter START timezone-aware (UTC)
encounters_clean["START"] = pd.to_datetime(
    encounters_clean["START"],
    errors="coerce",
    utc=True
)

# Create birthdate lookup
birthdate_lookup = patients_age.set_index("Id")["BIRTHDATE"]

# Match each encounter to the patient's birthdate
encounter_birthdates = encounters_clean["PATIENT"].map(
    birthdate_lookup
)

# Calculate age at encounter
encounters_clean["AGE_AT_ENCOUNTER"] = (
    encounters_clean["START"] - encounter_birthdates
).dt.days / 365.25

# Round age
encounters_clean["AGE_AT_ENCOUNTER"] = (
    encounters_clean["AGE_AT_ENCOUNTER"].round(1)
)

print("Age calculation completed successfully.")

encounters_clean[
    ["Id", "PATIENT", "START", "AGE_AT_ENCOUNTER"]
].head(10)
```

```python
# ============================================================
# STEP 4.13B — PATIENT-LEVEL CLINICAL UTILIZATION
# ============================================================

print("=" * 70)
print("STEP 4.13B — PATIENT-LEVEL CLINICAL UTILIZATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Create working copy
# ------------------------------------------------------------

patient_clinical_utilization = encounters_clean[
    ["PATIENT", "ENCOUNTERCLASS"]
].copy()


# ------------------------------------------------------------
# 2. Standardize encounter class for classification
# ------------------------------------------------------------

patient_clinical_utilization["ENCOUNTERCLASS"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    .astype("string")
    .str.strip()
    .str.lower()
)


# ------------------------------------------------------------
# 3. Create encounter-type indicators
# ------------------------------------------------------------

patient_clinical_utilization["Is_Inpatient"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "inpatient"
).astype(int)


patient_clinical_utilization["Is_Emergency"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "emergency"
).astype(int)


patient_clinical_utilization["Is_Outpatient"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "outpatient"
).astype(int)


patient_clinical_utilization["Is_Ambulatory"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "ambulatory"
).astype(int)


patient_clinical_utilization["Is_UrgentCare"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "urgentcare"
).astype(int)


patient_clinical_utilization["Is_Wellness"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    == "wellness"
).astype(int)


# ------------------------------------------------------------
# 4. Define high-acuity encounters
#
# High-acuity = Emergency + Inpatient
# ------------------------------------------------------------

patient_clinical_utilization["Is_High_Acuity"] = (
    patient_clinical_utilization["ENCOUNTERCLASS"]
    .isin(["emergency", "inpatient"])
).astype(int)


# ------------------------------------------------------------
# 5. Aggregate encounters at patient level
# ------------------------------------------------------------

patient_clinical_utilization = (
    patient_clinical_utilization
    .groupby("PATIENT")
    .agg(
        Encounter_Count=("PATIENT", "size"),

        Inpatient_Encounters=("Is_Inpatient", "sum"),

        Emergency_Encounters=("Is_Emergency", "sum"),

        Outpatient_Encounters=("Is_Outpatient", "sum"),

        Ambulatory_Encounters=("Is_Ambulatory", "sum"),

        UrgentCare_Encounters=("Is_UrgentCare", "sum"),

        Wellness_Encounters=("Is_Wellness", "sum"),

        High_Acuity_Encounters=("Is_High_Acuity", "sum")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 6. Calculate utilization rates
# ------------------------------------------------------------

patient_clinical_utilization["Emergency_Rate"] = (
    patient_clinical_utilization["Emergency_Encounters"]
    / patient_clinical_utilization["Encounter_Count"]
    * 100
)


patient_clinical_utilization["Inpatient_Rate"] = (
    patient_clinical_utilization["Inpatient_Encounters"]
    / patient_clinical_utilization["Encounter_Count"]
    * 100
)


patient_clinical_utilization["High_Acuity_Rate"] = (
    patient_clinical_utilization["High_Acuity_Encounters"]
    / patient_clinical_utilization["Encounter_Count"]
    * 100
)


# ------------------------------------------------------------
# 7. Protect against division-by-zero
# ------------------------------------------------------------

rate_columns = [
    "Emergency_Rate",
    "Inpatient_Rate",
    "High_Acuity_Rate"
]

patient_clinical_utilization[rate_columns] = (
    patient_clinical_utilization[rate_columns]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


# ------------------------------------------------------------
# 8. Round utilization rates
# ------------------------------------------------------------

patient_clinical_utilization[rate_columns] = (
    patient_clinical_utilization[rate_columns]
    .round(2)
)


# ------------------------------------------------------------
# 9. Sort by total utilization
# ------------------------------------------------------------

patient_clinical_utilization = (
    patient_clinical_utilization
    .sort_values(
        "Encounter_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 10. Display top 10 patients
# ------------------------------------------------------------

print("\nTOP 10 PATIENTS BY TOTAL ENCOUNTERS")

display(
    patient_clinical_utilization.head(10)
)
```

```python
# ============================================================
# STEP 4.13C — PATIENT-LEVEL UTILIZATION VALIDATION
# ============================================================

print("=" * 70)
print("STEP 4.13C — PATIENT-LEVEL UTILIZATION VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Number of patients
# ------------------------------------------------------------

print(
    "Patients:",
    patient_clinical_utilization["PATIENT"].nunique()
)


# ------------------------------------------------------------
# 2. Total encounters
# ------------------------------------------------------------

print(
    "Total encounters:",
    patient_clinical_utilization["Encounter_Count"].sum()
)


# ------------------------------------------------------------
# 3. Encounter-type totals
# ------------------------------------------------------------

print(
    "Total inpatient encounters:",
    patient_clinical_utilization[
        "Inpatient_Encounters"
    ].sum()
)


print(
    "Total emergency encounters:",
    patient_clinical_utilization[
        "Emergency_Encounters"
    ].sum()
)


print(
    "Total outpatient encounters:",
    patient_clinical_utilization[
        "Outpatient_Encounters"
    ].sum()
)


print(
    "Total ambulatory encounters:",
    patient_clinical_utilization[
        "Ambulatory_Encounters"
    ].sum()
)


print(
    "Total urgent-care encounters:",
    patient_clinical_utilization[
        "UrgentCare_Encounters"
    ].sum()
)


print(
    "Total wellness encounters:",
    patient_clinical_utilization[
        "Wellness_Encounters"
    ].sum()
)


print(
    "Total high-acuity encounters:",
    patient_clinical_utilization[
        "High_Acuity_Encounters"
    ].sum()
)


# ------------------------------------------------------------
# 4. Reconciliation check
# ------------------------------------------------------------

encounter_type_total = (
    patient_clinical_utilization["Inpatient_Encounters"].sum()
    + patient_clinical_utilization["Emergency_Encounters"].sum()
    + patient_clinical_utilization["Outpatient_Encounters"].sum()
    + patient_clinical_utilization["Ambulatory_Encounters"].sum()
    + patient_clinical_utilization["UrgentCare_Encounters"].sum()
    + patient_clinical_utilization["Wellness_Encounters"].sum()
)


total_encounters = (
    patient_clinical_utilization["Encounter_Count"].sum()
)


print("\nENCOUNTER RECONCILIATION")
print("Encounter-type total:", encounter_type_total)
print("Patient-level total:", total_encounters)


if encounter_type_total == total_encounters:
    print("STATUS: PASS")
else:
    print("STATUS: FAIL")


# ------------------------------------------------------------
# 5. High-acuity reconciliation
# ------------------------------------------------------------

expected_high_acuity = (
    patient_clinical_utilization["Inpatient_Encounters"].sum()
    +
    patient_clinical_utilization["Emergency_Encounters"].sum()
)

actual_high_acuity = (
    patient_clinical_utilization["High_Acuity_Encounters"].sum()
)


print("\nHIGH-ACUITY RECONCILIATION")
print("Expected:", expected_high_acuity)
print("Actual:", actual_high_acuity)


if expected_high_acuity == actual_high_acuity:
    print("STATUS: PASS")
else:
    print("STATUS: FAIL")


# ------------------------------------------------------------
# 6. Rate validation
# ------------------------------------------------------------

rate_validation = pd.DataFrame({
    "Metric": [
        "Minimum Emergency Rate",
        "Maximum Emergency Rate",
        "Minimum Inpatient Rate",
        "Maximum Inpatient Rate",
        "Minimum High-Acuity Rate",
        "Maximum High-Acuity Rate"
    ],
    "Value": [
        patient_clinical_utilization["Emergency_Rate"].min(),
        patient_clinical_utilization["Emergency_Rate"].max(),
        patient_clinical_utilization["Inpatient_Rate"].min(),
        patient_clinical_utilization["Inpatient_Rate"].max(),
        patient_clinical_utilization["High_Acuity_Rate"].min(),
        patient_clinical_utilization["High_Acuity_Rate"].max()
    ]
})

print("\nRATE VALIDATION")

display(rate_validation)
```

```python
# ============================================================
# STEP 4.14A — PATIENT-LEVEL FINANCIAL BURDEN
# ============================================================

print("=" * 70)
print("STEP 4.14A — PATIENT-LEVEL FINANCIAL BURDEN")
print("=" * 70)

# ------------------------------------------------------------
# 1. Inspect financial columns first
# ------------------------------------------------------------

print("\nENCOUNTER FINANCIAL COLUMNS:")
print([
    col for col in encounters_clean.columns
    if any(term in col.upper() for term in
           ["COST", "COVERAGE", "RESPONS", "CLAIM"])
])

print("\nPROCEDURE FINANCIAL COLUMNS:")
print([
    col for col in procedures_clean.columns
    if any(term in col.upper() for term in
           ["COST", "COVERAGE", "RESPONS", "CLAIM"])
])
```

```python
# ============================================================
# STEP 4.14A — ENCOUNTER PATIENT RESPONSIBILITY
# ============================================================

print("=" * 70)
print("STEP 4.14A — ENCOUNTER PATIENT RESPONSIBILITY")
print("=" * 70)


# ------------------------------------------------------------
# 1. Ensure financial columns are numeric
# ------------------------------------------------------------

financial_columns = [
    "BASE_ENCOUNTER_COST",
    "TOTAL_CLAIM_COST",
    "PAYER_COVERAGE"
]

for col in financial_columns:
    encounters_clean[col] = pd.to_numeric(
        encounters_clean[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# 2. Calculate patient responsibility
# ------------------------------------------------------------

encounters_clean["PATIENT_RESPONSIBILITY"] = (
    encounters_clean["TOTAL_CLAIM_COST"]
    - encounters_clean["PAYER_COVERAGE"]
)


# ------------------------------------------------------------
# 3. Validate the calculation
# ------------------------------------------------------------

print("\nPATIENT RESPONSIBILITY SUMMARY")

display(
    encounters_clean[
        [
            "TOTAL_CLAIM_COST",
            "PAYER_COVERAGE",
            "PATIENT_RESPONSIBILITY"
        ]
    ].describe()
)


# ------------------------------------------------------------
# 4. Check for negative patient responsibility
# ------------------------------------------------------------

negative_patient_responsibility = (
    encounters_clean["PATIENT_RESPONSIBILITY"] < 0
).sum()

print(
    "\nNegative patient responsibility records:",
    negative_patient_responsibility
)


# ------------------------------------------------------------
# 5. Check missing values
# ------------------------------------------------------------

print(
    "Missing patient responsibility:",
    encounters_clean["PATIENT_RESPONSIBILITY"].isna().sum()
)


print("\nStep 4.14A completed successfully.")
```

```python
# ============================================================
# STEP 4.14B — PATIENT-LEVEL FINANCIAL AGGREGATION
# ============================================================
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.14B — PATIENT-LEVEL FINANCIAL AGGREGATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Aggregate encounter-level financial burden by patient
# ------------------------------------------------------------

patient_financial_burden = (
    encounters_clean
    .groupby("PATIENT")
    .agg(
        Encounter_Count=("PATIENT", "size"),

        Total_Base_Encounter_Cost=(
            "BASE_ENCOUNTER_COST",
            "sum"
        ),

        Total_Claim_Cost=(
            "TOTAL_CLAIM_COST",
            "sum"
        ),

        Total_Payer_Coverage=(
            "PAYER_COVERAGE",
            "sum"
        ),

        Total_Patient_Responsibility=(
            "PATIENT_RESPONSIBILITY",
            "sum"
        )
    )
    .reset_index()
)

# ============================================================
# FINANCIAL RECONCILIATION CHECK
# ============================================================

encounters_clean["FINANCIAL_RECONCILIATION_DIFF"] = (
    encounters_clean["TOTAL_CLAIM_COST"]
    - (
        encounters_clean["PAYER_COVERAGE"]
        + encounters_clean["PATIENT_RESPONSIBILITY"]
    )
)

print("\nFinancial reconciliation check:")

print(
    "Maximum absolute difference:",
    encounters_clean[
        "FINANCIAL_RECONCILIATION_DIFF"
    ].abs().max()
)

print(
    "Records with non-zero difference:",
    (
        encounters_clean["FINANCIAL_RECONCILIATION_DIFF"]
        .abs() > 0.01
    ).sum()
)

# ------------------------------------------------------------
# 2. Aggregate procedure cost by patient
# ------------------------------------------------------------

patient_procedure_cost = (
    procedures_clean
    .groupby("PATIENT")
    .agg(
        Procedure_Count=("PATIENT", "size"),

        Total_Procedure_Cost=(
            "BASE_COST",
            "sum"
        )
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Merge procedure financial information
# ------------------------------------------------------------

patient_financial_burden = patient_financial_burden.merge(
    patient_procedure_cost,
    on="PATIENT",
    how="left"
)


# ------------------------------------------------------------
# 4. Patients without procedures
# ------------------------------------------------------------

patient_financial_burden["Procedure_Count"] = (
    patient_financial_burden["Procedure_Count"]
    .fillna(0)
    .astype(int)
)

patient_financial_burden["Total_Procedure_Cost"] = (
    patient_financial_burden["Total_Procedure_Cost"]
    .fillna(0)
)


# ------------------------------------------------------------
# 5. Calculate average financial burden per encounter
# ------------------------------------------------------------

patient_financial_burden["Average_Cost_Per_Encounter"] = (
    patient_financial_burden["Total_Claim_Cost"]
    / patient_financial_burden["Encounter_Count"]
)

patient_financial_burden["Average_Payer_Coverage_Per_Encounter"] = (
    patient_financial_burden["Total_Payer_Coverage"]
    / patient_financial_burden["Encounter_Count"]
)

patient_financial_burden["Average_Patient_Responsibility_Per_Encounter"] = (
    patient_financial_burden["Total_Patient_Responsibility"]
    / patient_financial_burden["Encounter_Count"]
)


# ------------------------------------------------------------
# 6. Calculate patient responsibility rate
# ------------------------------------------------------------

patient_financial_burden["Patient_Responsibility_Rate"] = (
    patient_financial_burden["Total_Patient_Responsibility"]
    / patient_financial_burden["Total_Claim_Cost"]
    * 100
)


# ------------------------------------------------------------
# 7. Calculate payer coverage rate
# ------------------------------------------------------------

patient_financial_burden["Payer_Coverage_Rate"] = (
    patient_financial_burden["Total_Payer_Coverage"]
    / patient_financial_burden["Total_Claim_Cost"]
    * 100
)


# ------------------------------------------------------------
# 8. Protect against division-by-zero
# ------------------------------------------------------------

rate_columns = [
    "Patient_Responsibility_Rate",
    "Payer_Coverage_Rate"
]

average_columns = [
    "Average_Cost_Per_Encounter",
    "Average_Payer_Coverage_Per_Encounter",
    "Average_Patient_Responsibility_Per_Encounter"
]

patient_financial_burden[rate_columns + average_columns] = (
    patient_financial_burden[rate_columns + average_columns]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


# ------------------------------------------------------------
# 9. Round analytical metrics
# ------------------------------------------------------------

patient_financial_burden[[
    "Total_Base_Encounter_Cost",
    "Total_Claim_Cost",
    "Total_Payer_Coverage",
    "Total_Patient_Responsibility",
    "Total_Procedure_Cost",
    "Average_Cost_Per_Encounter",
    "Average_Payer_Coverage_Per_Encounter",
    "Average_Patient_Responsibility_Per_Encounter"
]] = patient_financial_burden[[
    "Total_Base_Encounter_Cost",
    "Total_Claim_Cost",
    "Total_Payer_Coverage",
    "Total_Patient_Responsibility",
    "Total_Procedure_Cost",
    "Average_Cost_Per_Encounter",
    "Average_Payer_Coverage_Per_Encounter",
    "Average_Patient_Responsibility_Per_Encounter"
]].round(2)


# ------------------------------------------------------------
# 10. Sort by total patient responsibility
# ------------------------------------------------------------

patient_financial_burden = (
    patient_financial_burden
    .sort_values(
        "Total_Patient_Responsibility",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
#  TOP 20% PATIENT COST CONCENTRATION
# ------------------------------------------------------------

top_20_count = max(
    1,
    int(np.ceil(len(patient_financial_burden) * 0.20))
)

patient_top20_percent_share = (
    patient_financial_burden
    .sort_values(
        "Total_Claim_Cost",
        ascending=False
    )
    .head(top_20_count)["Total_Claim_Cost"]
    .sum()
    / patient_financial_burden["Total_Claim_Cost"].sum()
    * 100
)

print(
    f"\nTop 20% of patients account for "
    f"{patient_top20_percent_share:.2f}% "
    f"of total claim cost."
)
# ------------------------------------------------------------
# 11. Display top 10
# ------------------------------------------------------------

print("\nTOP 10 PATIENTS BY TOTAL PATIENT RESPONSIBILITY")

display(
    patient_financial_burden.head(10)
)
```

```python
# ============================================================
# STEP 4.14C — PATIENT-LEVEL FINANCIAL VALIDATION
# ============================================================
import numpy as np

print("=" * 70)
print("STEP 4.14C — PATIENT-LEVEL FINANCIAL VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# Patient count
# ------------------------------------------------------------

print(
    "\nPatients:",
    len(patient_financial_burden)
)


# ------------------------------------------------------------
# Encounter reconciliation
# ------------------------------------------------------------

print(
    "Total encounters:",
    patient_financial_burden["Encounter_Count"].sum()
)

print(
    "Expected encounters:",
    len(encounters_clean)
)


# ------------------------------------------------------------
# Financial reconciliation
# ------------------------------------------------------------

print("\nFINANCIAL RECONCILIATION")

financial_validation = pd.DataFrame({
    "Metric": [
        "Total Base Encounter Cost",
        "Total Claim Cost",
        "Total Payer Coverage",
        "Total Patient Responsibility",
        "Total Procedure Cost"
    ],

    "Patient_Level_Total": [
        patient_financial_burden[
            "Total_Base_Encounter_Cost"
        ].sum(),

        patient_financial_burden[
            "Total_Claim_Cost"
        ].sum(),

        patient_financial_burden[
            "Total_Payer_Coverage"
        ].sum(),

        patient_financial_burden[
            "Total_Patient_Responsibility"
        ].sum(),

        patient_financial_burden[
            "Total_Procedure_Cost"
        ].sum()
    ],

    "Source_Level_Total": [
        encounters_clean[
            "BASE_ENCOUNTER_COST"
        ].sum(),

        encounters_clean[
            "TOTAL_CLAIM_COST"
        ].sum(),

        encounters_clean[
            "PAYER_COVERAGE"
        ].sum(),

        encounters_clean[
            "PATIENT_RESPONSIBILITY"
        ].sum(),

        procedures_clean[
            "BASE_COST"
        ].sum()
    ]
})


financial_validation["Difference"] = (
    financial_validation["Patient_Level_Total"]
    - financial_validation["Source_Level_Total"]
)

financial_validation["Status"] = np.where(
    np.isclose(
        financial_validation["Patient_Level_Total"],
        financial_validation["Source_Level_Total"],
        rtol=1e-9,
        atol=1e-2
    ),
    "PASS",
    "FAIL"
)

display(financial_validation)


# ------------------------------------------------------------
# Rate validation
# ------------------------------------------------------------

print("\nRATE VALIDATION")

rate_validation = pd.DataFrame({
    "Metric": [
        "Minimum Patient Responsibility Rate",
        "Maximum Patient Responsibility Rate",
        "Minimum Payer Coverage Rate",
        "Maximum Payer Coverage Rate"
    ],

    "Value": [
        patient_financial_burden[
            "Patient_Responsibility_Rate"
        ].min(),

        patient_financial_burden[
            "Patient_Responsibility_Rate"
        ].max(),

        patient_financial_burden[
            "Payer_Coverage_Rate"
        ].min(),

        patient_financial_burden[
            "Payer_Coverage_Rate"
        ].max()
    ]
})

display(rate_validation)
```

```python
# ======================================================================
# STEP 4.15A — DEMOGRAPHIC × UTILIZATION PREPARATION
# ======================================================================

print("=" * 70)
print("STEP 4.15A — DEMOGRAPHIC × UTILIZATION PREPARATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Confirm required source datasets exist
# ----------------------------------------------------------------------

required_objects = [
    "patients_clean",
    "patient_clinical_utilization"
]

for obj in required_objects:
    if obj not in globals():
        raise NameError(
            f"Required dataset '{obj}' is not available. "
            f"Run the earlier patient-level analysis first."
        )


# ----------------------------------------------------------------------
# 2. Display available patient demographic columns
# ----------------------------------------------------------------------

demographic_candidates = [
    "GENDER",
    "RACE",
    "ETHNICITY",
    "MARITAL",
    "AGE_GROUP"
]

available_demographics = [
    col for col in demographic_candidates
    if col in patients_clean.columns
]

missing_demographics = [
    col for col in demographic_candidates
    if col not in patients_clean.columns
]

print("\nAVAILABLE DEMOGRAPHIC COLUMNS:")
print(available_demographics)

if missing_demographics:
    print("\nNOT FOUND:")
    print(missing_demographics)


# ----------------------------------------------------------------------
# 3. Confirm patient-level utilization columns
# ----------------------------------------------------------------------

utilization_columns = [
    "PATIENT",
    "Encounter_Count",
    "Inpatient_Encounters",
    "Emergency_Encounters",
    "High_Acuity_Encounters",
    "Emergency_Rate",
    "High_Acuity_Rate"
]

missing_utilization = [
    col for col in utilization_columns
    if col not in patient_clinical_utilization.columns
]

if missing_utilization:
    raise KeyError(
        f"Missing utilization columns: {missing_utilization}"
    )

print("\nUTILIZATION COLUMNS VERIFIED:")
print(utilization_columns)


# ----------------------------------------------------------------------
# 4. Select demographic columns that actually exist
# ----------------------------------------------------------------------

demographic_columns = available_demographics.copy()

if not demographic_columns:
    raise ValueError(
        "None of the expected demographic columns were found "
        "in patients_clean."
    )


# ----------------------------------------------------------------------
# 5. Create a clean demographic source table
# ----------------------------------------------------------------------

patient_demographics = patients_clean[
    ["Id"] + demographic_columns
].copy()

patient_demographics = patient_demographics.rename(
    columns={"Id": "PATIENT"}
)


# ----------------------------------------------------------------------
# 6. Standardize categorical demographic values
# ----------------------------------------------------------------------

for col in demographic_columns:

    patient_demographics[col] = (
        patient_demographics[col]
        .astype("string")
        .str.strip()
    )

    patient_demographics[col] = (
        patient_demographics[col]
        .replace(
            {
                "": "Unknown",
                "nan": "Unknown",
                "None": "Unknown",
                "<NA>": "Unknown"
            }
        )
        .fillna("Unknown")
    )


# ----------------------------------------------------------------------
# 7. Merge demographics with validated patient-level utilization
# ----------------------------------------------------------------------

patient_demographic_utilization = patient_demographics.merge(
    patient_clinical_utilization,
    on="PATIENT",
    how="inner",
    validate="one_to_one"
)


# ----------------------------------------------------------------------
# 8. Basic merge validation
# ----------------------------------------------------------------------

print("\nMERGE VALIDATION")
print("-" * 70)

print(
    "Patients in patients_clean:",
    patients_clean["Id"].nunique()
)

print(
    "Patients in utilization table:",
    patient_clinical_utilization["PATIENT"].nunique()
)

print(
    "Patients after demographic/utilization merge:",
    patient_demographic_utilization["PATIENT"].nunique()
)


# ----------------------------------------------------------------------
# 9. Display resulting structure
# ----------------------------------------------------------------------

print("\nDEMOGRAPHIC × UTILIZATION COLUMNS:")
print(patient_demographic_utilization.columns.tolist())


# ----------------------------------------------------------------------
# 10. Preview
# ----------------------------------------------------------------------

patient_demographic_utilization.head()
```

```python
# ======================================================================
# STEP 4.15B — DEMOGRAPHIC × UTILIZATION AGGREGATION
# ======================================================================

print("=" * 70)
print("STEP 4.15B — DEMOGRAPHIC × UTILIZATION AGGREGATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Define aggregation function
# ----------------------------------------------------------------------

def create_demographic_utilization_summary(
    dataframe,
    demographic_column
):

    summary = (
        dataframe
        .groupby(demographic_column, dropna=False)
        .agg(
            Patient_Count=("PATIENT", "nunique"),

            Total_Encounters=("Encounter_Count", "sum"),

            Average_Encounters_Per_Patient=(
                "Encounter_Count",
                "mean"
            ),

            Median_Encounters_Per_Patient=(
                "Encounter_Count",
                "median"
            ),

            Total_Inpatient_Encounters=(
                "Inpatient_Encounters",
                "sum"
            ),

            Total_Emergency_Encounters=(
                "Emergency_Encounters",
                "sum"
            ),

            Total_High_Acuity_Encounters=(
                "High_Acuity_Encounters",
                "sum"
            ),

            Average_Emergency_Rate=(
                "Emergency_Rate",
                "mean"
            ),

            Average_High_Acuity_Rate=(
                "High_Acuity_Rate",
                "mean"
            )
        )
        .reset_index()
    )


    # ------------------------------------------------------------------
    # Calculate group-level rates
    # ------------------------------------------------------------------

    summary["Emergency_Rate"] = (
        summary["Total_Emergency_Encounters"]
        / summary["Total_Encounters"]
        * 100
    )


    summary["High_Acuity_Rate"] = (
        summary["Total_High_Acuity_Encounters"]
        / summary["Total_Encounters"]
        * 100
    )


    # ------------------------------------------------------------------
    # Replace invalid mathematical results
    # ------------------------------------------------------------------

    rate_columns = [
        "Emergency_Rate",
        "High_Acuity_Rate",
        "Average_Emergency_Rate",
        "Average_High_Acuity_Rate"
    ]

    summary[rate_columns] = (
        summary[rate_columns]
        .replace([float("inf"), -float("inf")], 0)
        .fillna(0)
    )


    # ------------------------------------------------------------------
    # Sort by total utilization
    # ------------------------------------------------------------------

    summary = (
        summary
        .sort_values(
            "Total_Encounters",
            ascending=False
        )
        .reset_index(drop=True)
    )


    return summary


# ----------------------------------------------------------------------
# 2. Create demographic summaries
# ----------------------------------------------------------------------

demographic_utilization_summaries = {}


for demographic in demographic_columns:

    demographic_utilization_summaries[demographic] = (
        create_demographic_utilization_summary(
            patient_demographic_utilization,
            demographic
        )
    )


# ----------------------------------------------------------------------
# 3. Create convenient named tables
# ----------------------------------------------------------------------

if "GENDER" in demographic_utilization_summaries:
    gender_utilization = demographic_utilization_summaries["GENDER"]

if "RACE" in demographic_utilization_summaries:
    race_utilization = demographic_utilization_summaries["RACE"]

if "ETHNICITY" in demographic_utilization_summaries:
    ethnicity_utilization = demographic_utilization_summaries["ETHNICITY"]

if "MARITAL" in demographic_utilization_summaries:
    marital_utilization = demographic_utilization_summaries["MARITAL"]

if "AGE_GROUP" in demographic_utilization_summaries:
    age_group_utilization = demographic_utilization_summaries["AGE_GROUP"]


# ----------------------------------------------------------------------
# 4. Display each demographic analysis
# ----------------------------------------------------------------------

for demographic, summary in demographic_utilization_summaries.items():

    print("\n")
    print("=" * 70)
    print(f"{demographic} × UTILIZATION")
    print("=" * 70)

    display(summary)


# ----------------------------------------------------------------------
# 5. Identify groups with highest utilization
# ----------------------------------------------------------------------

print("\n")
print("=" * 70)
print("HIGHEST-UTILIZATION GROUPS")
print("=" * 70)


for demographic, summary in demographic_utilization_summaries.items():

    if len(summary) == 0:
        continue

    highest_group = summary.iloc[0]

    print(
        f"\n{demographic}: "
        f"{highest_group[demographic]} "
        f"→ "
        f"{highest_group['Total_Encounters']:,.0f} encounters "
        f"across "
        f"{highest_group['Patient_Count']:,.0f} patients"
    )
```

```python
# ======================================================================
# STEP 4.15C — DEMOGRAPHIC × UTILIZATION VALIDATION
# ======================================================================

print("=" * 70)
print("STEP 4.15C — DEMOGRAPHIC × UTILIZATION VALIDATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Patient-level merge validation
# ----------------------------------------------------------------------

source_patient_count = patients_clean["Id"].nunique()

utilization_patient_count = (
    patient_clinical_utilization["PATIENT"].nunique()
)

merged_patient_count = (
    patient_demographic_utilization["PATIENT"].nunique()
)


print("\nPATIENT COUNT VALIDATION")
print("-" * 70)

print("Patients in patients_clean:", source_patient_count)
print("Patients in utilization:", utilization_patient_count)
print("Patients after merge:", merged_patient_count)


patient_count_status = (
    source_patient_count
    == utilization_patient_count
    == merged_patient_count
)

print(
    "STATUS:",
    "PASS" if patient_count_status else "FAIL"
)


# ----------------------------------------------------------------------
# 2. Encounter reconciliation
# ----------------------------------------------------------------------

source_encounter_total = (
    patient_clinical_utilization["Encounter_Count"].sum()
)

demographic_encounter_results = []


for demographic, summary in demographic_utilization_summaries.items():

    grouped_total = summary["Total_Encounters"].sum()

    difference = grouped_total - source_encounter_total

    demographic_encounter_results.append(
        {
            "Demographic": demographic,
            "Patient_Level_Total": grouped_total,
            "Source_Level_Total": source_encounter_total,
            "Difference": difference,
            "Status": (
                "PASS"
                if abs(difference) < 1e-9
                else "FAIL"
            )
        }
    )


demographic_encounter_validation = pd.DataFrame(
    demographic_encounter_results
)


print("\nENCOUNTER RECONCILIATION")
print("-" * 70)

display(demographic_encounter_validation)


# ----------------------------------------------------------------------
# 3. High-acuity reconciliation
# ----------------------------------------------------------------------

source_high_acuity_total = (
    patient_clinical_utilization[
        "High_Acuity_Encounters"
    ].sum()
)

high_acuity_results = []


for demographic, summary in demographic_utilization_summaries.items():

    grouped_total = (
        summary["Total_High_Acuity_Encounters"]
        .sum()
    )

    difference = (
        grouped_total
        - source_high_acuity_total
    )

    high_acuity_results.append(
        {
            "Demographic": demographic,
            "Patient_Level_Total": grouped_total,
            "Source_Level_Total": source_high_acuity_total,
            "Difference": difference,
            "Status": (
                "PASS"
                if abs(difference) < 1e-9
                else "FAIL"
            )
        }
    )


high_acuity_validation = pd.DataFrame(
    high_acuity_results
)


print("\nHIGH-ACUITY RECONCILIATION")
print("-" * 70)

display(high_acuity_validation)


# ----------------------------------------------------------------------
# 4. Emergency reconciliation
# ----------------------------------------------------------------------

source_emergency_total = (
    patient_clinical_utilization[
        "Emergency_Encounters"
    ].sum()
)

emergency_results = []


for demographic, summary in demographic_utilization_summaries.items():

    grouped_total = (
        summary["Total_Emergency_Encounters"]
        .sum()
    )

    difference = (
        grouped_total
        - source_emergency_total
    )

    emergency_results.append(
        {
            "Demographic": demographic,
            "Patient_Level_Total": grouped_total,
            "Source_Level_Total": source_emergency_total,
            "Difference": difference,
            "Status": (
                "PASS"
                if abs(difference) < 1e-9
                else "FAIL"
            )
        }
    )


emergency_validation = pd.DataFrame(
    emergency_results
)


print("\nEMERGENCY RECONCILIATION")
print("-" * 70)

display(emergency_validation)


# ----------------------------------------------------------------------
# 5. Validate rate ranges
# ----------------------------------------------------------------------

rate_validation = pd.DataFrame(
    {
        "Metric": [
            "Minimum Emergency Rate",
            "Maximum Emergency Rate",
            "Minimum High-Acuity Rate",
            "Maximum High-Acuity Rate"
        ],
        "Value": [
            patient_demographic_utilization[
                "Emergency_Rate"
            ].min(),

            patient_demographic_utilization[
                "Emergency_Rate"
            ].max(),

            patient_demographic_utilization[
                "High_Acuity_Rate"
            ].min(),

            patient_demographic_utilization[
                "High_Acuity_Rate"
            ].max()
        ]
    }
)


print("\nRATE VALIDATION")
print("-" * 70)

display(rate_validation)


# ----------------------------------------------------------------------
# 6. Final validation status
# ----------------------------------------------------------------------

all_encounter_pass = (
    demographic_encounter_validation["Status"]
    .eq("PASS")
    .all()
)

all_high_acuity_pass = (
    high_acuity_validation["Status"]
    .eq("PASS")
    .all()
)

all_emergency_pass = (
    emergency_validation["Status"]
    .eq("PASS")
    .all()
)

rates_valid = (
    rate_validation["Value"].between(0, 100).all()
)


overall_status = all(
    [
        patient_count_status,
        all_encounter_pass,
        all_high_acuity_pass,
        all_emergency_pass,
        rates_valid
    ]
)


print("\n")
print("=" * 70)
print("STEP 4.15 VALIDATION SUMMARY")
print("=" * 70)

print(
    "Patient count validation:",
    "PASS" if patient_count_status else "FAIL"
)

print(
    "Encounter reconciliation:",
    "PASS" if all_encounter_pass else "FAIL"
)

print(
    "High-acuity reconciliation:",
    "PASS" if all_high_acuity_pass else "FAIL"
)

print(
    "Emergency reconciliation:",
    "PASS" if all_emergency_pass else "FAIL"
)

print(
    "Rate validation:",
    "PASS" if rates_valid else "FAIL"
)

print("-" * 70)

print(
    "OVERALL STATUS:",
    "PASS" if overall_status else "FAIL"
)
```

```python
# ======================================================================
# STEP 4.16A — ENCOUNTER TYPE × COST PREPARATION
# ======================================================================

print("=" * 70)
print("STEP 4.16A — ENCOUNTER TYPE × COST PREPARATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Confirm required dataset exists
# ----------------------------------------------------------------------

if "encounters_clean" not in globals():
    raise NameError(
        "encounters_clean is not available. "
        "Run the encounter cleaning steps first."
    )


# ----------------------------------------------------------------------
# 2. Required columns
# ----------------------------------------------------------------------

required_columns = [
    "ENCOUNTERCLASS",
    "BASE_ENCOUNTER_COST",
    "TOTAL_CLAIM_COST",
    "PAYER_COVERAGE",
    "PATIENT_RESPONSIBILITY"
]

missing_columns = [
    col for col in required_columns
    if col not in encounters_clean.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )


print("\nREQUIRED FINANCIAL COLUMNS VERIFIED:")
print(required_columns)


# ----------------------------------------------------------------------
# 3. Create analytical copy
# ----------------------------------------------------------------------

encounter_type_cost = encounters_clean[
    required_columns
].copy()


# ----------------------------------------------------------------------
# 4. Standardize encounter class
# ----------------------------------------------------------------------

encounter_type_cost["ENCOUNTERCLASS"] = (
    encounter_type_cost["ENCOUNTERCLASS"]
    .astype("string")
    .str.strip()
    .str.lower()
    .fillna("unknown")
)


# ----------------------------------------------------------------------
# 5. Convert financial columns to numeric
# ----------------------------------------------------------------------

financial_columns = [
    "BASE_ENCOUNTER_COST",
    "TOTAL_CLAIM_COST",
    "PAYER_COVERAGE",
    "PATIENT_RESPONSIBILITY"
]

for col in financial_columns:
    encounter_type_cost[col] = pd.to_numeric(
        encounter_type_cost[col],
        errors="coerce"
    )


# ----------------------------------------------------------------------
# 6. Check missing financial values
# ----------------------------------------------------------------------

print("\nMISSING FINANCIAL VALUES:")
print(
    encounter_type_cost[financial_columns]
    .isna()
    .sum()
)


# ----------------------------------------------------------------------
# 7. Display encounter class distribution
# ----------------------------------------------------------------------

print("\nENCOUNTER CLASS DISTRIBUTION:")
display(
    encounter_type_cost["ENCOUNTERCLASS"]
    .value_counts()
    .rename_axis("Encounter_Class")
    .reset_index(name="Encounter_Count")
)


# ----------------------------------------------------------------------
# 8. Preview
# ----------------------------------------------------------------------

print("\nENCOUNTER TYPE × COST SOURCE:")
display(encounter_type_cost.head())
```

```python
# ======================================================================
# STEP 4.16B — ENCOUNTER TYPE × COST AGGREGATION
# ======================================================================

print("=" * 70)
print("STEP 4.16B — ENCOUNTER TYPE × COST AGGREGATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Aggregate financial metrics by encounter class
# ----------------------------------------------------------------------

encounter_type_cost_summary = (
    encounter_type_cost
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("ENCOUNTERCLASS", "size"),

        Total_Base_Encounter_Cost=(
            "BASE_ENCOUNTER_COST",
            "sum"
        ),

        Total_Claim_Cost=(
            "TOTAL_CLAIM_COST",
            "sum"
        ),

        Total_Payer_Coverage=(
            "PAYER_COVERAGE",
            "sum"
        ),

        Total_Patient_Responsibility=(
            "PATIENT_RESPONSIBILITY",
            "sum"
        ),

        Average_Base_Encounter_Cost=(
            "BASE_ENCOUNTER_COST",
            "mean"
        ),

        Average_Claim_Cost=(
            "TOTAL_CLAIM_COST",
            "mean"
        ),

        Average_Payer_Coverage=(
            "PAYER_COVERAGE",
            "mean"
        ),

        Average_Patient_Responsibility=(
            "PATIENT_RESPONSIBILITY",
            "mean"
        )
    )
    .reset_index()
)


# ----------------------------------------------------------------------
# 2. Calculate financial rates
# ----------------------------------------------------------------------

encounter_type_cost_summary["Patient_Responsibility_Rate"] = (
    encounter_type_cost_summary[
        "Total_Patient_Responsibility"
    ]
    /
    encounter_type_cost_summary[
        "Total_Claim_Cost"
    ]
    * 100
)


encounter_type_cost_summary["Payer_Coverage_Rate"] = (
    encounter_type_cost_summary[
        "Total_Payer_Coverage"
    ]
    /
    encounter_type_cost_summary[
        "Total_Claim_Cost"
    ]
    * 100
)


# ----------------------------------------------------------------------
# 3. Calculate cost contribution
# ----------------------------------------------------------------------

total_claim_cost = (
    encounter_type_cost_summary[
        "Total_Claim_Cost"
    ].sum()
)


encounter_type_cost_summary["Claim_Cost_Share"] = (
    encounter_type_cost_summary[
        "Total_Claim_Cost"
    ]
    /
    total_claim_cost
    * 100
)


# ----------------------------------------------------------------------
# 4. Safely handle invalid rates
# ----------------------------------------------------------------------

rate_columns = [
    "Patient_Responsibility_Rate",
    "Payer_Coverage_Rate",
    "Claim_Cost_Share"
]

encounter_type_cost_summary[rate_columns] = (
    encounter_type_cost_summary[rate_columns]
    .replace([float("inf"), -float("inf")], 0)
    .fillna(0)
)


# ----------------------------------------------------------------------
# 5. Sort by total claim cost
# ----------------------------------------------------------------------

encounter_type_cost_summary = (
    encounter_type_cost_summary
    .sort_values(
        "Total_Claim_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ----------------------------------------------------------------------
# 6. Display full result
# ----------------------------------------------------------------------

print("\nENCOUNTER TYPE × COST SUMMARY")
print("-" * 70)

display(encounter_type_cost_summary)
```

```python
# ======================================================================
# STEP 4.16C — ENCOUNTER TYPE × COST VALIDATION
# ======================================================================

print("=" * 70)
print("STEP 4.16C — ENCOUNTER TYPE × COST VALIDATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. SOURCE-LEVEL TOTALS
# ----------------------------------------------------------------------

source_totals = {
    "Encounter_Count": len(encounters_clean),

    "Total_Base_Encounter_Cost": (
        encounters_clean["BASE_ENCOUNTER_COST"].sum()
    ),

    "Total_Claim_Cost": (
        encounters_clean["TOTAL_CLAIM_COST"].sum()
    ),

    "Total_Payer_Coverage": (
        encounters_clean["PAYER_COVERAGE"].sum()
    ),

    "Total_Patient_Responsibility": (
        encounters_clean["PATIENT_RESPONSIBILITY"].sum()
    )
}


# ----------------------------------------------------------------------
# 2. GROUPED TOTALS
# ----------------------------------------------------------------------

grouped_totals = {
    "Encounter_Count": (
        encounter_type_cost_summary["Encounter_Count"].sum()
    ),

    "Total_Base_Encounter_Cost": (
        encounter_type_cost_summary["Total_Base_Encounter_Cost"].sum()
    ),

    "Total_Claim_Cost": (
        encounter_type_cost_summary["Total_Claim_Cost"].sum()
    ),

    "Total_Payer_Coverage": (
        encounter_type_cost_summary["Total_Payer_Coverage"].sum()
    ),

    "Total_Patient_Responsibility": (
        encounter_type_cost_summary["Total_Patient_Responsibility"].sum()
    )
}


# ----------------------------------------------------------------------
# 3. FINANCIAL RECONCILIATION
# ----------------------------------------------------------------------

validation_rows = []

for metric in source_totals:

    source_value = source_totals[metric]
    grouped_value = grouped_totals[metric]

    difference = grouped_value - source_value

    validation_rows.append({
        "Metric": metric,
        "Grouped_Total": grouped_value,
        "Source_Total": source_value,
        "Difference": difference,
        "Status": (
            "PASS"
            if abs(difference) < 0.01
            else "FAIL"
        )
    })


encounter_type_cost_validation = pd.DataFrame(
    validation_rows
)


print("\nFINANCIAL RECONCILIATION")
print("-" * 70)

display(encounter_type_cost_validation)


# ----------------------------------------------------------------------
# 4. ENCOUNTER COUNT RECONCILIATION
# ----------------------------------------------------------------------

source_encounter_count = len(encounters_clean)

grouped_encounter_count = (
    encounter_type_cost_summary["Encounter_Count"].sum()
)


print("\nENCOUNTER COUNT RECONCILIATION")
print("-" * 70)

print(
    "Source encounters:",
    source_encounter_count
)

print(
    "Grouped encounters:",
    grouped_encounter_count
)

print(
    "STATUS:",
    "PASS"
    if source_encounter_count == grouped_encounter_count
    else "FAIL"
)


# ----------------------------------------------------------------------
# 5. RATE VALIDATION
# ----------------------------------------------------------------------

rate_validation = pd.DataFrame({
    "Metric": [
        "Minimum Patient Responsibility Rate",
        "Maximum Patient Responsibility Rate",
        "Minimum Payer Coverage Rate",
        "Maximum Payer Coverage Rate"
    ],

    "Value": [
        encounter_type_cost_summary[
            "Patient_Responsibility_Rate"
        ].min(),

        encounter_type_cost_summary[
            "Patient_Responsibility_Rate"
        ].max(),

        encounter_type_cost_summary[
            "Payer_Coverage_Rate"
        ].min(),

        encounter_type_cost_summary[
            "Payer_Coverage_Rate"
        ].max()
    ]
})


print("\nRATE VALIDATION")
print("-" * 70)

display(rate_validation)


# ----------------------------------------------------------------------
# 6. CHECK THAT RATES ARE WITHIN VALID RANGE
# ----------------------------------------------------------------------

rates_pass = (
    rate_validation["Value"]
    .between(0, 100)
    .all()
)


# ----------------------------------------------------------------------
# 7. OVERALL VALIDATION STATUS
# ----------------------------------------------------------------------

financial_reconciliation_pass = (
    encounter_type_cost_validation["Status"]
    .eq("PASS")
    .all()
)

encounter_count_pass = (
    source_encounter_count == grouped_encounter_count
)


overall_status = all([
    financial_reconciliation_pass,
    encounter_count_pass,
    rates_pass
])


# ----------------------------------------------------------------------
# 8. FINAL SUMMARY
# ----------------------------------------------------------------------

print("\n")
print("=" * 70)
print("STEP 4.16 VALIDATION SUMMARY")
print("=" * 70)

print(
    "Financial reconciliation:",
    "PASS" if financial_reconciliation_pass else "FAIL"
)

print(
    "Encounter count reconciliation:",
    "PASS" if encounter_count_pass else "FAIL"
)

print(
    "Rate validation:",
    "PASS" if rates_pass else "FAIL"
)

print("-" * 70)

print(
    "OVERALL STATUS:",
    "PASS" if overall_status else "FAIL"
)

print("=" * 70)
```

```python
# ======================================================================
# STEP 4.17A — PAYER FINANCIAL DISTRIBUTION
# ======================================================================

print("=" * 70)
print("STEP 4.17A — PAYER FINANCIAL DISTRIBUTION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Confirm payer column
# ----------------------------------------------------------------------

print("\nPAYER COLUMN CHECK")
print("-" * 70)

print(
    [
        col for col in encounters_clean.columns
        if "PAYER" in col.upper()
    ]
)


# ----------------------------------------------------------------------
# 2. Inspect payer distribution
# ----------------------------------------------------------------------

payer_distribution = (
    encounters_clean
    .groupby("PAYER", dropna=False)
    .size()
    .reset_index(name="Encounter_Count")
    .sort_values(
        "Encounter_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\nPAYER DISTRIBUTION")
print("-" * 70)

display(payer_distribution)


# ----------------------------------------------------------------------
# 3. Number of unique payers
# ----------------------------------------------------------------------

print("\nUNIQUE PAYERS")
print("-" * 70)

print(
    "Unique payers:",
    encounters_clean["PAYER"].nunique(dropna=False)
)
```

```python
# ======================================================================
# STEP 4.17B — PAYER × PATIENT RESPONSIBILITY
# ======================================================================

print("=" * 70)
print("STEP 4.17B — PAYER × PATIENT RESPONSIBILITY")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Aggregate financial burden by payer
# ----------------------------------------------------------------------

payer_patient_responsibility = (
    encounters_clean
    .groupby("PAYER", dropna=False)
    .agg(
        Encounter_Count=("PAYER", "size"),

        Total_Base_Encounter_Cost=(
            "BASE_ENCOUNTER_COST",
            "sum"
        ),

        Total_Claim_Cost=(
            "TOTAL_CLAIM_COST",
            "sum"
        ),

        Total_Payer_Coverage=(
            "PAYER_COVERAGE",
            "sum"
        ),

        Total_Patient_Responsibility=(
            "PATIENT_RESPONSIBILITY",
            "sum"
        )
    )
    .reset_index()
)


# ----------------------------------------------------------------------
# 2. Calculate average financial burden per encounter
# ----------------------------------------------------------------------

payer_patient_responsibility[
    "Average_Claim_Cost_Per_Encounter"
] = (
    payer_patient_responsibility["Total_Claim_Cost"]
    / payer_patient_responsibility["Encounter_Count"]
)


payer_patient_responsibility[
    "Average_Payer_Coverage_Per_Encounter"
] = (
    payer_patient_responsibility["Total_Payer_Coverage"]
    / payer_patient_responsibility["Encounter_Count"]
)


payer_patient_responsibility[
    "Average_Patient_Responsibility_Per_Encounter"
] = (
    payer_patient_responsibility[
        "Total_Patient_Responsibility"
    ]
    / payer_patient_responsibility["Encounter_Count"]
)


# ----------------------------------------------------------------------
# 3. Calculate financial responsibility rates
# ----------------------------------------------------------------------

payer_patient_responsibility[
    "Patient_Responsibility_Rate"
] = (
    payer_patient_responsibility[
        "Total_Patient_Responsibility"
    ]
    / payer_patient_responsibility["Total_Claim_Cost"]
    * 100
)


payer_patient_responsibility[
    "Payer_Coverage_Rate"
] = (
    payer_patient_responsibility[
        "Total_Payer_Coverage"
    ]
    / payer_patient_responsibility["Total_Claim_Cost"]
    * 100
)


# ----------------------------------------------------------------------
# 4. Handle possible division-by-zero values
# ----------------------------------------------------------------------

rate_columns = [
    "Patient_Responsibility_Rate",
    "Payer_Coverage_Rate"
]

average_columns = [
    "Average_Claim_Cost_Per_Encounter",
    "Average_Payer_Coverage_Per_Encounter",
    "Average_Patient_Responsibility_Per_Encounter"
]

payer_patient_responsibility[
    rate_columns + average_columns
] = (
    payer_patient_responsibility[
        rate_columns + average_columns
    ]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)


# ----------------------------------------------------------------------
# 5. Sort by total patient responsibility
# ----------------------------------------------------------------------

payer_patient_responsibility = (
    payer_patient_responsibility
    .sort_values(
        "Total_Patient_Responsibility",
        ascending=False
    )
    .reset_index(drop=True)
)


# ----------------------------------------------------------------------
# 6. Display complete payer analysis
# ----------------------------------------------------------------------

print("\nPAYER × PATIENT RESPONSIBILITY SUMMARY")
print("-" * 70)

display(payer_patient_responsibility)


# ----------------------------------------------------------------------
# 7. Top payers by patient financial burden
# ----------------------------------------------------------------------

print("\nTOP 10 PAYERS BY TOTAL PATIENT RESPONSIBILITY")
print("-" * 70)

display(
    payer_patient_responsibility[
        [
            "PAYER",
            "Encounter_Count",
            "Total_Claim_Cost",
            "Total_Payer_Coverage",
            "Total_Patient_Responsibility",
            "Average_Patient_Responsibility_Per_Encounter",
            "Patient_Responsibility_Rate",
            "Payer_Coverage_Rate"
        ]
    ]
    .head(10)
)
```

```python
# ======================================================================
# STEP 4.17C — PAYER × PATIENT RESPONSIBILITY VALIDATION
# ======================================================================

print("=" * 70)
print("STEP 4.17C — PAYER × PATIENT RESPONSIBILITY VALIDATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Source-level totals
# ----------------------------------------------------------------------

source_totals = {
    "Encounter_Count": len(encounters_clean),

    "Total_Base_Encounter_Cost": (
        encounters_clean["BASE_ENCOUNTER_COST"].sum()
    ),

    "Total_Claim_Cost": (
        encounters_clean["TOTAL_CLAIM_COST"].sum()
    ),

    "Total_Payer_Coverage": (
        encounters_clean["PAYER_COVERAGE"].sum()
    ),

    "Total_Patient_Responsibility": (
        encounters_clean["PATIENT_RESPONSIBILITY"].sum()
    )
}


# ----------------------------------------------------------------------
# 2. Payer-level grouped totals
# ----------------------------------------------------------------------

grouped_totals = {
    "Encounter_Count": (
        payer_patient_responsibility["Encounter_Count"].sum()
    ),

    "Total_Base_Encounter_Cost": (
        payer_patient_responsibility[
            "Total_Base_Encounter_Cost"
        ].sum()
    ),

    "Total_Claim_Cost": (
        payer_patient_responsibility[
            "Total_Claim_Cost"
        ].sum()
    ),

    "Total_Payer_Coverage": (
        payer_patient_responsibility[
            "Total_Payer_Coverage"
        ].sum()
    ),

    "Total_Patient_Responsibility": (
        payer_patient_responsibility[
            "Total_Patient_Responsibility"
        ].sum()
    )
}


# ----------------------------------------------------------------------
# 3. Financial reconciliation
# ----------------------------------------------------------------------

validation_rows = []

for metric in source_totals:

    source_value = source_totals[metric]
    grouped_value = grouped_totals[metric]

    difference = grouped_value - source_value

    validation_rows.append(
        {
            "Metric": metric,
            "Grouped_Total": grouped_value,
            "Source_Total": source_value,
            "Difference": difference,
            "Status": (
                "PASS"
                if abs(difference) < 0.01
                else "FAIL"
            )
        }
    )


payer_financial_validation = pd.DataFrame(
    validation_rows
)


print("\nFINANCIAL RECONCILIATION")
print("-" * 70)

display(payer_financial_validation)


# ----------------------------------------------------------------------
# 4. Validate payer count
# ----------------------------------------------------------------------

source_payer_count = (
    encounters_clean["PAYER"]
    .nunique(dropna=False)
)

grouped_payer_count = len(
    payer_patient_responsibility
)


print("\nPAYER COUNT VALIDATION")
print("-" * 70)

print(
    "Source unique payers:",
    source_payer_count
)

print(
    "Grouped payers:",
    grouped_payer_count
)

print(
    "STATUS:",
    "PASS"
    if source_payer_count == grouped_payer_count
    else "FAIL"
)


# ----------------------------------------------------------------------
# 5. Validate encounter counts
# ----------------------------------------------------------------------

source_encounter_count = len(
    encounters_clean
)

grouped_encounter_count = (
    payer_patient_responsibility[
        "Encounter_Count"
    ].sum()
)


print("\nENCOUNTER COUNT RECONCILIATION")
print("-" * 70)

print(
    "Source encounters:",
    source_encounter_count
)

print(
    "Grouped encounters:",
    grouped_encounter_count
)

print(
    "STATUS:",
    "PASS"
    if source_encounter_count == grouped_encounter_count
    else "FAIL"
)


# ----------------------------------------------------------------------
# 6. Validate financial rates
# ----------------------------------------------------------------------

rate_validation = pd.DataFrame(
    {
        "Metric": [
            "Minimum Patient Responsibility Rate",
            "Maximum Patient Responsibility Rate",
            "Minimum Payer Coverage Rate",
            "Maximum Payer Coverage Rate"
        ],

        "Value": [
            payer_patient_responsibility[
                "Patient_Responsibility_Rate"
            ].min(),

            payer_patient_responsibility[
                "Patient_Responsibility_Rate"
            ].max(),

            payer_patient_responsibility[
                "Payer_Coverage_Rate"
            ].min(),

            payer_patient_responsibility[
                "Payer_Coverage_Rate"
            ].max()
        ]
    }
)


print("\nRATE VALIDATION")
print("-" * 70)

display(rate_validation)


# ----------------------------------------------------------------------
# 7. Validate claim-cost composition
#
# Total Claim Cost should equal:
# Payer Coverage + Patient Responsibility
# ----------------------------------------------------------------------

claim_composition = (
    payer_patient_responsibility[
        "Total_Payer_Coverage"
    ]
    +
    payer_patient_responsibility[
        "Total_Patient_Responsibility"
    ]
)

claim_total = (
    payer_patient_responsibility[
        "Total_Claim_Cost"
    ]
)

claim_difference = (
    claim_composition - claim_total
).abs().max()


print("\nCLAIM COST COMPOSITION VALIDATION")
print("-" * 70)

print(
    "Maximum absolute difference:",
    claim_difference
)

print(
    "STATUS:",
    "PASS"
    if claim_difference < 0.01
    else "FAIL"
)


# ----------------------------------------------------------------------
# 8. Overall validation status
# ----------------------------------------------------------------------

financial_reconciliation_pass = (
    payer_financial_validation["Status"]
    .eq("PASS")
    .all()
)

payer_count_pass = (
    source_payer_count == grouped_payer_count
)

encounter_count_pass = (
    source_encounter_count == grouped_encounter_count
)

rates_pass = (
    rate_validation["Value"]
    .between(0, 100)
    .all()
)

claim_composition_pass = (
    claim_difference < 0.01
)


overall_status = all(
    [
        financial_reconciliation_pass,
        payer_count_pass,
        encounter_count_pass,
        rates_pass,
        claim_composition_pass
    ]
)


# ----------------------------------------------------------------------
# 9. Final summary
# ----------------------------------------------------------------------

print("\n")
print("=" * 70)
print("STEP 4.17 VALIDATION SUMMARY")
print("=" * 70)

print(
    "Financial reconciliation:",
    "PASS"
    if financial_reconciliation_pass
    else "FAIL"
)

print(
    "Payer count reconciliation:",
    "PASS"
    if payer_count_pass
    else "FAIL"
)

print(
    "Encounter count reconciliation:",
    "PASS"
    if encounter_count_pass
    else "FAIL"
)

print(
    "Rate validation:",
    "PASS"
    if rates_pass
    else "FAIL"
)

print(
    "Claim cost composition:",
    "PASS"
    if claim_composition_pass
    else "FAIL"
)

print("-" * 70)

print(
    "OVERALL STATUS:",
    "PASS"
    if overall_status
    else "FAIL"
)
```

```python
# ======================================================================
# STEP 4.18A — PROCEDURE DISTRIBUTION
# ======================================================================

print("=" * 70)
print("STEP 4.18A — PROCEDURE DISTRIBUTION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Confirm required procedure columns
# ----------------------------------------------------------------------

required_procedure_columns = [
    "PATIENT",
    "ENCOUNTER",
    "CODE",
    "DESCRIPTION",
    "BASE_COST"
]

missing_columns = [
    col
    for col in required_procedure_columns
    if col not in procedures_clean.columns
]


print("\nREQUIRED COLUMN CHECK")
print("-" * 70)

if missing_columns:
    print("Missing columns:", missing_columns)
else:
    print("STATUS: PASS")
    print("All required procedure columns are available.")


# ----------------------------------------------------------------------
# 2. Basic procedure statistics
# ----------------------------------------------------------------------

print("\nPROCEDURE DATASET SUMMARY")
print("-" * 70)

print(
    "Total procedures:",
    len(procedures_clean)
)

print(
    "Unique procedure codes:",
    procedures_clean["CODE"].nunique()
)

print(
    "Unique procedure descriptions:",
    procedures_clean["DESCRIPTION"].nunique()
)

print(
    "Unique patients with procedures:",
    procedures_clean["PATIENT"].nunique()
)

print(
    "Unique encounters with procedures:",
    procedures_clean["ENCOUNTER"].nunique()
)


# ----------------------------------------------------------------------
# 3. Procedure frequency by code
# ----------------------------------------------------------------------

procedure_code_distribution = (
    procedures_clean
    .groupby(
        ["CODE", "DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("CODE", "size"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Average_Procedure_Cost=("BASE_COST", "mean")
    )
    .reset_index()
)


# ----------------------------------------------------------------------
# 4. Sort by procedure frequency
# ----------------------------------------------------------------------

procedure_code_distribution = (
    procedure_code_distribution
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ----------------------------------------------------------------------
# 5. Display top procedures
# ----------------------------------------------------------------------

print("\nTOP 20 PROCEDURES BY FREQUENCY")
print("-" * 70)

display(
    procedure_code_distribution.head(20)
)
```

```python
# ======================================================================
# STEP 4.18B — PROCEDURE × COST / UTILIZATION
# ======================================================================

print("=" * 70)
print("STEP 4.18B — PROCEDURE × COST / UTILIZATION")
print("=" * 70)


# ----------------------------------------------------------------------
# 1. Procedure-level cost and utilization summary
# ----------------------------------------------------------------------

procedure_cost_summary = (
    procedures_clean
    .groupby(["CODE", "DESCRIPTION"], dropna=False)
    .agg(
        Procedure_Count=("CODE", "size"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Average_Procedure_Cost=("BASE_COST", "mean"),
        Median_Procedure_Cost=("BASE_COST", "median"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique")
    )
    .reset_index()
)


# ----------------------------------------------------------------------
# 2. Calculate utilization and cost metrics
# ----------------------------------------------------------------------

procedure_cost_summary["Cost_Per_Patient"] = (
    procedure_cost_summary["Total_Procedure_Cost"]
    / procedure_cost_summary["Unique_Patients"]
    .replace(0, np.nan)
)

procedure_cost_summary["Procedures_Per_Patient"] = (
    procedure_cost_summary["Procedure_Count"]
    / procedure_cost_summary["Unique_Patients"]
    .replace(0, np.nan)
)

procedure_cost_summary["Procedures_Per_Encounter"] = (
    procedure_cost_summary["Procedure_Count"]
    / procedure_cost_summary["Unique_Encounters"]
    .replace(0, np.nan)
)


# ----------------------------------------------------------------------
# 3. Top procedures by total cost
# ----------------------------------------------------------------------

top_procedures_by_cost = (
    procedure_cost_summary
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


print("\nTOP 20 PROCEDURES BY TOTAL COST")
print("-" * 70)

display(top_procedures_by_cost)


# ----------------------------------------------------------------------
# 4. Top procedures by frequency
# ----------------------------------------------------------------------

top_procedures_by_frequency = (
    procedure_cost_summary
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


print("\nTOP 20 PROCEDURES BY FREQUENCY")
print("-" * 70)

display(top_procedures_by_frequency)


# ----------------------------------------------------------------------
# 5. Most expensive procedures by average cost
# ----------------------------------------------------------------------

top_procedures_by_average_cost = (
    procedure_cost_summary
    .sort_values(
        "Average_Procedure_Cost",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


print("\nTOP 20 PROCEDURES BY AVERAGE COST")
print("-" * 70)

display(top_procedures_by_average_cost)


# ----------------------------------------------------------------------
# 6. High-utilization procedures
# ----------------------------------------------------------------------

high_utilization_procedures = (
    procedure_cost_summary
    .sort_values(
        [
            "Procedure_Count",
            "Total_Procedure_Cost"
        ],
        ascending=[False, False]
    )
    .head(20)
    .reset_index(drop=True)
)


print("\nTOP 20 HIGH-UTILIZATION PROCEDURES")
print("-" * 70)

display(high_utilization_procedures)


# ----------------------------------------------------------------------
# 7. Procedure cost concentration
# ----------------------------------------------------------------------

total_procedure_cost = (
    procedures_clean["BASE_COST"].sum()
)

top_10_procedure_cost = (
    procedure_cost_summary
    .nlargest(
        10,
        "Total_Procedure_Cost"
    )["Total_Procedure_Cost"]
    .sum()
)

top_20_procedure_cost = (
    procedure_cost_summary
    .nlargest(
        20,
        "Total_Procedure_Cost"
    )["Total_Procedure_Cost"]
    .sum()
)

top_10_cost_share = (
    top_10_procedure_cost / total_procedure_cost * 100
)

top_20_cost_share = (
    top_20_procedure_cost / total_procedure_cost * 100
)


print("\nPROCEDURE COST CONCENTRATION")
print("-" * 70)

print(
    f"Total procedure cost: "
    f"{total_procedure_cost:,.2f}"
)

print(
    f"Top 10 procedures total cost: "
    f"{top_10_procedure_cost:,.2f}"
)

print(
    f"Top 10 procedure cost share: "
    f"{top_10_cost_share:.2f}%"
)

print(
    f"Top 20 procedures total cost: "
    f"{top_20_procedure_cost:,.2f}"
)

print(
    f"Top 20 procedure cost share: "
    f"{top_20_cost_share:.2f}%"
)


# ----------------------------------------------------------------------
# 8. Procedure utilization concentration
# ----------------------------------------------------------------------

total_procedure_count = len(procedures_clean)

top_10_procedure_count = (
    procedure_cost_summary
    .nlargest(
        10,
        "Procedure_Count"
    )["Procedure_Count"]
    .sum()
)

top_20_procedure_count = (
    procedure_cost_summary
    .nlargest(
        20,
        "Procedure_Count"
    )["Procedure_Count"]
    .sum()
)

top_10_utilization_share = (
    top_10_procedure_count
    / total_procedure_count
    * 100
)

top_20_utilization_share = (
    top_20_procedure_count
    / total_procedure_count
    * 100
)


print("\nPROCEDURE UTILIZATION CONCENTRATION")
print("-" * 70)

print(
    f"Total procedures: "
    f"{total_procedure_count:,}"
)

print(
    f"Top 10 procedure volume: "
    f"{top_10_procedure_count:,}"
)

print(
    f"Top 10 procedure utilization share: "
    f"{top_10_utilization_share:.2f}%"
)

print(
    f"Top 20 procedure volume: "
    f"{top_20_procedure_count:,}"
)

print(
    f"Top 20 procedure utilization share: "
    f"{top_20_utilization_share:.2f}%"
)


# ----------------------------------------------------------------------
# 9. Overall procedure × cost metrics
# ----------------------------------------------------------------------

procedure_cost_metrics = pd.DataFrame(
    {
        "Metric": [
            "Total Procedures",
            "Unique Procedure Codes",
            "Unique Procedure Descriptions",
            "Unique Patients",
            "Unique Encounters",
            "Total Procedure Cost",
            "Average Procedure Cost",
            "Median Procedure Cost",
            "Top 10 Cost Share (%)",
            "Top 20 Cost Share (%)",
            "Top 10 Utilization Share (%)",
            "Top 20 Utilization Share (%)"
        ],
        "Value": [
            len(procedures_clean),
            procedures_clean["CODE"].nunique(),
            procedures_clean["DESCRIPTION"].nunique(),
            procedures_clean["PATIENT"].nunique(),
            procedures_clean["ENCOUNTER"].nunique(),
            procedures_clean["BASE_COST"].sum(),
            procedures_clean["BASE_COST"].mean(),
            procedures_clean["BASE_COST"].median(),
            top_10_cost_share,
            top_20_cost_share,
            top_10_utilization_share,
            top_20_utilization_share
        ]
    }
)


print("\nOVERALL PROCEDURE × COST / UTILIZATION METRICS")
print("-" * 70)

display(procedure_cost_metrics)
```

```python
# ----------------------------------------------------------------------------
# 1. SELECT CORRECT STEP 4.18A OUTPUT + REQUIRED COLUMN CHECK
# ----------------------------------------------------------------------------

# Step 4.18A generated the complete procedure-level summary as
# `procedure_cost_summary`.
#
# Do NOT use the older/basic `procedure_summary`, which only contains:
# ['Procedure', 'Procedure_Count']
#
# `procedure_cost_summary` contains the complete 11-column dataset required
# for Step 4.18B.

if "procedure_cost_summary" not in globals():
    raise NameError(
        "procedure_cost_summary does not exist. "
        "Please run STEP 4.18A first."
    )

# Use the detailed Step 4.18A output
procedure_summary = procedure_cost_summary.copy()


required_columns = [
    "CODE",
    "DESCRIPTION",
    "Procedure_Count",
    "Total_Procedure_Cost",
    "Average_Procedure_Cost",
    "Median_Procedure_Cost",
    "Unique_Patients",
    "Unique_Encounters",
    "Cost_Per_Patient",
    "Procedures_Per_Patient",
    "Procedures_Per_Encounter"
]


missing_columns = [
    col for col in required_columns
    if col not in procedure_summary.columns
]


if missing_columns:
    print("\nSTATUS: FAIL")
    print("Missing columns:", missing_columns)

    print("\nCurrent procedure_summary columns:")
    print(procedure_summary.columns.tolist())

    raise ValueError(
        "procedure_cost_summary does not contain all required "
        "Step 4.18B columns."
    )


print("\nSOURCE DATAFRAME")
print("-" * 70)
print("Using: procedure_cost_summary")
print(f"Shape: {procedure_summary.shape}")
print("Columns:")
print(procedure_summary.columns.tolist())


print("\nREQUIRED COLUMN CHECK")
print("-" * 70)
print("STATUS: PASS")
print("All required procedure columns are available.")
```

```python
# ============================================================================
# STEP 4.18B — PROCEDURE COST & UTILIZATION ANALYSIS
# ----------------------------------------------------------------------------
# Purpose:
#   Identify procedures that are:
#   1. High-cost
#   2. High-utilization
#   3. High-cost per patient
#   4. Highly repetitive per patient
#   5. Potential cost/utilization outliers
#
# IMPORTANT:
#   Step 4.18A creates:
#       procedure_cost_summary
#
#   procedure_summary is a different dataframe containing only:
#       ['Procedure', 'Procedure_Count']
#
#   Therefore, this step uses procedure_cost_summary.
# ============================================================================

print("=" * 70)
print("STEP 4.18B — PROCEDURE COST & UTILIZATION ANALYSIS")
print("=" * 70)


# ============================================================================
# 1. USE THE CORRECT STEP 4.18A OUTPUT
# ============================================================================

# Step 4.18A generated the complete procedure-level summary as:
# procedure_cost_summary

procedure_analysis = procedure_cost_summary.copy()


# ============================================================================
# 2. REQUIRED COLUMN CHECK
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "Procedure_Count",
    "Total_Procedure_Cost",
    "Average_Procedure_Cost",
    "Median_Procedure_Cost",
    "Unique_Patients",
    "Unique_Encounters",
    "Cost_Per_Patient",
    "Procedures_Per_Patient",
    "Procedures_Per_Encounter"
]

missing_columns = [
    col for col in required_columns
    if col not in procedure_analysis.columns
]


if missing_columns:

    print("\nSTATUS: FAIL")
    print("Missing columns:", missing_columns)

    print("\nCurrent procedure_analysis columns:")
    print(procedure_analysis.columns.tolist())

    raise ValueError(
        "procedure_cost_summary does not contain all required "
        "Step 4.18A columns."
    )


print("\nREQUIRED COLUMN CHECK")
print("-" * 70)
print("STATUS: PASS")
print("All required procedure columns are available.")

print(f"Procedure records available: {len(procedure_analysis):,}")


# ============================================================================
# 3. DATA TYPE / NUMERIC VALIDATION
# ============================================================================

numeric_columns = [
    "Procedure_Count",
    "Total_Procedure_Cost",
    "Average_Procedure_Cost",
    "Median_Procedure_Cost",
    "Unique_Patients",
    "Unique_Encounters",
    "Cost_Per_Patient",
    "Procedures_Per_Patient",
    "Procedures_Per_Encounter"
]

for col in numeric_columns:

    procedure_analysis[col] = pd.to_numeric(
        procedure_analysis[col],
        errors="coerce"
    )


# Remove records where the key cost/utilization measures cannot be evaluated
procedure_analysis = procedure_analysis.dropna(
    subset=[
        "Procedure_Count",
        "Total_Procedure_Cost",
        "Average_Procedure_Cost",
        "Unique_Patients",
        "Cost_Per_Patient",
        "Procedures_Per_Patient"
    ]
).copy()


print("\nDATA VALIDATION")
print("-" * 70)
print(f"Valid procedure records: {len(procedure_analysis):,}")
print(
    f"Missing values in key metrics: "
    f"{procedure_analysis[[
        'Procedure_Count',
        'Total_Procedure_Cost',
        'Average_Procedure_Cost',
        'Cost_Per_Patient',
        'Procedures_Per_Patient'
    ]].isna().sum().sum():,}"
)


# ============================================================================
# 4. HIGH-COST PROCEDURES
# ----------------------------------------------------------------------------
# Ranked by total procedure cost.
# ============================================================================

high_cost_procedures = (
    procedure_analysis
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .head(10)
    .copy()
)


print("\nTOP 10 PROCEDURES BY TOTAL COST")
print("-" * 70)

print(
    high_cost_procedures[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Total_Procedure_Cost",
            "Average_Procedure_Cost",
            "Unique_Patients",
            "Cost_Per_Patient"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 5. HIGH-UTILIZATION PROCEDURES
# ----------------------------------------------------------------------------
# Ranked by number of procedure records.
# ============================================================================

high_utilization_procedures = (
    procedure_analysis
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .head(10)
    .copy()
)


print("\nTOP 10 PROCEDURES BY UTILIZATION")
print("-" * 70)

print(
    high_utilization_procedures[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Total_Procedure_Cost",
            "Average_Procedure_Cost",
            "Unique_Patients",
            "Procedures_Per_Patient"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 6. HIGH UNIT-COST PROCEDURES
# ----------------------------------------------------------------------------
# Ranked by average cost per procedure.
#
# NOTE:
#   This is different from total cost.
#   A procedure can have a very high unit cost but low utilization.
# ============================================================================

high_unit_cost_procedures = (
    procedure_analysis
    .sort_values(
        "Average_Procedure_Cost",
        ascending=False
    )
    .head(10)
    .copy()
)


print("\nTOP 10 PROCEDURES BY AVERAGE COST")
print("-" * 70)

print(
    high_unit_cost_procedures[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Average_Procedure_Cost",
            "Median_Procedure_Cost",
            "Total_Procedure_Cost",
            "Unique_Patients"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 7. HIGH COST PER PATIENT
# ----------------------------------------------------------------------------
# Identifies procedures generating the highest cumulative cost per patient.
# ============================================================================

high_cost_per_patient = (
    procedure_analysis
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .head(10)
    .copy()
)


print("\nTOP 10 PROCEDURES BY COST PER PATIENT")
print("-" * 70)

print(
    high_cost_per_patient[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_Per_Patient",
            "Procedures_Per_Patient"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 8. HIGHLY REPEATED PROCEDURES
# ----------------------------------------------------------------------------
# Ranked by average number of procedures per patient.
# ============================================================================

high_repeat_procedures = (
    procedure_analysis
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .head(10)
    .copy()
)


print("\nTOP 10 PROCEDURES BY PROCEDURES PER PATIENT")
print("-" * 70)

print(
    high_repeat_procedures[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Procedures_Per_Patient",
            "Procedures_Per_Encounter",
            "Total_Procedure_Cost"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 9. COST × UTILIZATION CLASSIFICATION
# ----------------------------------------------------------------------------
# Uses the median across procedure codes as the threshold.
#
# Four profiles:
#   High Cost / High Utilization
#   High Cost / Low Utilization
#   Low Cost / High Utilization
#   Low Cost / Low Utilization
# ============================================================================

median_cost = procedure_analysis[
    "Total_Procedure_Cost"
].median()

median_volume = procedure_analysis[
    "Procedure_Count"
].median()


procedure_analysis["Cost_Profile"] = np.where(
    procedure_analysis["Total_Procedure_Cost"] >= median_cost,
    "High Cost",
    "Low Cost"
)


procedure_analysis["Utilization_Profile"] = np.where(
    procedure_analysis["Procedure_Count"] >= median_volume,
    "High Utilization",
    "Low Utilization"
)


procedure_analysis["Procedure_Profile"] = (
    procedure_analysis["Cost_Profile"]
    + " / "
    + procedure_analysis["Utilization_Profile"]
)


print("\nPROCEDURE COST × UTILIZATION PROFILE")
print("-" * 70)

print(f"Median procedure total cost: {median_cost:,.2f}")
print(f"Median procedure volume: {median_volume:,.2f}")

profile_counts = (
    procedure_analysis[
        "Procedure_Profile"
    ]
    .value_counts()
    .rename_axis("Profile")
    .reset_index(name="Procedure_Count")
)


print(
    profile_counts.to_string(index=False)
)


# ============================================================================
# 10. HIGH-COST + HIGH-UTILIZATION PROCEDURES
# ----------------------------------------------------------------------------

high_cost_high_utilization = (
    procedure_analysis[
        (procedure_analysis["Total_Procedure_Cost"] >= median_cost)
        &
        (procedure_analysis["Procedure_Count"] >= median_volume)
    ]
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .copy()
)


print("\nHIGH-COST + HIGH-UTILIZATION PROCEDURES")
print("-" * 70)

if len(high_cost_high_utilization) > 0:

    print(
        high_cost_high_utilization[
            [
                "CODE",
                "DESCRIPTION",
                "Procedure_Count",
                "Total_Procedure_Cost",
                "Average_Procedure_Cost",
                "Unique_Patients",
                "Cost_Per_Patient"
            ]
        ].to_string(index=False)
    )

else:

    print("No high-cost/high-utilization procedures identified.")


# ============================================================================
# 11. HIGH-COST + LOW-UTILIZATION PROCEDURES
# ----------------------------------------------------------------------------
# These are particularly important because they may represent:
#   - expensive procedures
#   - relatively low frequency
#   - potential high-impact cost drivers
# ============================================================================

high_cost_low_utilization = (
    procedure_analysis[
        (procedure_analysis["Total_Procedure_Cost"] >= median_cost)
        &
        (procedure_analysis["Procedure_Count"] < median_volume)
    ]
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .copy()
)


print("\nHIGH-COST + LOW-UTILIZATION PROCEDURES")
print("-" * 70)

if len(high_cost_low_utilization) > 0:

    print(
        high_cost_low_utilization[
            [
                "CODE",
                "DESCRIPTION",
                "Procedure_Count",
                "Total_Procedure_Cost",
                "Average_Procedure_Cost",
                "Unique_Patients",
                "Cost_Per_Patient"
            ]
        ]
        .head(10)
        .to_string(index=False)
    )

else:

    print("No high-cost/low-utilization procedures identified.")


# ============================================================================
# 12. LOW-COST + HIGH-UTILIZATION PROCEDURES
# ----------------------------------------------------------------------------
# These procedures may have low unit cost but create substantial aggregate
# workload/cost because of high frequency.
# ============================================================================

low_cost_high_utilization = (
    procedure_analysis[
        (procedure_analysis["Total_Procedure_Cost"] < median_cost)
        &
        (procedure_analysis["Procedure_Count"] >= median_volume)
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .copy()
)


print("\nLOW-COST + HIGH-UTILIZATION PROCEDURES")
print("-" * 70)

if len(low_cost_high_utilization) > 0:

    print(
        low_cost_high_utilization[
            [
                "CODE",
                "DESCRIPTION",
                "Procedure_Count",
                "Total_Procedure_Cost",
                "Average_Procedure_Cost",
                "Unique_Patients",
                "Procedures_Per_Patient"
            ]
        ]
        .head(10)
        .to_string(index=False)
    )

else:

    print("No low-cost/high-utilization procedures identified.")


# ============================================================================
# 13. LOW-COST + LOW-UTILIZATION PROCEDURES
# ============================================================================

low_cost_low_utilization = (
    procedure_analysis[
        (procedure_analysis["Total_Procedure_Cost"] < median_cost)
        &
        (procedure_analysis["Procedure_Count"] < median_volume)
    ]
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .copy()
)


print("\nLOW-COST + LOW-UTILIZATION PROCEDURES")
print("-" * 70)

print(
    low_cost_low_utilization[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Total_Procedure_Cost",
            "Average_Procedure_Cost"
        ]
    ]
    .head(10)
    .to_string(index=False)
)


# ============================================================================
# 14. PROCEDURE COST CONCENTRATION
# ----------------------------------------------------------------------------
# Measures how concentrated total procedure expenditure is among the
# highest-cost procedure codes.
# ============================================================================

total_cost = procedure_analysis[
    "Total_Procedure_Cost"
].sum()


top_10_cost = (
    procedure_analysis
    .nlargest(
        min(10, len(procedure_analysis)),
        "Total_Procedure_Cost"
    )[
        "Total_Procedure_Cost"
    ]
    .sum()
)


top_20_cost = (
    procedure_analysis
    .nlargest(
        min(20, len(procedure_analysis)),
        "Total_Procedure_Cost"
    )[
        "Total_Procedure_Cost"
    ]
    .sum()
)


top_10_cost_share = (
    (top_10_cost / total_cost) * 100
    if total_cost > 0
    else 0
)


top_20_cost_share = (
    (top_20_cost / total_cost) * 100
    if total_cost > 0
    else 0
)


print("\nPROCEDURE COST CONCENTRATION")
print("-" * 70)

print(
    f"Total procedure cost: "
    f"{total_cost:,.2f}"
)

print(
    f"Top 10 procedure cost: "
    f"{top_10_cost:,.2f}"
)

print(
    f"Top 10 procedure cost share: "
    f"{top_10_cost_share:.2f}%"
)

print(
    f"Top 20 procedure cost: "
    f"{top_20_cost:,.2f}"
)

print(
    f"Top 20 procedure cost share: "
    f"{top_20_cost_share:.2f}%"
)


# ============================================================================
# 15. PROCEDURE UTILIZATION CONCENTRATION
# ============================================================================

total_procedures = procedure_analysis[
    "Procedure_Count"
].sum()


top_10_volume = (
    procedure_analysis
    .nlargest(
        min(10, len(procedure_analysis)),
        "Procedure_Count"
    )[
        "Procedure_Count"
    ]
    .sum()
)


top_20_volume = (
    procedure_analysis
    .nlargest(
        min(20, len(procedure_analysis)),
        "Procedure_Count"
    )[
        "Procedure_Count"
    ]
    .sum()
)


top_10_utilization_share = (
    (top_10_volume / total_procedures) * 100
    if total_procedures > 0
    else 0
)


top_20_utilization_share = (
    (top_20_volume / total_procedures) * 100
    if total_procedures > 0
    else 0
)


print("\nPROCEDURE UTILIZATION CONCENTRATION")
print("-" * 70)

print(
    f"Total procedures: "
    f"{total_procedures:,.0f}"
)

print(
    f"Top 10 procedure volume: "
    f"{top_10_volume:,.0f}"
)

print(
    f"Top 10 procedure utilization share: "
    f"{top_10_utilization_share:.2f}%"
)

print(
    f"Top 20 procedure volume: "
    f"{top_20_volume:,.0f}"
)

print(
    f"Top 20 procedure utilization share: "
    f"{top_20_utilization_share:.2f}%"
)


# ============================================================================
# 16. OVERALL PROCEDURE × COST / UTILIZATION METRICS
# ============================================================================

overall_metrics = pd.DataFrame({
    "Metric": [
        "Total Procedures",
        "Unique Procedure Codes",
        "Unique Procedure Descriptions",
        "Unique Patients",
        "Unique Encounters",
        "Total Procedure Cost",
        "Average Procedure Cost",
        "Median Procedure Cost",
        "Top 10 Cost Share (%)",
        "Top 20 Cost Share (%)",
        "Top 10 Utilization Share (%)",
        "Top 20 Utilization Share (%)"
    ],

    "Value": [
        procedure_analysis["Procedure_Count"].sum(),

        procedure_analysis["CODE"].nunique(),

        procedure_analysis["DESCRIPTION"].nunique(),

        procedure_analysis["Unique_Patients"].max(),

        procedure_analysis["Unique_Encounters"].sum(),

        total_cost,

        (
            total_cost / total_procedures
            if total_procedures > 0
            else 0
        ),

        procedure_analysis["Average_Procedure_Cost"].median(),

        top_10_cost_share,

        top_20_cost_share,

        top_10_utilization_share,

        top_20_utilization_share
    ]
})


print("\nOVERALL PROCEDURE × COST / UTILIZATION METRICS")
print("-" * 70)

print(
    overall_metrics.to_string(index=False)
)


# ============================================================================
# 17. SAVE KEY OUTPUT DATAFRAMES
# ----------------------------------------------------------------------------
# These variables can be used by later analysis/dashboard steps.
# ============================================================================

# Complete procedure-level analysis
procedure_cost_utilization_analysis = procedure_analysis.copy()


# Top cost procedures
top_procedures_by_cost = (
    procedure_analysis
    .nlargest(
        min(20, len(procedure_analysis)),
        "Total_Procedure_Cost"
    )
    .copy()
)


# Top utilization procedures
top_procedures_by_frequency = (
    procedure_analysis
    .nlargest(
        min(20, len(procedure_analysis)),
        "Procedure_Count"
    )
    .copy()
)


# Top average-cost procedures
top_procedures_by_average_cost = (
    procedure_analysis
    .nlargest(
        min(20, len(procedure_analysis)),
        "Average_Procedure_Cost"
    )
    .copy()
)


# ============================================================================
# 18. FINAL SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18B COMPLETE")
print("=" * 70)

print(
    f"Procedures analysed: "
    f"{len(procedure_analysis):,}"
)

print(
    f"Total procedure records: "
    f"{total_procedures:,.0f}"
)

print(
    f"Total procedure cost: "
    f"{total_cost:,.2f}"
)

print(
    f"High-cost procedures identified: "
    f"{len(high_cost_procedures):,}"
)

print(
    f"High-utilization procedures identified: "
    f"{len(high_utilization_procedures):,}"
)

print(
    f"High-cost/high-utilization procedures: "
    f"{len(high_cost_high_utilization):,}"
)

print(
    f"High-cost/low-utilization procedures: "
    f"{len(high_cost_low_utilization):,}"
)

print(
    f"Low-cost/high-utilization procedures: "
    f"{len(low_cost_high_utilization):,}"
)

print(
    f"Top 10 cost concentration: "
    f"{top_10_cost_share:.2f}%"
)

print(
    f"Top 20 cost concentration: "
    f"{top_20_cost_share:.2f}%"
)

print(
    f"Top 10 utilization concentration: "
    f"{top_10_utilization_share:.2f}%"
)

print(
    f"Top 20 utilization concentration: "
    f"{top_20_utilization_share:.2f}%"
)

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# PROCEDURE CODE ↔ DESCRIPTION CONSISTENCY CHECK
# ============================================================================

print("=" * 70)
print("PROCEDURE CODE ↔ DESCRIPTION CONSISTENCY CHECK")
print("=" * 70)

code_description_map = (
    procedures_clean
    .groupby("CODE")["DESCRIPTION"]
    .nunique()
    .sort_values(ascending=False)
)

multi_description_codes = (
    code_description_map[code_description_map > 1]
    .reset_index(name="Unique_Descriptions")
)

print("\nCODES WITH MULTIPLE DESCRIPTIONS")
print("-" * 70)

print(
    f"Codes with multiple descriptions: "
    f"{len(multi_description_codes):,}"
)

print(
    multi_description_codes
    .to_string(index=False)
)


# Show the actual mappings
if len(multi_description_codes) > 0:

    print("\nACTUAL CODE → DESCRIPTION MAPPINGS")
    print("-" * 70)

    for code in multi_description_codes["CODE"].head(20):

        descriptions = (
            procedures_clean.loc[
                procedures_clean["CODE"] == code,
                "DESCRIPTION"
            ]
            .dropna()
            .unique()
        )

        print(f"\nCODE: {code}")

        for description in descriptions:
            print(f"  - {description}")


# ============================================================================
# DESCRIPTION → CODE CHECK
# ============================================================================

description_code_map = (
    procedures_clean
    .groupby("DESCRIPTION")["CODE"]
    .nunique()
    .sort_values(ascending=False)
)

multi_code_descriptions = (
    description_code_map[description_code_map > 1]
    .reset_index(name="Unique_Codes")
)

print("\n\nDESCRIPTIONS ASSOCIATED WITH MULTIPLE CODES")
print("-" * 70)

print(
    f"Descriptions with multiple codes: "
    f"{len(multi_code_descriptions):,}"
)

print(
    multi_code_descriptions
    .head(20)
    .to_string(index=False)
)


# ============================================================================
# EXACT DUPLICATE ROW CHECK
# ============================================================================

exact_duplicate_count = procedures_clean.duplicated().sum()

print("\n\nEXACT DUPLICATE RECORD CHECK")
print("-" * 70)

print(
    f"Exact duplicate rows: "
    f"{exact_duplicate_count:,}"
)

if exact_duplicate_count == 0:
    print("STATUS: No exact duplicate procedure records found.")
else:
    print("STATUS: Exact duplicate records exist.")
```

```python
# ============================================================================
# STEP 4.18C — PROCEDURE COST DATA QUALITY & ANOMALY ANALYSIS
# ----------------------------------------------------------------------------
# Purpose:
#   1. Validate procedure cost data quality
#   2. Investigate repeated BASE_COST values
#   3. Quantify the impact of BASE_COST = 431
#   4. Investigate CODE ↔ DESCRIPTION inconsistencies
#   5. Identify extreme procedure cost values
#   6. Perform sensitivity analysis excluding BASE_COST = 431
#
# Important:
#   - No records are deleted.
#   - procedures_clean remains unchanged.
#   - BASE_COST = 431 is FLAGGED, not removed.
# ============================================================================

print("=" * 70)
print("STEP 4.18C — PROCEDURE COST DATA QUALITY & ANOMALY ANALYSIS")
print("=" * 70)


# ============================================================================
# 1. REQUIRED DATA CHECK
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "PATIENT",
    "ENCOUNTER",
    "BASE_COST"
]

missing_columns = [
    col for col in required_columns
    if col not in procedures_clean.columns
]

if missing_columns:

    print("\nSTATUS: FAIL")
    print("Missing columns:", missing_columns)

    raise ValueError(
        "Required procedure cost-quality columns are missing."
    )

print("\nREQUIRED COLUMN CHECK")
print("-" * 70)
print("STATUS: PASS")
print("All required columns are available.")


# ============================================================================
# 2. WORKING COPY
# ----------------------------------------------------------------------------
# Do NOT modify procedures_clean.
# ============================================================================

procedure_quality = procedures_clean.copy()


# ============================================================================
# 3. BASIC DATA QUALITY METRICS
# ============================================================================

total_records = len(procedure_quality)

missing_cost = procedure_quality["BASE_COST"].isna().sum()

zero_cost = (
    procedure_quality["BASE_COST"]
    .eq(0)
    .sum()
)

negative_cost = (
    procedure_quality["BASE_COST"]
    .lt(0)
    .sum()
)

positive_cost = (
    procedure_quality["BASE_COST"]
    .gt(0)
    .sum()
)

print("\nBASIC PROCEDURE COST DATA QUALITY")
print("-" * 70)

print(f"Total procedure records: {total_records:,}")
print(f"Missing BASE_COST: {missing_cost:,}")
print(f"Zero BASE_COST: {zero_cost:,}")
print(f"Negative BASE_COST: {negative_cost:,}")
print(f"Positive BASE_COST: {positive_cost:,}")


# ============================================================================
# 4. EXACT DUPLICATE CHECK
# ============================================================================

exact_duplicates = procedure_quality.duplicated().sum()

print("\nEXACT DUPLICATE CHECK")
print("-" * 70)

print(f"Exact duplicate records: {exact_duplicates:,}")

if exact_duplicates == 0:
    print("STATUS: PASS — No exact duplicate records detected.")
else:
    print("STATUS: REVIEW — Exact duplicate records detected.")


# ============================================================================
# 5. CODE ↔ DESCRIPTION CONSISTENCY
# ============================================================================

code_description_counts = (
    procedure_quality
    .groupby("CODE")["DESCRIPTION"]
    .nunique()
)

multi_description_codes = (
    code_description_counts[
        code_description_counts > 1
    ]
    .sort_values(ascending=False)
)

description_code_counts = (
    procedure_quality
    .groupby("DESCRIPTION")["CODE"]
    .nunique()
)

multi_code_descriptions = (
    description_code_counts[
        description_code_counts > 1
    ]
    .sort_values(ascending=False)
)

print("\nCODE ↔ DESCRIPTION CONSISTENCY")
print("-" * 70)

print(
    f"Codes with multiple descriptions: "
    f"{len(multi_description_codes):,}"
)

print(
    f"Descriptions with multiple codes: "
    f"{len(multi_code_descriptions):,}"
)

if len(multi_description_codes) > 0:

    print("\nCODES WITH MULTIPLE DESCRIPTIONS")
    print("-" * 70)

    print(
        multi_description_codes
        .rename("Unique_Descriptions")
        .reset_index()
        .to_string(index=False)
    )


# ============================================================================
# 6. CREATE BASE_COST = 431 FLAG
# ----------------------------------------------------------------------------
# This is a FLAG only. No records are removed.
# ============================================================================

procedure_quality["BASE_COST_431_FLAG"] = np.where(
    procedure_quality["BASE_COST"].eq(431),
    "BASE_COST_431",
    "OTHER_COST"
)

cost_431_records = (
    procedure_quality["BASE_COST"]
    .eq(431)
    .sum()
)

cost_431_share_records = (
    cost_431_records / total_records * 100
    if total_records > 0
    else 0
)

print("\nBASE_COST = 431 ANALYSIS")
print("-" * 70)

print(
    f"Records with BASE_COST = 431: "
    f"{cost_431_records:,}"
)

print(
    f"Share of all procedure records: "
    f"{cost_431_share_records:.2f}%"
)


# ============================================================================
# 7. COST IMPACT OF BASE_COST = 431
# ============================================================================

total_cost_all = (
    procedure_quality["BASE_COST"]
    .sum()
)

cost_431_total = (
    procedure_quality.loc[
        procedure_quality["BASE_COST"] == 431,
        "BASE_COST"
    ]
    .sum()
)

cost_non_431_total = (
    procedure_quality.loc[
        procedure_quality["BASE_COST"] != 431,
        "BASE_COST"
    ]
    .sum()
)

cost_431_share = (
    cost_431_total / total_cost_all * 100
    if total_cost_all != 0
    else 0
)

print("\nBASE_COST = 431 COST IMPACT")
print("-" * 70)

print(
    f"Total procedure cost: "
    f"{total_cost_all:,.2f}"
)

print(
    f"Cost from BASE_COST = 431 records: "
    f"{cost_431_total:,.2f}"
)

print(
    f"Cost from non-431 records: "
    f"{cost_non_431_total:,.2f}"
)

print(
    f"431 cost share: "
    f"{cost_431_share:.2f}%"
)


# ============================================================================
# 8. AVERAGE / MEDIAN COST COMPARISON
# ============================================================================

all_average_cost = (
    procedure_quality["BASE_COST"]
    .mean()
)

all_median_cost = (
    procedure_quality["BASE_COST"]
    .median()
)

non_431 = procedure_quality[
    procedure_quality["BASE_COST"] != 431
].copy()

non_431_average_cost = (
    non_431["BASE_COST"]
    .mean()
)

non_431_median_cost = (
    non_431["BASE_COST"]
    .median()
)

print("\nCOST DISTRIBUTION COMPARISON")
print("-" * 70)

print(
    f"Average cost — all records: "
    f"{all_average_cost:,.2f}"
)

print(
    f"Median cost — all records: "
    f"{all_median_cost:,.2f}"
)

print(
    f"Average cost — excluding 431: "
    f"{non_431_average_cost:,.2f}"
)

print(
    f"Median cost — excluding 431: "
    f"{non_431_median_cost:,.2f}"
)


# ============================================================================
# 9. FREQUENT COST VALUES
# ============================================================================

cost_frequency = (
    procedure_quality["BASE_COST"]
    .value_counts(dropna=False)
    .reset_index()
)

cost_frequency.columns = [
    "BASE_COST",
    "Record_Count"
]

print("\nTOP 20 MOST FREQUENT BASE_COST VALUES")
print("-" * 70)

print(
    cost_frequency
    .head(20)
    .to_string(index=False)
)


# ============================================================================
# 10. HIGH-COST RECORDS
# ----------------------------------------------------------------------------
# Identify extreme individual procedure costs.
# ============================================================================

cost_95th_percentile = (
    procedure_quality["BASE_COST"]
    .quantile(0.95)
)

cost_99th_percentile = (
    procedure_quality["BASE_COST"]
    .quantile(0.99)
)

high_cost_95 = procedure_quality[
    procedure_quality["BASE_COST"] >= cost_95th_percentile
].copy()

high_cost_99 = procedure_quality[
    procedure_quality["BASE_COST"] >= cost_99th_percentile
].copy()

print("\nPROCEDURE COST OUTLIER THRESHOLDS")
print("-" * 70)

print(
    f"95th percentile cost: "
    f"{cost_95th_percentile:,.2f}"
)

print(
    f"99th percentile cost: "
    f"{cost_99th_percentile:,.2f}"
)

print(
    f"Records at/above 95th percentile: "
    f"{len(high_cost_95):,}"
)

print(
    f"Records at/above 99th percentile: "
    f"{len(high_cost_99):,}"
)


# ============================================================================
# 11. TOP 20 INDIVIDUAL PROCEDURE COST RECORDS
# ============================================================================

top_individual_costs = (
    procedure_quality
    .sort_values("BASE_COST", ascending=False)
    .head(20)
    .copy()
)

print("\nTOP 20 INDIVIDUAL PROCEDURE COST RECORDS")
print("-" * 70)

print(
    top_individual_costs[
        [
            "CODE",
            "DESCRIPTION",
            "PATIENT",
            "ENCOUNTER",
            "BASE_COST",
            "BASE_COST_431_FLAG"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 12. PROCEDURE-LEVEL IMPACT OF BASE_COST = 431
# ----------------------------------------------------------------------------
# Identify procedures where most/all records carry BASE_COST = 431.
# ============================================================================

procedure_431_analysis = (
    procedure_quality
    .groupby(
        ["CODE", "DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("BASE_COST", "size"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Average_Procedure_Cost=("BASE_COST", "mean"),
        Median_Procedure_Cost=("BASE_COST", "median"),
        Cost_431_Count=(
            "BASE_COST",
            lambda x: (x == 431).sum()
        )
    )
    .reset_index()
)

procedure_431_analysis["Cost_431_Share"] = (
    procedure_431_analysis["Cost_431_Count"]
    / procedure_431_analysis["Procedure_Count"]
    * 100
)

procedure_431_analysis = (
    procedure_431_analysis
    .sort_values(
        "Cost_431_Share",
        ascending=False
    )
)

print("\nPROCEDURES MOST AFFECTED BY BASE_COST = 431")
print("-" * 70)

print(
    procedure_431_analysis[
        [
            "CODE",
            "DESCRIPTION",
            "Procedure_Count",
            "Cost_431_Count",
            "Cost_431_Share",
            "Average_Procedure_Cost"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


# ============================================================================
# 13. SENSITIVITY ANALYSIS — TOTAL COST
# ----------------------------------------------------------------------------

records_all = len(procedure_quality)
records_excluding_431 = len(non_431)

total_cost_excluding_431 = (
    non_431["BASE_COST"].sum()
)

cost_reduction = (
    total_cost_all - total_cost_excluding_431
)

print("\nSENSITIVITY ANALYSIS — BASE_COST = 431")
print("-" * 70)

print("SCENARIO 1 — ALL PROCEDURE COST RECORDS")
print(
    f"Records: {records_all:,}"
)

print(
    f"Total cost: {total_cost_all:,.2f}"
)

print("\nSCENARIO 2 — EXCLUDING BASE_COST = 431")
print(
    f"Records: {records_excluding_431:,}"
)

print(
    f"Total cost: {total_cost_excluding_431:,.2f}"
)

print(
    f"Cost difference attributable to 431 records: "
    f"{cost_reduction:,.2f}"
)


# ============================================================================
# 14. COST CONCENTRATION — ALL RECORDS
# ============================================================================

procedure_cost_all = (
    procedure_quality
    .groupby(
        ["CODE", "DESCRIPTION"],
        dropna=False
    )["BASE_COST"]
    .sum()
    .sort_values(ascending=False)
)

top_10_all_cost = (
    procedure_cost_all
    .head(10)
    .sum()
)

top_20_all_cost = (
    procedure_cost_all
    .head(20)
    .sum()
)

top_10_all_share = (
    top_10_all_cost / total_cost_all * 100
    if total_cost_all > 0
    else 0
)

top_20_all_share = (
    top_20_all_cost / total_cost_all * 100
    if total_cost_all > 0
    else 0
)


# ============================================================================
# 15. COST CONCENTRATION — EXCLUDING 431
# ============================================================================

procedure_cost_non_431 = (
    non_431
    .groupby(
        ["CODE", "DESCRIPTION"],
        dropna=False
    )["BASE_COST"]
    .sum()
    .sort_values(ascending=False)
)

top_10_non_431_cost = (
    procedure_cost_non_431
    .head(10)
    .sum()
)

top_20_non_431_cost = (
    procedure_cost_non_431
    .head(20)
    .sum()
)

top_10_non_431_share = (
    top_10_non_431_cost
    / total_cost_excluding_431
    * 100
    if total_cost_excluding_431 > 0
    else 0
)

top_20_non_431_share = (
    top_20_non_431_cost
    / total_cost_excluding_431
    * 100
    if total_cost_excluding_431 > 0
    else 0
)

print("\nCOST CONCENTRATION SENSITIVITY")
print("-" * 70)

print(
    f"Top 10 cost share — all records: "
    f"{top_10_all_share:.2f}%"
)

print(
    f"Top 20 cost share — all records: "
    f"{top_20_all_share:.2f}%"
)

print(
    f"Top 10 cost share — excluding 431: "
    f"{top_10_non_431_share:.2f}%"
)

print(
    f"Top 20 cost share — excluding 431: "
    f"{top_20_non_431_share:.2f}%"
)


# ============================================================================
# 16. CREATE ANALYTICAL SUMMARY TABLE
# ============================================================================

procedure_cost_quality_metrics = pd.DataFrame({
    "Metric": [
        "Total Procedure Records",
        "Missing BASE_COST",
        "Zero BASE_COST",
        "Negative BASE_COST",
        "Exact Duplicate Records",
        "Records with BASE_COST = 431",
        "BASE_COST = 431 Record Share (%)",
        "Total Procedure Cost",
        "Cost from BASE_COST = 431",
        "BASE_COST = 431 Cost Share (%)",
        "Total Cost Excluding 431",
        "Average Procedure Cost",
        "Median Procedure Cost",
        "Average Cost Excluding 431",
        "Median Cost Excluding 431",
        "95th Percentile Cost",
        "99th Percentile Cost",
        "Codes with Multiple Descriptions",
        "Descriptions with Multiple Codes",
        "Top 10 Cost Share — All (%)",
        "Top 20 Cost Share — All (%)",
        "Top 10 Cost Share — Excluding 431 (%)",
        "Top 20 Cost Share — Excluding 431 (%)"
    ],
    "Value": [
        total_records,
        missing_cost,
        zero_cost,
        negative_cost,
        exact_duplicates,
        cost_431_records,
        cost_431_share_records,
        total_cost_all,
        cost_431_total,
        cost_431_share,
        total_cost_excluding_431,
        all_average_cost,
        all_median_cost,
        non_431_average_cost,
        non_431_median_cost,
        cost_95th_percentile,
        cost_99th_percentile,
        len(multi_description_codes),
        len(multi_code_descriptions),
        top_10_all_share,
        top_20_all_share,
        top_10_non_431_share,
        top_20_non_431_share
    ]
})


# ============================================================================
# 17. FINAL SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18C COMPLETE")
print("=" * 70)

print(
    f"Total procedure records analysed: "
    f"{total_records:,}"
)

print(
    f"Exact duplicate records: "
    f"{exact_duplicates:,}"
)

print(
    f"Codes with multiple descriptions: "
    f"{len(multi_description_codes):,}"
)

print(
    f"Records with BASE_COST = 431: "
    f"{cost_431_records:,}"
)

print(
    f"BASE_COST = 431 record share: "
    f"{cost_431_share_records:.2f}%"
)

print(
    f"Total procedure cost: "
    f"{total_cost_all:,.2f}"
)

print(
    f"BASE_COST = 431 cost share: "
    f"{cost_431_share:.2f}%"
)

print(
    f"Total cost excluding 431: "
    f"{total_cost_excluding_431:,.2f}"
)

print(
    f"Top 10 cost share — all records: "
    f"{top_10_all_share:.2f}%"
)

print(
    f"Top 10 cost share — excluding 431: "
    f"{top_10_non_431_share:.2f}%"
)

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18D — PROCEDURE STANDARDIZATION & ANALYTICAL FLAGS
# ----------------------------------------------------------------------------
# Purpose:
#   Prepare procedure-level data for downstream analysis without modifying
#   or deleting the original procedure records.
#
# Objectives:
#   1. Preserve the original CODE and DESCRIPTION values
#   2. Identify code-description inconsistencies
#   3. Create canonical descriptions for known inconsistent codes
#   4. Flag BASE_COST = 431 records
#   5. Flag non-431 cost records
#   6. Flag high-cost and extreme-cost records
#   7. Create analytical cost bands
#   8. Validate the resulting analytical dataset
#
# IMPORTANT:
#   - No procedure records are deleted.
#   - No exact duplicates are removed.
#   - Original DESCRIPTION is preserved.
#   - BASE_COST = 431 is flagged, NOT treated as invalid.
#   - Canonical descriptions are analytical labels only.
# ============================================================================

print("=" * 70)
print("STEP 4.18D — PROCEDURE STANDARDIZATION & ANALYTICAL FLAGS")
print("=" * 70)


# ============================================================================
# 1. SOURCE DATA VALIDATION
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "BASE_COST",
    "PATIENT",
    "ENCOUNTER"
]

missing_columns = [
    col for col in required_columns
    if col not in procedures_clean.columns
]

if missing_columns:
    print("\nSTATUS: FAIL")
    print("Missing columns:", missing_columns)
    raise ValueError(
        "Required columns are missing from procedures_clean."
    )

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print("All required source columns are available.")


# ============================================================================
# 2. CREATE ANALYTICAL COPY
# ============================================================================

procedure_analysis = procedures_clean.copy()

print("\nANALYTICAL DATASET CREATED")
print("-" * 70)
print(f"Source records: {len(procedures_clean):,}")
print(f"Analytical records: {len(procedure_analysis):,}")


# ============================================================================
# 3. PRESERVE ORIGINAL IDENTIFIERS
# ============================================================================

procedure_analysis["ORIGINAL_CODE"] = procedure_analysis["CODE"]
procedure_analysis["ORIGINAL_DESCRIPTION"] = (
    procedure_analysis["DESCRIPTION"]
)

print("\nORIGINAL VALUES PRESERVED")
print("-" * 70)
print("Original CODE preserved in ORIGINAL_CODE.")
print("Original DESCRIPTION preserved in ORIGINAL_DESCRIPTION.")


# ============================================================================
# 4. CODE ↔ DESCRIPTION VARIANCE ANALYSIS
# ============================================================================

code_description_counts = (
    procedure_analysis
    .groupby("CODE")["DESCRIPTION"]
    .nunique()
)

codes_with_multiple_descriptions = (
    code_description_counts[
        code_description_counts > 1
    ]
    .index
    .tolist()
)

procedure_analysis["CODE_DESCRIPTION_VARIANCE_FLAG"] = (
    procedure_analysis["CODE"].isin(
        codes_with_multiple_descriptions
    )
)

print("\nCODE ↔ DESCRIPTION STANDARDIZATION")
print("-" * 70)
print(
    f"Codes with multiple descriptions: "
    f"{len(codes_with_multiple_descriptions):,}"
)

print(
    f"Records affected by code-description variance: "
    f"{procedure_analysis['CODE_DESCRIPTION_VARIANCE_FLAG'].sum():,}"
)


# ============================================================================
# 5. CANONICAL DESCRIPTION MAPPING
# ----------------------------------------------------------------------------
# The original DESCRIPTION is retained.
#
# These canonical labels are analytical groupings only and do not replace
# the source descriptions.
# ============================================================================

canonical_description_map = {

    # Physical examination variants
    "5880005": "Physical examination",

    # Depression screening variants
    "171207006": "Depression screening",

    # Chest X-ray variants
    "399208008": "Plain chest X-ray",

    # Cytopathology smear variants
    "90226004": (
        "Cytopathology procedure - preparation of smear - genital source"
    )
}

procedure_analysis["CANONICAL_DESCRIPTION"] = (
    procedure_analysis["CODE"]
    .map(canonical_description_map)
    .fillna(procedure_analysis["DESCRIPTION"])
)


# ============================================================================
# 6. CANONICALIZATION FLAG
# ============================================================================

procedure_analysis["CANONICAL_DESCRIPTION_FLAG"] = (
    procedure_analysis["DESCRIPTION"]
    != procedure_analysis["CANONICAL_DESCRIPTION"]
)

print("\nCANONICAL DESCRIPTION MAPPING")
print("-" * 70)

print(
    f"Records assigned a canonical description: "
    f"{procedure_analysis['CANONICAL_DESCRIPTION_FLAG'].sum():,}"
)

print("\nCODE-LEVEL CANONICAL MAPPINGS")

for code, canonical_description in canonical_description_map.items():

    descriptions = (
        procedure_analysis.loc[
            procedure_analysis["CODE"] == code,
            "DESCRIPTION"
        ]
        .dropna()
        .unique()
        .tolist()
    )

    print(f"\nCODE: {code}")
    print(f"Canonical description: {canonical_description}")
    print("Original descriptions:")

    for description in descriptions:
        print(f"  - {description}")


# ============================================================================
# 7. BASE_COST = 431 FLAG
# ----------------------------------------------------------------------------
# IMPORTANT:
#   BASE_COST = 431 is NOT treated as invalid.
#   It is flagged because STEP 4.18C demonstrated unusually high frequency.
# ============================================================================

procedure_analysis["BASE_COST_431_FLAG"] = (
    procedure_analysis["BASE_COST"] == 431
)

procedure_analysis["NON_431_COST_FLAG"] = (
    procedure_analysis["BASE_COST"] != 431
)

print("\nBASE_COST = 431 FLAGGING")
print("-" * 70)

cost_431_count = (
    procedure_analysis["BASE_COST_431_FLAG"].sum()
)

total_records = len(procedure_analysis)

cost_431_share = (
    cost_431_count / total_records * 100
)

print(f"Records with BASE_COST = 431: {cost_431_count:,}")
print(f"Total procedure records: {total_records:,}")
print(f"431 record share: {cost_431_share:.2f}%")


# ============================================================================
# 8. COST IMPACT FLAGS
# ============================================================================

procedure_analysis["BASE_COST_431_AMOUNT"] = np.where(
    procedure_analysis["BASE_COST_431_FLAG"],
    procedure_analysis["BASE_COST"],
    0
)

procedure_analysis["NON_431_COST_AMOUNT"] = np.where(
    procedure_analysis["NON_431_COST_FLAG"],
    procedure_analysis["BASE_COST"],
    0
)


# ============================================================================
# 9. COST OUTLIER THRESHOLDS
# ----------------------------------------------------------------------------
# Use the same thresholds established in STEP 4.18C.
# ============================================================================

cost_95th = procedure_analysis["BASE_COST"].quantile(0.95)
cost_99th = procedure_analysis["BASE_COST"].quantile(0.99)

procedure_analysis["HIGH_COST_FLAG"] = (
    procedure_analysis["BASE_COST"] >= cost_95th
)

procedure_analysis["EXTREME_COST_FLAG"] = (
    procedure_analysis["BASE_COST"] >= cost_99th
)

print("\nCOST OUTLIER FLAGS")
print("-" * 70)
print(f"95th percentile threshold: {cost_95th:,.2f}")
print(f"99th percentile threshold: {cost_99th:,.2f}")

print(
    f"Records flagged HIGH_COST: "
    f"{procedure_analysis['HIGH_COST_FLAG'].sum():,}"
)

print(
    f"Records flagged EXTREME_COST: "
    f"{procedure_analysis['EXTREME_COST_FLAG'].sum():,}"
)


# ============================================================================
# 10. COST BAND CLASSIFICATION
# ============================================================================

procedure_analysis["COST_BAND"] = np.select(
    [
        procedure_analysis["BASE_COST"] >= cost_99th,
        procedure_analysis["BASE_COST"] >= cost_95th,
        procedure_analysis["BASE_COST"] > 431,
        procedure_analysis["BASE_COST"] == 431
    ],
    [
        "Extreme Cost",
        "High Cost",
        "Standard Non-431 Cost",
        "431 Cost"
    ],
    default="Other"
)


# ============================================================================
# 11. ANALYTICAL INTERPRETATION FLAG
# ----------------------------------------------------------------------------
# This prevents downstream analysis from automatically interpreting 431
# as genuinely low-cost.
# ============================================================================

procedure_analysis["COST_INTERPRETATION_FLAG"] = np.where(
    procedure_analysis["BASE_COST_431_FLAG"],
    "Review — BASE_COST = 431",
    np.where(
        procedure_analysis["EXTREME_COST_FLAG"],
        "Extreme Cost",
        np.where(
            procedure_analysis["HIGH_COST_FLAG"],
            "High Cost",
            "Standard Cost"
        )
    )
)


# ============================================================================
# 12. PROCEDURE-LEVEL QUALITY FLAGS
# ============================================================================

procedure_analysis["PROCEDURE_DATA_QUALITY_FLAG"] = np.select(
    [
        procedure_analysis["CODE_DESCRIPTION_VARIANCE_FLAG"]
        & procedure_analysis["BASE_COST_431_FLAG"],

        procedure_analysis["CODE_DESCRIPTION_VARIANCE_FLAG"],

        procedure_analysis["BASE_COST_431_FLAG"]
    ],
    [
        "Code-Description Variance + 431 Cost",
        "Code-Description Variance",
        "431 Cost — Review"
    ],
    default="No Major Flag"
)


# ============================================================================
# 13. PATIENT / ENCOUNTER VALIDATION
# ============================================================================

procedure_analysis["MISSING_PATIENT_FLAG"] = (
    procedure_analysis["PATIENT"].isna()
)

procedure_analysis["MISSING_ENCOUNTER_FLAG"] = (
    procedure_analysis["ENCOUNTER"].isna()
)

print("\nPATIENT / ENCOUNTER QUALITY")
print("-" * 70)

print(
    f"Missing PATIENT: "
    f"{procedure_analysis['MISSING_PATIENT_FLAG'].sum():,}"
)

print(
    f"Missing ENCOUNTER: "
    f"{procedure_analysis['MISSING_ENCOUNTER_FLAG'].sum():,}"
)


# ============================================================================
# 14. OVERALL ANALYTICAL QUALITY FLAG
# ============================================================================

procedure_analysis["OVERALL_QUALITY_FLAG"] = np.select(
    [
        procedure_analysis["MISSING_PATIENT_FLAG"]
        | procedure_analysis["MISSING_ENCOUNTER_FLAG"],

        procedure_analysis["CODE_DESCRIPTION_VARIANCE_FLAG"]
        & procedure_analysis["BASE_COST_431_FLAG"],

        procedure_analysis["CODE_DESCRIPTION_VARIANCE_FLAG"],

        procedure_analysis["BASE_COST_431_FLAG"]
    ],
    [
        "Missing Patient/Encounter",
        "Code-Description Variance + 431 Cost",
        "Code-Description Variance",
        "431 Cost — Review"
    ],
    default="Clean for Analysis"
)


# ============================================================================
# 15. STANDARDIZATION SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STANDARDIZATION SUMMARY")
print("=" * 70)

summary = pd.DataFrame({
    "Metric": [
        "Total procedure records",
        "Unique procedure codes",
        "Unique original descriptions",
        "Codes with multiple descriptions",
        "Records with code-description variance",
        "Records with BASE_COST = 431",
        "431 cost record share (%)",
        "Records above 95th percentile",
        "Records above 99th percentile",
        "Missing patients",
        "Missing encounters"
    ],
    "Value": [
        len(procedure_analysis),
        procedure_analysis["CODE"].nunique(),
        procedure_analysis["DESCRIPTION"].nunique(),
        len(codes_with_multiple_descriptions),
        int(
            procedure_analysis[
                "CODE_DESCRIPTION_VARIANCE_FLAG"
            ].sum()
        ),
        int(
            procedure_analysis[
                "BASE_COST_431_FLAG"
            ].sum()
        ),
        round(cost_431_share, 2),
        int(
            procedure_analysis[
                "HIGH_COST_FLAG"
            ].sum()
        ),
        int(
            procedure_analysis[
                "EXTREME_COST_FLAG"
            ].sum()
        ),
        int(
            procedure_analysis[
                "MISSING_PATIENT_FLAG"
            ].sum()
        ),
        int(
            procedure_analysis[
                "MISSING_ENCOUNTER_FLAG"
            ].sum()
        )
    ]
})

print(summary.to_string(index=False))


# ============================================================================
# 16. COST BAND DISTRIBUTION
# ============================================================================

print("\nCOST BAND DISTRIBUTION")
print("-" * 70)

cost_band_summary = (
    procedure_analysis["COST_BAND"]
    .value_counts()
    .rename_axis("Cost_Band")
    .reset_index(name="Record_Count")
)

cost_band_summary["Record_Share_%"] = (
    cost_band_summary["Record_Count"]
    / len(procedure_analysis)
    * 100
)

print(
    cost_band_summary.to_string(index=False)
)


# ============================================================================
# 17. PROCEDURE CODE STANDARDIZATION SUMMARY
# ============================================================================

print("\nCODE → CANONICAL DESCRIPTION SUMMARY")
print("-" * 70)

standardized_codes = (
    procedure_analysis[
        procedure_analysis["CODE"].isin(
            canonical_description_map.keys()
        )
    ][
        [
            "CODE",
            "DESCRIPTION",
            "CANONICAL_DESCRIPTION"
        ]
    ]
    .drop_duplicates()
    .sort_values(["CODE", "DESCRIPTION"])
)

print(
    standardized_codes.to_string(index=False)
)


# ============================================================================
# 18. PROCEDURES WITH 100% BASE_COST = 431
# ----------------------------------------------------------------------------
# These procedures are especially important for later interpretation.
# ============================================================================

procedure_431_summary = (
    procedure_analysis
    .groupby(
        ["CODE", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("BASE_COST", "size"),
        Cost_431_Count=("BASE_COST_431_FLAG", "sum"),
        Average_Cost=("BASE_COST", "mean"),
        Total_Cost=("BASE_COST", "sum")
    )
    .reset_index()
)

procedure_431_summary["Cost_431_Share_%"] = (
    procedure_431_summary["Cost_431_Count"]
    / procedure_431_summary["Procedure_Count"]
    * 100
)

procedure_431_summary = (
    procedure_431_summary
    .sort_values(
        ["Cost_431_Share_%", "Procedure_Count"],
        ascending=[False, False]
    )
)

print("\nPROCEDURES WITH 100% BASE_COST = 431")
print("-" * 70)

all_431_procedures = procedure_431_summary[
    procedure_431_summary["Cost_431_Share_%"] == 100
]

print(
    all_431_procedures.head(20).to_string(index=False)
)

print(
    f"\nProcedure groups with 100% BASE_COST = 431: "
    f"{len(all_431_procedures):,}"
)


# ============================================================================
# 19. HIGH-COST NON-431 PROCEDURES
# ----------------------------------------------------------------------------
# This is useful because it isolates genuinely high observed costs from the
# large 431-cost mass.
# ============================================================================

high_cost_non_431 = (
    procedure_analysis[
        procedure_analysis["NON_431_COST_FLAG"]
        & procedure_analysis["HIGH_COST_FLAG"]
    ]
    .sort_values("BASE_COST", ascending=False)
    .head(20)
    .copy()
)

print("\nTOP 20 HIGH-COST NON-431 PROCEDURE RECORDS")
print("-" * 70)

print(
    high_cost_non_431[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "PATIENT",
            "ENCOUNTER",
            "BASE_COST",
            "COST_BAND",
            "COST_INTERPRETATION_FLAG"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 20. CREATE PROCEDURE-LEVEL ANALYTICAL SUMMARY
# ----------------------------------------------------------------------------
# This summary is based on canonical descriptions but retains the original
# procedure-code structure.
# ============================================================================

procedure_analytical_summary = (
    procedure_analysis
    .groupby(
        ["CODE", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("BASE_COST", "size"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Average_Procedure_Cost=("BASE_COST", "mean"),
        Median_Procedure_Cost=("BASE_COST", "median"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Cost_431_Count=("BASE_COST_431_FLAG", "sum")
    )
    .reset_index()
)

procedure_analytical_summary["Cost_Per_Patient"] = np.where(
    procedure_analytical_summary["Unique_Patients"] > 0,
    procedure_analytical_summary["Total_Procedure_Cost"]
    / procedure_analytical_summary["Unique_Patients"],
    np.nan
)

procedure_analytical_summary["Procedures_Per_Patient"] = np.where(
    procedure_analytical_summary["Unique_Patients"] > 0,
    procedure_analytical_summary["Procedure_Count"]
    / procedure_analytical_summary["Unique_Patients"],
    np.nan
)

procedure_analytical_summary["Procedures_Per_Encounter"] = np.where(
    procedure_analytical_summary["Unique_Encounters"] > 0,
    procedure_analytical_summary["Procedure_Count"]
    / procedure_analytical_summary["Unique_Encounters"],
    np.nan
)

procedure_analytical_summary["Cost_431_Share_%"] = np.where(
    procedure_analytical_summary["Procedure_Count"] > 0,
    procedure_analytical_summary["Cost_431_Count"]
    / procedure_analytical_summary["Procedure_Count"]
    * 100,
    np.nan
)


# ============================================================================
# 21. FINAL VALIDATION
# ============================================================================

print("\n" + "=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

validation_checks = {
    "Record count preserved":
        len(procedure_analysis) == len(procedures_clean),

    "Original CODE preserved":
        "ORIGINAL_CODE" in procedure_analysis.columns,

    "Original DESCRIPTION preserved":
        "ORIGINAL_DESCRIPTION" in procedure_analysis.columns,

    "Canonical description created":
        "CANONICAL_DESCRIPTION" in procedure_analysis.columns,

    "431 flag created":
        "BASE_COST_431_FLAG" in procedure_analysis.columns,

    "High-cost flag created":
        "HIGH_COST_FLAG" in procedure_analysis.columns,

    "Extreme-cost flag created":
        "EXTREME_COST_FLAG" in procedure_analysis.columns,

    "Cost band created":
        "COST_BAND" in procedure_analysis.columns,

    "No records deleted":
        len(procedure_analysis) == 47701
}

validation_results = pd.DataFrame(
    list(validation_checks.items()),
    columns=["Validation", "Result"]
)

print(validation_results.to_string(index=False))

if not all(validation_checks.values()):
    raise ValueError(
        "STEP 4.18D validation failed. "
        "Review the validation results above."
    )


# ============================================================================
# 22. FINAL OUTPUTS
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18D COMPLETE")
print("=" * 70)

print(
    f"Total procedure records: "
    f"{len(procedure_analysis):,}"
)

print(
    f"Unique procedure codes: "
    f"{procedure_analysis['CODE'].nunique():,}"
)

print(
    f"Codes with description variance: "
    f"{len(codes_with_multiple_descriptions):,}"
)

print(
    f"Records with BASE_COST = 431: "
    f"{procedure_analysis['BASE_COST_431_FLAG'].sum():,}"
)

print(
    f"High-cost records: "
    f"{procedure_analysis['HIGH_COST_FLAG'].sum():,}"
)

print(
    f"Extreme-cost records: "
    f"{procedure_analysis['EXTREME_COST_FLAG'].sum():,}"
)

print("\nAnalytical datasets created:")
print("  - procedure_analysis")
print("  - procedure_analytical_summary")
print("  - procedure_431_summary")
print("  - high_cost_non_431")
print("  - cost_band_summary")

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18E — PROCEDURE COST & UTILIZATION INSIGHT ANALYSIS
# ----------------------------------------------------------------------------
# Purpose:
#   Convert the standardized procedure dataset from STEP 4.18D into
#   decision-ready procedure cost/utilization insights.
#
# Analytical principles:
#   1. Preserve all procedure records.
#   2. Preserve ORIGINAL_CODE and ORIGINAL_DESCRIPTION.
#   3. Use CODE as the primary procedure identifier.
#   4. Treat BASE_COST = 431 as an analytical flag, NOT as a deleted record.
#   5. Separate observed cost from 431-flagged/default-cost exposure.
#   6. Prioritize high-cost, high-utilization and high-cost-per-patient
#      procedures.
#   7. Avoid interpreting BASE_COST = 431 as automatically "low cost".
# ============================================================================

print("=" * 70)
print("STEP 4.18E — PROCEDURE COST & UTILIZATION INSIGHT ANALYSIS")
print("=" * 70)


# ============================================================================
# 1. SOURCE DATA CHECK
# ============================================================================

required_source_columns = [
    "CODE",
    "DESCRIPTION",
    "PATIENT",
    "ENCOUNTER",
    "BASE_COST"
]

# procedure_analysis was created in STEP 4.18D
if "procedure_analysis" not in globals():
    raise ValueError(
        "procedure_analysis is not available. "
        "Please run STEP 4.18D before STEP 4.18E."
    )

missing_source_columns = [
    col for col in required_source_columns
    if col not in procedure_analysis.columns
]

if missing_source_columns:
    print("\nSTATUS: FAIL")
    print("Missing columns:", missing_source_columns)
    raise ValueError(
        "Required columns are missing from procedure_analysis."
    )

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print("procedure_analysis is available.")
print(f"Procedure records available: {len(procedure_analysis):,}")


# ============================================================================
# 2. PREPARE ANALYTICAL DATASET
# ============================================================================

procedure_e = procedure_analysis.copy()

# ---------------------------------------------------------------------------
# Ensure cost flag exists
# ---------------------------------------------------------------------------

if "BASE_COST_431_FLAG" not in procedure_e.columns:
    procedure_e["BASE_COST_431_FLAG"] = np.where(
        procedure_e["BASE_COST"].eq(431),
        "BASE_COST_431",
        "OTHER_COST"
    )

# ---------------------------------------------------------------------------
# Ensure canonical description exists
# ---------------------------------------------------------------------------

if "CANONICAL_DESCRIPTION" not in procedure_e.columns:

    procedure_e["CANONICAL_DESCRIPTION"] = (
        procedure_e["DESCRIPTION"]
        .astype(str)
        .str.strip()
    )

# Fill any missing canonical descriptions
procedure_e["CANONICAL_DESCRIPTION"] = (
    procedure_e["CANONICAL_DESCRIPTION"]
    .replace(["", "nan", "None"], np.nan)
    .fillna(procedure_e["DESCRIPTION"])
)

# ---------------------------------------------------------------------------
# Create non-431 cost
# ---------------------------------------------------------------------------

procedure_e["NON_431_COST"] = np.where(
    procedure_e["BASE_COST"].ne(431),
    procedure_e["BASE_COST"],
    0
)

# ---------------------------------------------------------------------------
# Cost classification
# ---------------------------------------------------------------------------

if "COST_BAND" not in procedure_e.columns:

    p95 = procedure_e["BASE_COST"].quantile(0.95)
    p99 = procedure_e["BASE_COST"].quantile(0.99)

    procedure_e["COST_BAND"] = np.select(
        [
            procedure_e["BASE_COST"].eq(431),
            procedure_e["BASE_COST"] >= p99,
            procedure_e["BASE_COST"] >= p95
        ],
        [
            "431 Cost",
            "Extreme Cost",
            "High Cost"
        ],
        default="Standard Non-431 Cost"
    )

# ============================================================================
# 3. BASIC ANALYTICAL VALIDATION
# ============================================================================

print("\nANALYTICAL DATASET VALIDATION")
print("-" * 70)

print(f"Records preserved: {len(procedure_e):,}")
print(f"Unique procedure codes: {procedure_e['CODE'].nunique():,}")
print(
    f"Unique original descriptions: "
    f"{procedure_e['DESCRIPTION'].nunique():,}"
)
print(
    f"Unique canonical descriptions: "
    f"{procedure_e['CANONICAL_DESCRIPTION'].nunique():,}"
)

print(
    f"BASE_COST = 431 records: "
    f"{procedure_e['BASE_COST'].eq(431).sum():,}"
)

print(
    f"Non-431 records: "
    f"{procedure_e['BASE_COST'].ne(431).sum():,}"
)


# ============================================================================
# 4. BUILD PROCEDURE-LEVEL INSIGHT SUMMARY
# ============================================================================

procedure_insight_summary = (
    procedure_e
    .groupby(
        ["CODE", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("CODE", "size"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Average_Procedure_Cost=("BASE_COST", "mean"),
        Median_Procedure_Cost=("BASE_COST", "median"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Cost_431_Count=("BASE_COST", lambda x: (x == 431).sum()),
        Non_431_Count=("BASE_COST", lambda x: (x != 431).sum()),
        Non_431_Total_Cost=("NON_431_COST", "sum")
    )
    .reset_index()
)

# ---------------------------------------------------------------------------
# Derived metrics
# ---------------------------------------------------------------------------

procedure_insight_summary["Cost_Per_Patient"] = (
    procedure_insight_summary["Total_Procedure_Cost"]
    / procedure_insight_summary["Unique_Patients"].replace(0, np.nan)
)

procedure_insight_summary["Procedures_Per_Patient"] = (
    procedure_insight_summary["Procedure_Count"]
    / procedure_insight_summary["Unique_Patients"].replace(0, np.nan)
)

procedure_insight_summary["Procedures_Per_Encounter"] = (
    procedure_insight_summary["Procedure_Count"]
    / procedure_insight_summary["Unique_Encounters"].replace(0, np.nan)
)

procedure_insight_summary["Cost_431_Share"] = (
    procedure_insight_summary["Cost_431_Count"]
    / procedure_insight_summary["Procedure_Count"].replace(0, np.nan)
    * 100
)

procedure_insight_summary["Non_431_Cost_Share"] = np.where(
    procedure_insight_summary["Total_Procedure_Cost"] > 0,
    procedure_insight_summary["Non_431_Total_Cost"]
    / procedure_insight_summary["Total_Procedure_Cost"]
    * 100,
    0
)

procedure_insight_summary["Procedure_Utilization_Profile"] = np.where(
    procedure_insight_summary["Procedure_Count"]
    >= procedure_insight_summary["Procedure_Count"].median(),
    "High Utilization",
    "Low Utilization"
)

procedure_insight_summary["Cost_Profile"] = np.where(
    procedure_insight_summary["Non_431_Total_Cost"]
    >= procedure_insight_summary["Non_431_Total_Cost"].median(),
    "High Non-431 Cost",
    "Low Non-431 Cost"
)

procedure_insight_summary["Cost_Data_Profile"] = np.select(
    [
        procedure_insight_summary["Cost_431_Share"].eq(100),
        procedure_insight_summary["Cost_431_Share"].gt(50),
        procedure_insight_summary["Cost_431_Share"].gt(0)
    ],
    [
        "100% 431 Cost",
        "Majority 431 Cost",
        "Mixed 431 / Non-431 Cost"
    ],
    default="Non-431 Cost"
)


# ============================================================================
# 5. HIGH-COST PROCEDURES — NON-431 FOCUS
# ============================================================================

high_cost_non431_procedures = (
    procedure_insight_summary[
        procedure_insight_summary["Non_431_Count"] > 0
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY NON-431 TOTAL COST")
print("-" * 70)

print(
    high_cost_non431_procedures[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Count",
            "Non_431_Total_Cost",
            "Average_Procedure_Cost",
            "Unique_Patients",
            "Cost_Per_Patient",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 6. HIGH UTILIZATION PROCEDURES
# ============================================================================

high_utilization_procedure_insights = (
    procedure_insight_summary
    .sort_values(
        ["Procedure_Count", "Unique_Patients"],
        ascending=[False, False]
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY UTILIZATION")
print("-" * 70)

print(
    high_utilization_procedure_insights[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Procedures_Per_Patient",
            "Procedures_Per_Encounter",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 7. HIGH COST PER PATIENT
# ============================================================================

high_cost_per_patient_insights = (
    procedure_insight_summary[
        (procedure_insight_summary["Unique_Patients"] > 0) &
        (procedure_insight_summary["Non_431_Count"] > 0)
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY COST PER PATIENT")
print("-" * 70)

print(
    high_cost_per_patient_insights[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 8. HIGH REPEAT-UTILIZATION PROCEDURES
# ============================================================================

high_repeat_utilization = (
    procedure_insight_summary[
        procedure_insight_summary["Unique_Patients"] > 0
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY PROCEDURES PER PATIENT")
print("-" * 70)

print(
    high_repeat_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Procedures_Per_Patient",
            "Procedures_Per_Encounter",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 9. HIGH-COST + HIGH-UTILIZATION
# ============================================================================

cost_threshold = procedure_insight_summary[
    "Non_431_Total_Cost"
].median()

volume_threshold = procedure_insight_summary[
    "Procedure_Count"
].median()

high_cost_high_utilization = (
    procedure_insight_summary[
        (procedure_insight_summary["Non_431_Total_Cost"] >= cost_threshold) &
        (procedure_insight_summary["Procedure_Count"] >= volume_threshold) &
        (procedure_insight_summary["Non_431_Count"] > 0)
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .copy()
)

print("\nHIGH NON-431 COST + HIGH UTILIZATION PROCEDURES")
print("-" * 70)

print(
    high_cost_high_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Count",
            "Non_431_Total_Cost",
            "Unique_Patients",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 10. HIGH COST + LOW UTILIZATION
# ============================================================================

high_cost_low_utilization = (
    procedure_insight_summary[
        (procedure_insight_summary["Non_431_Total_Cost"] >= cost_threshold) &
        (procedure_insight_summary["Procedure_Count"] < volume_threshold) &
        (procedure_insight_summary["Non_431_Count"] > 0)
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .copy()
)

print("\nHIGH NON-431 COST + LOW UTILIZATION PROCEDURES")
print("-" * 70)

print(
    high_cost_low_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Count",
            "Non_431_Total_Cost",
            "Average_Procedure_Cost",
            "Cost_Per_Patient",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 11. HIGH UTILIZATION + LOW COST
# ============================================================================

low_cost_high_utilization = (
    procedure_insight_summary[
        (procedure_insight_summary["Non_431_Total_Cost"] < cost_threshold) &
        (procedure_insight_summary["Procedure_Count"] >= volume_threshold)
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .copy()
)

print("\nLOW NON-431 COST + HIGH UTILIZATION PROCEDURES")
print("-" * 70)

print(
    low_cost_high_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Procedures_Per_Patient",
            "Total_Procedure_Cost",
            "Non_431_Total_Cost",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 12. 100% BASE_COST = 431 PROCEDURES
# ============================================================================

procedures_100pct_431 = (
    procedure_insight_summary[
        procedure_insight_summary["Cost_431_Share"].eq(100)
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .copy()
)

print("\nPROCEDURES WITH 100% BASE_COST = 431")
print("-" * 70)

print(
    procedures_100pct_431[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)

print(
    f"\nProcedure groups with 100% BASE_COST = 431: "
    f"{len(procedures_100pct_431):,}"
)


# ============================================================================
# 13. COST / UTILIZATION PRIORITY MATRIX
# ============================================================================

procedure_priority = procedure_insight_summary.copy()

procedure_priority["Priority_Profile"] = np.select(
    [
        (
            (procedure_priority["Non_431_Total_Cost"] >= cost_threshold) &
            (procedure_priority["Procedure_Count"] >= volume_threshold) &
            (procedure_priority["Non_431_Count"] > 0)
        ),

        (
            (procedure_priority["Non_431_Total_Cost"] >= cost_threshold) &
            (procedure_priority["Procedure_Count"] < volume_threshold) &
            (procedure_priority["Non_431_Count"] > 0)
        ),

        (
            (procedure_priority["Non_431_Total_Cost"] < cost_threshold) &
            (procedure_priority["Procedure_Count"] >= volume_threshold)
        ),

        (
            procedure_priority["Cost_431_Share"].eq(100)
        )
    ],
    [
        "PRIORITY 1 — High Cost / High Utilization",
        "PRIORITY 2 — High Cost / Low Utilization",
        "PRIORITY 3 — High Utilization / Lower Non-431 Cost",
        "PRIORITY 4 — 431 Cost Dominated"
    ],
    default="STANDARD"
)

priority_counts = (
    procedure_priority["Priority_Profile"]
    .value_counts()
    .rename_axis("Priority_Profile")
    .reset_index(name="Procedure_Group_Count")
)

print("\nPROCEDURE PRIORITY MATRIX")
print("-" * 70)

print(
    priority_counts.to_string(index=False)
)


# ============================================================================
# 14. COST CONCENTRATION
# ============================================================================

total_observed_cost = procedure_insight_summary[
    "Total_Procedure_Cost"
].sum()

total_non431_cost = procedure_insight_summary[
    "Non_431_Total_Cost"
].sum()

top10_total_cost = (
    procedure_insight_summary
    .nlargest(10, "Total_Procedure_Cost")
    ["Total_Procedure_Cost"]
    .sum()
)

top20_total_cost = (
    procedure_insight_summary
    .nlargest(20, "Total_Procedure_Cost")
    ["Total_Procedure_Cost"]
    .sum()
)

top10_non431_cost = (
    procedure_insight_summary
    .nlargest(10, "Non_431_Total_Cost")
    ["Non_431_Total_Cost"]
    .sum()
)

top20_non431_cost = (
    procedure_insight_summary
    .nlargest(20, "Non_431_Total_Cost")
    ["Non_431_Total_Cost"]
    .sum()
)

print("\nPROCEDURE COST CONCENTRATION")
print("-" * 70)

print(
    f"Total observed procedure cost: "
    f"{total_observed_cost:,.2f}"
)

print(
    f"Total non-431 procedure cost: "
    f"{total_non431_cost:,.2f}"
)

print(
    f"Top 10 total-cost share: "
    f"{(top10_total_cost / total_observed_cost) * 100:.2f}%"
)

print(
    f"Top 20 total-cost share: "
    f"{(top20_total_cost / total_observed_cost) * 100:.2f}%"
)

print(
    f"Top 10 non-431 cost share: "
    f"{(top10_non431_cost / total_non431_cost) * 100:.2f}%"
)

print(
    f"Top 20 non-431 cost share: "
    f"{(top20_non431_cost / total_non431_cost) * 100:.2f}%"
)


# ============================================================================
# 15. UTILIZATION CONCENTRATION
# ============================================================================

total_procedures = procedure_insight_summary[
    "Procedure_Count"
].sum()

top10_volume = (
    procedure_insight_summary
    .nlargest(10, "Procedure_Count")
    ["Procedure_Count"]
    .sum()
)

top20_volume = (
    procedure_insight_summary
    .nlargest(20, "Procedure_Count")
    ["Procedure_Count"]
    .sum()
)

print("\nPROCEDURE UTILIZATION CONCENTRATION")
print("-" * 70)

print(
    f"Total procedure records: "
    f"{total_procedures:,.0f}"
)

print(
    f"Top 10 procedure volume share: "
    f"{(top10_volume / total_procedures) * 100:.2f}%"
)

print(
    f"Top 20 procedure volume share: "
    f"{(top20_volume / total_procedures) * 100:.2f}%"
)


# ============================================================================
# 16. BASE_COST = 431 IMPACT SUMMARY
# ============================================================================

records_431 = procedure_e["BASE_COST"].eq(431).sum()
records_non431 = procedure_e["BASE_COST"].ne(431).sum()

cost_431 = procedure_e.loc[
    procedure_e["BASE_COST"].eq(431),
    "BASE_COST"
].sum()

cost_non431 = procedure_e.loc[
    procedure_e["BASE_COST"].ne(431),
    "BASE_COST"
].sum()

print("\nBASE_COST = 431 ANALYTICAL IMPACT")
print("-" * 70)

print(
    f"431 records: {records_431:,}"
)

print(
    f"Non-431 records: {records_non431:,}"
)

print(
    f"431 record share: "
    f"{(records_431 / len(procedure_e)) * 100:.2f}%"
)

print(
    f"431 cost share: "
    f"{(cost_431 / total_observed_cost) * 100:.2f}%"
)

print(
    f"Non-431 cost share: "
    f"{(cost_non431 / total_observed_cost) * 100:.2f}%"
)


# ============================================================================
# 17. ANALYTICAL FLAGS
# ============================================================================

procedure_priority["High_Cost_Flag"] = (
    procedure_priority["Non_431_Total_Cost"] >= cost_threshold
)

procedure_priority["High_Utilization_Flag"] = (
    procedure_priority["Procedure_Count"] >= volume_threshold
)

procedure_priority["High_Cost_Per_Patient_Flag"] = (
    procedure_priority["Cost_Per_Patient"]
    >= procedure_priority["Cost_Per_Patient"].median()
)

procedure_priority["High_Repeat_Utilization_Flag"] = (
    procedure_priority["Procedures_Per_Patient"]
    >= procedure_priority["Procedures_Per_Patient"].median()
)

procedure_priority["431_Dominated_Flag"] = (
    procedure_priority["Cost_431_Share"].eq(100)
)


# ============================================================================
# 18. FINAL DECISION-READY DATASET
# ============================================================================

procedure_priority = procedure_priority.sort_values(
    [
        "Priority_Profile",
        "Non_431_Total_Cost",
        "Procedure_Count"
    ],
    ascending=[True, False, False]
).reset_index(drop=True)


# ============================================================================
# 19. OVERALL ANALYTICAL METRICS
# ============================================================================

procedure_insight_metrics = pd.DataFrame({
    "Metric": [
        "Total Procedure Records",
        "Unique Procedure Codes",
        "Unique Canonical Descriptions",
        "Total Observed Procedure Cost",
        "Total Non-431 Procedure Cost",
        "431 Record Share (%)",
        "431 Cost Share (%)",
        "Top 10 Total Cost Share (%)",
        "Top 20 Total Cost Share (%)",
        "Top 10 Non-431 Cost Share (%)",
        "Top 20 Non-431 Cost Share (%)",
        "Top 10 Utilization Share (%)",
        "Top 20 Utilization Share (%)",
        "High Cost / High Utilization Groups",
        "High Cost / Low Utilization Groups",
        "Low Cost / High Utilization Groups",
        "431-Dominated Groups"
    ],
    "Value": [
        total_procedures,
        procedure_insight_summary["CODE"].nunique(),
        procedure_insight_summary["CANONICAL_DESCRIPTION"].nunique(),
        total_observed_cost,
        total_non431_cost,
        (records_431 / len(procedure_e)) * 100,
        (cost_431 / total_observed_cost) * 100,
        (top10_total_cost / total_observed_cost) * 100,
        (top20_total_cost / total_observed_cost) * 100,
        (top10_non431_cost / total_non431_cost) * 100,
        (top20_non431_cost / total_non431_cost) * 100,
        (top10_volume / total_procedures) * 100,
        (top20_volume / total_procedures) * 100,
        (
            procedure_priority["Priority_Profile"]
            .eq("PRIORITY 1 — High Cost / High Utilization")
            .sum()
        ),
        (
            procedure_priority["Priority_Profile"]
            .eq("PRIORITY 2 — High Cost / Low Utilization")
            .sum()
        ),
        (
            procedure_priority["Priority_Profile"]
            .eq("PRIORITY 3 — High Utilization / Lower Non-431 Cost")
            .sum()
        ),
        (
            procedure_priority["Priority_Profile"]
            .eq("PRIORITY 4 — 431 Cost Dominated")
            .sum()
        )
    ]
})


# ============================================================================
# 20. FINAL VALIDATION
# ============================================================================

print("\n" + "=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

validation_results = pd.DataFrame({
    "Validation": [
        "Record count preserved",
        "No records deleted",
        "Original CODE preserved",
        "Original DESCRIPTION preserved",
        "BASE_COST = 431 flag available",
        "Non-431 cost available",
        "Procedure-level summary created",
        "Priority profile created"
    ],
    "Result": [
        len(procedure_e) == len(procedure_analysis),
        True,
        "ORIGINAL_CODE" in procedure_e.columns,
        "ORIGINAL_DESCRIPTION" in procedure_e.columns,
        "BASE_COST_431_FLAG" in procedure_e.columns,
        "NON_431_COST" in procedure_e.columns,
        len(procedure_insight_summary) > 0,
        "Priority_Profile" in procedure_priority.columns
    ]
})

print(validation_results.to_string(index=False))


# ============================================================================
# 21. FINAL OUTPUT SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18E COMPLETE")
print("=" * 70)

print(
    f"Total procedure records: "
    f"{len(procedure_e):,}"
)

print(
    f"Unique procedure codes: "
    f"{procedure_insight_summary['CODE'].nunique():,}"
)

print(
    f"Total observed procedure cost: "
    f"{total_observed_cost:,.2f}"
)

print(
    f"Total non-431 procedure cost: "
    f"{total_non431_cost:,.2f}"
)

print(
    f"431 record share: "
    f"{(records_431 / len(procedure_e)) * 100:.2f}%"
)

print(
    f"431 cost share: "
    f"{(cost_431 / total_observed_cost) * 100:.2f}%"
)

print(
    f"Top 10 non-431 cost concentration: "
    f"{(top10_non431_cost / total_non431_cost) * 100:.2f}%"
)

print(
    f"Top 20 non-431 cost concentration: "
    f"{(top20_non431_cost / total_non431_cost) * 100:.2f}%"
)

print(
    f"Priority 1 groups: "
    f"{(
        procedure_priority['Priority_Profile']
        .eq('PRIORITY 1 — High Cost / High Utilization')
    ).sum():,}"
)

print(
    f"Priority 2 groups: "
    f"{(
        procedure_priority['Priority_Profile']
        .eq('PRIORITY 2 — High Cost / Low Utilization')
    ).sum():,}"
)

print(
    f"431-dominated groups: "
    f"{(
        procedure_priority['Priority_Profile']
        .eq('PRIORITY 4 — 431 Cost Dominated')
    ).sum():,}"
)


# ============================================================================
# DATASETS CREATED
# ============================================================================

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_e")
print("  - procedure_insight_summary")
print("  - high_cost_non431_procedures")
print("  - high_utilization_procedure_insights")
print("  - high_cost_per_patient_insights")
print("  - high_repeat_utilization")
print("  - high_cost_high_utilization")
print("  - high_cost_low_utilization")
print("  - low_cost_high_utilization")
print("  - procedures_100pct_431")
print("  - procedure_priority")
print("  - procedure_insight_metrics")

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18F — PROCEDURE STANDARDIZATION VALIDATION & CLINICAL/COST SEGMENTATION
# ----------------------------------------------------------------------------
# Purpose:
#   Validate procedure standardization, preserve source integrity, and create
#   clinically and financially meaningful analytical segments.
#
# Principles:
#   - DO NOT delete procedure records.
#   - DO NOT modify ORIGINAL_CODE or ORIGINAL_DESCRIPTION.
#   - Treat BASE_COST = 431 as an analytical flag, not automatically invalid.
#   - Use CODE as the stable procedure identifier.
#   - Validate canonical description consistency.
#   - Separate 431-dominated activity from meaningful non-431 cost analysis.
#   - Preserve both volume and financial perspectives.
# ============================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18F — PROCEDURE STANDARDIZATION VALIDATION &")
print("CLINICAL/COST SEGMENTATION")
print("=" * 70)


# ============================================================================
# 1. SOURCE DATA CHECK
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "ORIGINAL_CODE",
    "ORIGINAL_DESCRIPTION",
    "BASE_COST",
    "BASE_COST_431_FLAG",
    "COST_BAND",
    "COST_INTERPRETATION_FLAG",
    "PATIENT",
    "ENCOUNTER",
    "CANONICAL_DESCRIPTION"
]

missing_columns = [
    col for col in required_columns
    if col not in procedure_analysis.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

df = procedure_analysis.copy()

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print(f"Procedure records available: {len(df):,}")


# ============================================================================
# 2. RECORD PRESERVATION CHECK
# ============================================================================

source_record_count = len(df)

print("\nRECORD PRESERVATION")
print("-" * 70)
print(f"Source records: {source_record_count:,}")
print(f"Analytical records: {len(df):,}")
print(
    f"Record count preserved: "
    f"{source_record_count == len(df)}"
)

if source_record_count != len(df):
    raise ValueError(
        "Record count changed unexpectedly."
    )


# ============================================================================
# 3. ORIGINAL VALUE PRESERVATION
# ============================================================================

original_code_preserved = (
    df["ORIGINAL_CODE"].notna().all()
)

original_description_preserved = (
    df["ORIGINAL_DESCRIPTION"].notna().all()
)

print("\nORIGINAL VALUE PRESERVATION")
print("-" * 70)
print(
    f"Original CODE preserved: "
    f"{original_code_preserved}"
)
print(
    f"Original DESCRIPTION preserved: "
    f"{original_description_preserved}"
)


# ============================================================================
# 4. CODE → DESCRIPTION STANDARDIZATION VALIDATION
# ============================================================================

code_description_summary = (
    df.groupby("CODE")
      .agg(
          Unique_Original_Descriptions=(
              "ORIGINAL_DESCRIPTION",
              "nunique"
          ),
          Unique_Canonical_Descriptions=(
              "CANONICAL_DESCRIPTION",
              "nunique"
          ),
          Procedure_Count=("CODE", "size")
      )
      .reset_index()
)

code_variance = code_description_summary[
    code_description_summary[
        "Unique_Original_Descriptions"
    ] > 1
].copy()

canonical_variance = code_description_summary[
    code_description_summary[
        "Unique_Canonical_Descriptions"
    ] > 1
].copy()

print("\nCODE → DESCRIPTION STANDARDIZATION VALIDATION")
print("-" * 70)

print(
    "Codes with multiple original descriptions:",
    len(code_variance)
)

print(
    "Codes with multiple canonical descriptions:",
    len(canonical_variance)
)

print("\nCODES WITH ORIGINAL DESCRIPTION VARIANCE")
print("-" * 70)

if len(code_variance) > 0:
    print(
        code_variance[
            [
                "CODE",
                "Unique_Original_Descriptions",
                "Unique_Canonical_Descriptions",
                "Procedure_Count"
            ]
        ].to_string(index=False)
    )
else:
    print("None")


print("\nCODES WITH MULTIPLE CANONICAL DESCRIPTIONS")
print("-" * 70)

if len(canonical_variance) > 0:
    print(
        canonical_variance[
            [
                "CODE",
                "Unique_Original_Descriptions",
                "Unique_Canonical_Descriptions",
                "Procedure_Count"
            ]
        ].to_string(index=False)
    )
else:
    print("None")


# ============================================================================
# 5. CANONICAL DESCRIPTION QUALITY
# ============================================================================

missing_canonical = (
    df["CANONICAL_DESCRIPTION"]
    .isna()
    .sum()
)

blank_canonical = (
    df["CANONICAL_DESCRIPTION"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("\nCANONICAL DESCRIPTION QUALITY")
print("-" * 70)
print(f"Missing canonical descriptions: {missing_canonical:,}")
print(f"Blank canonical descriptions: {blank_canonical:,}")

canonical_quality_pass = (
    missing_canonical == 0
    and blank_canonical == 0
)

print(
    "Canonical description availability:",
    "PASS" if canonical_quality_pass else "REVIEW"
)


# ============================================================================
# 6. EXPLICIT REVIEW OF THE FOUR KNOWN VARIANT CODES
# ============================================================================

known_variant_codes = [
    "5880005",
    "171207006",
    "399208008",
    "90226004"
]

variant_review = (
    df[
        df["CODE"].astype(str).isin(known_variant_codes)
    ]
    .groupby("CODE")
    .agg(
        Original_Descriptions=(
            "ORIGINAL_DESCRIPTION",
            lambda x: sorted(
                pd.Series(x.dropna())
                .astype(str)
                .unique()
                .tolist()
            )
        ),
        Canonical_Descriptions=(
            "CANONICAL_DESCRIPTION",
            lambda x: sorted(
                pd.Series(x.dropna())
                .astype(str)
                .unique()
                .tolist()
            )
        ),
        Procedure_Count=("CODE", "size")
    )
    .reset_index()
)

print("\nKNOWN DESCRIPTION-VARIANT CODES")
print("-" * 70)

if len(variant_review) > 0:
    for _, row in variant_review.iterrows():

        print(f"\nCODE: {row['CODE']}")
        print(
            f"Procedure records: "
            f"{row['Procedure_Count']:,}"
        )

        print("Original descriptions:")

        for desc in row["Original_Descriptions"]:
            print(f"  - {desc}")

        print("Canonical descriptions:")

        for desc in row["Canonical_Descriptions"]:
            print(f"  - {desc}")


# ============================================================================
# 7. BASE_COST = 431 SEGMENTATION
# ============================================================================

df["IS_431_COST"] = (
    pd.to_numeric(
        df["BASE_COST"],
        errors="coerce"
    ) == 431
)

df["IS_NON_431_COST"] = ~df["IS_431_COST"]

print("\nBASE_COST = 431 SEGMENTATION")
print("-" * 70)

count_431 = int(df["IS_431_COST"].sum())
count_non_431 = int(df["IS_NON_431_COST"].sum())

print(f"431 records: {count_431:,}")
print(f"Non-431 records: {count_non_431:,}")
print(
    f"431 record share: "
    f"{count_431 / len(df) * 100:.2f}%"
)


# ============================================================================
# 8. COST AGGREGATION
# ============================================================================

total_cost = df["BASE_COST"].sum()

cost_431 = df.loc[
    df["IS_431_COST"],
    "BASE_COST"
].sum()

cost_non_431 = df.loc[
    df["IS_NON_431_COST"],
    "BASE_COST"
].sum()

print("\nCOST SEGMENTATION")
print("-" * 70)
print(f"Total procedure cost: {total_cost:,.2f}")
print(f"431 cost: {cost_431:,.2f}")
print(f"Non-431 cost: {cost_non_431:,.2f}")
print(
    f"431 cost share: "
    f"{cost_431 / total_cost * 100:.2f}%"
)
print(
    f"Non-431 cost share: "
    f"{cost_non_431 / total_cost * 100:.2f}%"
)


# ============================================================================
# 9. PROCEDURE-LEVEL CLINICAL / COST SEGMENTATION
# ============================================================================

procedure_segmentation = (
    df.groupby(
        [
            "CODE",
            "CANONICAL_DESCRIPTION"
        ],
        dropna=False
    )
    .agg(
        Procedure_Count=("CODE", "size"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Total_Cost=("BASE_COST", "sum"),
        Average_Cost=("BASE_COST", "mean"),
        Median_Cost=("BASE_COST", "median"),
        Cost_431_Count=("IS_431_COST", "sum"),
        Non_431_Count=("IS_NON_431_COST", "sum"),
        Non_431_Cost=(
            "BASE_COST",
            lambda x: x[x != 431].sum()
        )
    )
    .reset_index()
)

procedure_segmentation[
    "Cost_431_Share"
] = (
    procedure_segmentation["Cost_431_Count"]
    / procedure_segmentation["Procedure_Count"]
    * 100
)

procedure_segmentation[
    "Procedures_Per_Patient"
] = (
    procedure_segmentation["Procedure_Count"]
    / procedure_segmentation["Unique_Patients"]
)

procedure_segmentation[
    "Procedures_Per_Encounter"
] = (
    procedure_segmentation["Procedure_Count"]
    / procedure_segmentation["Unique_Encounters"]
)

procedure_segmentation[
    "Cost_Per_Patient"
] = (
    procedure_segmentation["Total_Cost"]
    / procedure_segmentation["Unique_Patients"]
)


# ============================================================================
# 10. COST INTENSITY SEGMENTS
# ============================================================================

non_431_cost_values = (
    df.loc[
        df["IS_NON_431_COST"],
        "BASE_COST"
    ]
)

if len(non_431_cost_values) > 0:

    p50_non431 = non_431_cost_values.quantile(0.50)
    p75_non431 = non_431_cost_values.quantile(0.75)
    p95_non431 = non_431_cost_values.quantile(0.95)
    p99_non431 = non_431_cost_values.quantile(0.99)

else:

    p50_non431 = 0
    p75_non431 = 0
    p95_non431 = 0
    p99_non431 = 0


def assign_cost_intensity(row):

    if row["Cost_431_Share"] == 100:
        return "431_DOMINATED"

    if row["Non_431_Cost"] == 0:
        return "431_DOMINATED"

    avg_cost = row["Average_Cost"]

    if avg_cost >= p99_non431:
        return "EXTREME_NON_431_COST"

    if avg_cost >= p95_non431:
        return "HIGH_NON_431_COST"

    if avg_cost >= p75_non431:
        return "MODERATE_NON_431_COST"

    return "STANDARD_NON_431_COST"


procedure_segmentation[
    "Cost_Intensity"
] = procedure_segmentation.apply(
    assign_cost_intensity,
    axis=1
)


# ============================================================================
# 11. UTILIZATION INTENSITY SEGMENTS
# ============================================================================

procedure_volume_q75 = (
    procedure_segmentation[
        "Procedure_Count"
    ].quantile(0.75)
)

procedure_volume_q95 = (
    procedure_segmentation[
        "Procedure_Count"
    ].quantile(0.95)
)

repeat_q75 = (
    procedure_segmentation[
        "Procedures_Per_Patient"
    ].quantile(0.75)
)

repeat_q95 = (
    procedure_segmentation[
        "Procedures_Per_Patient"
    ].quantile(0.95)
)


def assign_utilization_intensity(row):

    volume = row["Procedure_Count"]
    repeat_rate = row["Procedures_Per_Patient"]

    if (
        volume >= procedure_volume_q95
        or repeat_rate >= repeat_q95
    ):
        return "EXTREME_UTILIZATION"

    if (
        volume >= procedure_volume_q75
        or repeat_rate >= repeat_q75
    ):
        return "HIGH_UTILIZATION"

    return "STANDARD_UTILIZATION"


procedure_segmentation[
    "Utilization_Intensity"
] = procedure_segmentation.apply(
    assign_utilization_intensity,
    axis=1
)


# ============================================================================
# 12. PRIORITY SEGMENT
# ============================================================================

def assign_priority(row):

    cost = row["Cost_Intensity"]
    utilization = row["Utilization_Intensity"]

    if cost in [
        "EXTREME_NON_431_COST",
        "HIGH_NON_431_COST"
    ] and utilization in [
        "EXTREME_UTILIZATION",
        "HIGH_UTILIZATION"
    ]:
        return "PRIORITY 1 — HIGH COST / HIGH UTILIZATION"

    if cost in [
        "EXTREME_NON_431_COST",
        "HIGH_NON_431_COST"
    ]:
        return "PRIORITY 2 — HIGH COST / LOWER UTILIZATION"

    if utilization in [
        "EXTREME_UTILIZATION",
        "HIGH_UTILIZATION"
    ]:
        return "PRIORITY 3 — HIGH UTILIZATION / LOWER COST"

    if row["Cost_431_Share"] == 100:
        return "PRIORITY 4 — 431 COST DOMINATED"

    return "STANDARD"


procedure_segmentation[
    "Priority_Profile"
] = procedure_segmentation.apply(
    assign_priority,
    axis=1
)


# ============================================================================
# 13. HIGH COST / HIGH UTILIZATION
# ============================================================================

high_cost_high_utilization = (
    procedure_segmentation[
        procedure_segmentation[
            "Priority_Profile"
        ]
        == "PRIORITY 1 — HIGH COST / HIGH UTILIZATION"
    ]
    .sort_values(
        "Non_431_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 14. HIGH COST / LOWER UTILIZATION
# ============================================================================

high_cost_low_utilization = (
    procedure_segmentation[
        procedure_segmentation[
            "Priority_Profile"
        ]
        == "PRIORITY 2 — HIGH COST / LOWER UTILIZATION"
    ]
    .sort_values(
        "Non_431_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 15. HIGH UTILIZATION / LOWER COST
# ============================================================================

high_utilization_low_cost = (
    procedure_segmentation[
        procedure_segmentation[
            "Priority_Profile"
        ]
        == "PRIORITY 3 — HIGH UTILIZATION / LOWER COST"
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 16. 431-DOMINATED PROCEDURES
# ============================================================================

procedures_431_dominated = (
    procedure_segmentation[
        procedure_segmentation[
            "Cost_431_Share"
        ] == 100
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 17. HIGH REPEAT UTILIZATION
# ============================================================================

high_repeat_utilization = (
    procedure_segmentation[
        procedure_segmentation[
            "Procedures_Per_Patient"
        ] >= repeat_q95
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 18. PROCEDURE COST CONCENTRATION
# ============================================================================

procedure_cost_rank = (
    procedure_segmentation
    .sort_values(
        "Non_431_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)

total_non_431_procedure_cost = (
    procedure_cost_rank["Non_431_Cost"].sum()
)

if total_non_431_procedure_cost > 0:

    top10_non431_share = (
        procedure_cost_rank.head(10)[
            "Non_431_Cost"
        ].sum()
        / total_non_431_procedure_cost
        * 100
    )

    top20_non431_share = (
        procedure_cost_rank.head(20)[
            "Non_431_Cost"
        ].sum()
        / total_non_431_procedure_cost
        * 100
    )

else:

    top10_non431_share = 0
    top20_non431_share = 0


# ============================================================================
# 19. UTILIZATION CONCENTRATION
# ============================================================================

procedure_volume_rank = (
    procedure_segmentation
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)

total_volume = (
    procedure_volume_rank[
        "Procedure_Count"
    ].sum()
)

top10_volume_share = (
    procedure_volume_rank.head(10)[
        "Procedure_Count"
    ].sum()
    / total_volume
    * 100
)

top20_volume_share = (
    procedure_volume_rank.head(20)[
        "Procedure_Count"
    ].sum()
    / total_volume
    * 100
)


# ============================================================================
# 20. PRIORITY DISTRIBUTION
# ============================================================================

priority_distribution = (
    procedure_segmentation[
        "Priority_Profile"
    ]
    .value_counts()
    .rename_axis("Priority_Profile")
    .reset_index(
        name="Procedure_Group_Count"
    )
)

priority_distribution[
    "Group_Share_%"
] = (
    priority_distribution[
        "Procedure_Group_Count"
    ]
    / len(procedure_segmentation)
    * 100
)


# ============================================================================
# 21. COST INTENSITY DISTRIBUTION
# ============================================================================

cost_intensity_distribution = (
    procedure_segmentation[
        "Cost_Intensity"
    ]
    .value_counts()
    .rename_axis("Cost_Intensity")
    .reset_index(
        name="Procedure_Group_Count"
    )
)


# ============================================================================
# 22. UTILIZATION INTENSITY DISTRIBUTION
# ============================================================================

utilization_distribution = (
    procedure_segmentation[
        "Utilization_Intensity"
    ]
    .value_counts()
    .rename_axis("Utilization_Intensity")
    .reset_index(
        name="Procedure_Group_Count"
    )
)


# ============================================================================
# 23. KEY METRICS
# ============================================================================

procedure_18f_metrics = pd.DataFrame({
    "Metric": [
        "Total procedure records",
        "Unique procedure codes",
        "Unique original descriptions",
        "Codes with original description variance",
        "Codes with canonical description variance",
        "Records with BASE_COST = 431",
        "431 record share (%)",
        "Total procedure cost",
        "431 procedure cost",
        "Non-431 procedure cost",
        "431 cost share (%)",
        "Non-431 cost share (%)",
        "High cost / high utilization groups",
        "High cost / lower utilization groups",
        "High utilization / lower cost groups",
        "431 dominated groups",
        "Top 10 non-431 procedure cost share (%)",
        "Top 20 non-431 procedure cost share (%)",
        "Top 10 procedure volume share (%)",
        "Top 20 procedure volume share (%)",
        "High repeat utilization groups"
    ],
    "Value": [
        len(df),
        df["CODE"].nunique(),
        df["ORIGINAL_DESCRIPTION"].nunique(),
        len(code_variance),
        len(canonical_variance),
        count_431,
        count_431 / len(df) * 100,
        total_cost,
        cost_431,
        cost_non_431,
        cost_431 / total_cost * 100,
        cost_non_431 / total_cost * 100,
        len(high_cost_high_utilization),
        len(high_cost_low_utilization),
        len(high_utilization_low_cost),
        len(procedures_431_dominated),
        top10_non431_share,
        top20_non431_share,
        top10_volume_share,
        top20_volume_share,
        len(high_repeat_utilization)
    ]
})


# ============================================================================
# 24. DISPLAY RESULTS
# ============================================================================

print("\nPROCEDURE PRIORITY DISTRIBUTION")
print("-" * 70)
print(
    priority_distribution.to_string(index=False)
)


print("\nCOST INTENSITY DISTRIBUTION")
print("-" * 70)
print(
    cost_intensity_distribution.to_string(index=False)
)


print("\nUTILIZATION INTENSITY DISTRIBUTION")
print("-" * 70)
print(
    utilization_distribution.to_string(index=False)
)


print("\nHIGH COST / HIGH UTILIZATION PROCEDURES")
print("-" * 70)

if len(high_cost_high_utilization) > 0:

    print(
        high_cost_high_utilization[
            [
                "CODE",
                "CANONICAL_DESCRIPTION",
                "Procedure_Count",
                "Non_431_Cost",
                "Unique_Patients",
                "Cost_Per_Patient",
                "Procedures_Per_Patient",
                "Cost_431_Share"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )

else:
    print("None identified.")


print("\nHIGH COST / LOWER UTILIZATION PROCEDURES")
print("-" * 70)

if len(high_cost_low_utilization) > 0:

    print(
        high_cost_low_utilization[
            [
                "CODE",
                "CANONICAL_DESCRIPTION",
                "Procedure_Count",
                "Non_431_Cost",
                "Average_Cost",
                "Cost_Per_Patient",
                "Cost_431_Share"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )

else:
    print("None identified.")


print("\nHIGH UTILIZATION / LOWER COST PROCEDURES")
print("-" * 70)

if len(high_utilization_low_cost) > 0:

    print(
        high_utilization_low_cost[
            [
                "CODE",
                "CANONICAL_DESCRIPTION",
                "Procedure_Count",
                "Unique_Patients",
                "Procedures_Per_Patient",
                "Total_Cost",
                "Cost_431_Share"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )

else:
    print("None identified.")


print("\n431-DOMINATED PROCEDURES")
print("-" * 70)

print(
    procedures_431_dominated[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Cost",
            "Cost_431_Share"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


print("\nHIGH REPEAT UTILIZATION")
print("-" * 70)

if len(high_repeat_utilization) > 0:

    print(
        high_repeat_utilization[
            [
                "CODE",
                "CANONICAL_DESCRIPTION",
                "Procedure_Count",
                "Unique_Patients",
                "Procedures_Per_Patient",
                "Unique_Encounters"
            ]
        ]
        .head(20)
        .to_string(index=False)
    )

else:
    print("None identified.")


# ============================================================================
# 25. FINAL VALIDATION
# ============================================================================

validation = {
    "Record count preserved":
        len(df) == source_record_count,

    "Original CODE preserved":
        df["ORIGINAL_CODE"].notna().all(),

    "Original DESCRIPTION preserved":
        df["ORIGINAL_DESCRIPTION"].notna().all(),

    "Canonical description available":
        canonical_quality_pass,

    "431 flag available":
        "IS_431_COST" in df.columns,

    "Non-431 flag available":
        "IS_NON_431_COST" in df.columns,

    "Procedure segmentation created":
        len(procedure_segmentation) > 0,

    "Priority profile created":
        "Priority_Profile" in procedure_segmentation.columns,

    "No records deleted":
        len(df) == source_record_count
}

validation_table = pd.DataFrame({
    "Validation": list(validation.keys()),
    "Result": list(validation.values())
})

print("\nFINAL VALIDATION")
print("-" * 70)
print(
    validation_table.to_string(index=False)
)


# ============================================================================
# 26. CREATE ANALYTICAL DATASETS
# ============================================================================

procedure_f = procedure_segmentation.copy()

procedure_f_priority = procedure_segmentation[
    [
        "CODE",
        "CANONICAL_DESCRIPTION",
        "Procedure_Count",
        "Unique_Patients",
        "Unique_Encounters",
        "Total_Cost",
        "Non_431_Cost",
        "Average_Cost",
        "Median_Cost",
        "Cost_Per_Patient",
        "Procedures_Per_Patient",
        "Procedures_Per_Encounter",
        "Cost_431_Share",
        "Cost_Intensity",
        "Utilization_Intensity",
        "Priority_Profile"
    ]
].copy()


# ============================================================================
# 27. EXPORT / NOTEBOOK OBJECTS
# ============================================================================

procedure_standardization_validation = code_description_summary.copy()

procedure_cost_segmentation = procedure_segmentation.copy()

procedure_priority_matrix = procedure_f_priority.copy()

procedure_f_summary = procedure_18f_metrics.copy()


# ============================================================================
# 28. COMPLETION SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18F COMPLETE")
print("=" * 70)

print(f"Total procedure records: {len(df):,}")
print(f"Unique procedure codes: {df['CODE'].nunique():,}")
print(
    f"Unique original descriptions: "
    f"{df['ORIGINAL_DESCRIPTION'].nunique():,}"
)

print(
    f"Codes with original description variance: "
    f"{len(code_variance):,}"
)

print(
    f"Codes with canonical description variance: "
    f"{len(canonical_variance):,}"
)

print(
    f"BASE_COST = 431 records: "
    f"{count_431:,}"
)

print(
    f"431 record share: "
    f"{count_431 / len(df) * 100:.2f}%"
)

print(
    f"Non-431 procedure cost: "
    f"{cost_non_431:,.2f}"
)

print(
    f"Top 10 non-431 cost concentration: "
    f"{top10_non431_share:.2f}%"
)

print(
    f"Top 20 non-431 cost concentration: "
    f"{top20_non431_share:.2f}%"
)

print(
    f"Top 10 procedure volume concentration: "
    f"{top10_volume_share:.2f}%"
)

print(
    f"Top 20 procedure volume concentration: "
    f"{top20_volume_share:.2f}%"
)

print(
    f"Priority 1 groups: "
    f"{len(high_cost_high_utilization):,}"
)

print(
    f"Priority 2 groups: "
    f"{len(high_cost_low_utilization):,}"
)

print(
    f"Priority 3 groups: "
    f"{len(high_utilization_low_cost):,}"
)

print(
    f"431-dominated groups: "
    f"{len(procedures_431_dominated):,}"
)

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_f")
print("  - procedure_f_summary")
print("  - procedure_standardization_validation")
print("  - procedure_cost_segmentation")
print("  - procedure_priority_matrix")
print("  - high_cost_high_utilization")
print("  - high_cost_low_utilization")
print("  - high_utilization_low_cost")
print("  - procedures_431_dominated")
print("  - high_repeat_utilization")
print("  - priority_distribution")
print("  - cost_intensity_distribution")
print("  - utilization_distribution")

all_pass = all(validation.values())

print("\nSTATUS:", "PASS" if all_pass else "REVIEW REQUIRED")

print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18G — PROCEDURE CODE-LEVEL CANONICALIZATION &
#              FINAL ANALYTICAL READINESS
# ============================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18G — PROCEDURE CODE-LEVEL CANONICALIZATION &")
print("             FINAL ANALYTICAL READINESS")
print("=" * 70)


# ============================================================================
# 1. FIND THE CORRECT ROW-LEVEL SOURCE DATASET
# ============================================================================

print("\nSOURCE DATA DISCOVERY")
print("-" * 70)

# procedure_analysis should contain the original row-level procedure records.
if "procedure_analysis" not in globals():
    raise NameError(
        "procedure_analysis is not available. "
        "Please run STEP 4.18D first."
    )

source_df = procedure_analysis.copy()

print("Using source dataset: procedure_analysis")
print(f"Source records: {len(source_df):,}")


# ============================================================================
# 2. CHECK REQUIRED COLUMNS
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "BASE_COST",
    "PATIENT",
    "ENCOUNTER"
]

missing_columns = [
    col for col in required_columns
    if col not in source_df.columns
]

if missing_columns:
    print("\nAVAILABLE COLUMNS")
    print("-" * 70)
    print(source_df.columns.tolist())

    raise ValueError(
        f"procedure_analysis is missing required columns: "
        f"{missing_columns}"
    )

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print(f"Procedure records available: {len(source_df):,}")


# ============================================================================
# 3. PRESERVE ORIGINAL VALUES
# ============================================================================

procedure_g = source_df.copy()

if "ORIGINAL_CODE" not in procedure_g.columns:
    procedure_g["ORIGINAL_CODE"] = procedure_g["CODE"].copy()

if "ORIGINAL_DESCRIPTION" not in procedure_g.columns:
    procedure_g["ORIGINAL_DESCRIPTION"] = (
        procedure_g["DESCRIPTION"].copy()
    )

print("\nORIGINAL VALUE PRESERVATION")
print("-" * 70)
print("ORIGINAL_CODE available: True")
print("ORIGINAL_DESCRIPTION available: True")


# ============================================================================
# 4. EXPLICIT CODE-LEVEL CANONICAL MAPPING
# ============================================================================

canonical_mapping = {
    "5880005": "Physical examination",
    "171207006": "Depression screening",
    "399208008": "Plain chest X-ray",
    "90226004": (
        "Cytopathology procedure - preparation of smear - genital source"
    )
}

print("\nEXPLICIT CANONICAL MAPPING")
print("-" * 70)

for code, description in canonical_mapping.items():
    print(f"{code} → {description}")


# ============================================================================
# 5. NORMALIZE CODE TYPE FOR MAPPING
# ============================================================================

procedure_g["_CODE_KEY"] = (
    procedure_g["CODE"]
    .astype("string")
    .str.strip()
)


# ============================================================================
# 6. BUILD FALLBACK CODE-LEVEL MAPPING
# ============================================================================
# For codes not explicitly defined above:
# use the most frequent original DESCRIPTION.
# ============================================================================

description_counts = (
    procedure_g
    .groupby(
        ["_CODE_KEY", "ORIGINAL_DESCRIPTION"],
        dropna=False
    )
    .size()
    .reset_index(name="Description_Count")
)

description_counts = description_counts.sort_values(
    [
        "_CODE_KEY",
        "Description_Count",
        "ORIGINAL_DESCRIPTION"
    ],
    ascending=[True, False, True]
)

fallback_mapping = (
    description_counts
    .drop_duplicates(
        subset=["_CODE_KEY"],
        keep="first"
    )
    .set_index("_CODE_KEY")["ORIGINAL_DESCRIPTION"]
    .to_dict()
)


# ============================================================================
# 7. APPLY EXPLICIT + FALLBACK CANONICALIZATION
# ============================================================================

explicit_mapping_normalized = {
    str(code): description
    for code, description in canonical_mapping.items()
}

full_canonical_mapping = fallback_mapping.copy()
full_canonical_mapping.update(
    explicit_mapping_normalized
)

procedure_g["CANONICAL_DESCRIPTION"] = (
    procedure_g["_CODE_KEY"]
    .map(full_canonical_mapping)
)


# ============================================================================
# 8. CANONICAL DESCRIPTION QUALITY
# ============================================================================

missing_canonical = (
    procedure_g["CANONICAL_DESCRIPTION"].isna().sum()
)

blank_canonical = (
    procedure_g["CANONICAL_DESCRIPTION"]
    .astype("string")
    .str.strip()
    .eq("")
    .sum()
)

print("\nCANONICAL DESCRIPTION QUALITY")
print("-" * 70)
print(f"Missing canonical descriptions: {missing_canonical:,}")
print(f"Blank canonical descriptions: {blank_canonical:,}")

if missing_canonical > 0 or blank_canonical > 0:
    raise ValueError(
        "Canonical description mapping is incomplete."
    )

print("STATUS: PASS")


# ============================================================================
# 9. VERIFY ONE CANONICAL DESCRIPTION PER CODE
# ============================================================================

code_canonical_validation = (
    procedure_g
    .groupby("_CODE_KEY")["CANONICAL_DESCRIPTION"]
    .nunique()
    .reset_index(
        name="Unique_Canonical_Descriptions"
    )
)

codes_with_multiple_canonical = (
    code_canonical_validation[
        code_canonical_validation[
            "Unique_Canonical_Descriptions"
        ] > 1
    ]
)

print("\nCODE → CANONICAL DESCRIPTION VALIDATION")
print("-" * 70)
print(
    f"Unique procedure codes: "
    f"{procedure_g['_CODE_KEY'].nunique():,}"
)

print(
    "Codes with multiple canonical descriptions: "
    f"{len(codes_with_multiple_canonical):,}"
)

if len(codes_with_multiple_canonical) > 0:
    print(codes_with_multiple_canonical)

    raise ValueError(
        "Canonicalization failed: one or more procedure "
        "codes still have multiple canonical descriptions."
    )

print(
    "Every CODE has exactly one canonical description: True"
)


# ============================================================================
# 10. VALIDATE THE FOUR KNOWN VARIANT CODES
# ============================================================================

print("\nKNOWN VARIANT CODE VALIDATION")
print("-" * 70)

for code, expected_description in canonical_mapping.items():

    subset = procedure_g[
        procedure_g["_CODE_KEY"] == str(code)
    ]

    actual_values = (
        subset["CANONICAL_DESCRIPTION"]
        .dropna()
        .unique()
        .tolist()
    )

    print(f"\nCODE: {code}")
    print(f"Records: {len(subset):,}")
    print(f"Expected: {expected_description}")
    print(f"Actual: {actual_values}")

    if actual_values != [expected_description]:
        raise ValueError(
            f"Canonical mapping failed for CODE {code}"
        )


# ============================================================================
# 11. RECORD COUNT PRESERVATION
# ============================================================================

record_count_preserved = (
    len(procedure_g) == len(source_df)
)

print("\nRECORD PRESERVATION")
print("-" * 70)
print(f"Source records: {len(source_df):,}")
print(f"Analytical records: {len(procedure_g):,}")
print(
    f"Record count preserved: {record_count_preserved}"
)

if not record_count_preserved:
    raise ValueError(
        "Record count changed during canonicalization."
    )


# ============================================================================
# 12. ORIGINAL VALUE PRESERVATION
# ============================================================================

original_code_preserved = (
    procedure_g["ORIGINAL_CODE"].astype("string").tolist()
    ==
    source_df["ORIGINAL_CODE"].astype("string").tolist()
    if "ORIGINAL_CODE" in source_df.columns
    else
    procedure_g["ORIGINAL_CODE"].astype("string").tolist()
    ==
    source_df["CODE"].astype("string").tolist()
)

original_description_preserved = (
    procedure_g["ORIGINAL_DESCRIPTION"].astype("string").tolist()
    ==
    source_df["ORIGINAL_DESCRIPTION"].astype("string").tolist()
    if "ORIGINAL_DESCRIPTION" in source_df.columns
    else
    procedure_g["ORIGINAL_DESCRIPTION"].astype("string").tolist()
    ==
    source_df["DESCRIPTION"].astype("string").tolist()
)

print("\nORIGINAL VALUES")
print("-" * 70)
print(
    f"Original CODE preserved: "
    f"{original_code_preserved}"
)
print(
    f"Original DESCRIPTION preserved: "
    f"{original_description_preserved}"
)

if not original_code_preserved:
    raise ValueError("Original CODE was not preserved.")

if not original_description_preserved:
    raise ValueError(
        "Original DESCRIPTION was not preserved."
    )


# ============================================================================
# 13. CANONICALIZATION IMPACT
# ============================================================================

procedure_g["DESCRIPTION_STANDARDIZED_FLAG"] = (
    procedure_g["ORIGINAL_DESCRIPTION"].astype("string")
    !=
    procedure_g["CANONICAL_DESCRIPTION"].astype("string")
)

standardized_records = (
    procedure_g["DESCRIPTION_STANDARDIZED_FLAG"].sum()
)

unchanged_records = (
    len(procedure_g) - standardized_records
)

print("\nCANONICALIZATION IMPACT")
print("-" * 70)
print(f"Records standardized: {standardized_records:,}")
print(f"Records unchanged: {unchanged_records:,}")
print(
    f"Standardized record share: "
    f"{standardized_records / len(procedure_g) * 100:.2f}%"
)


# ============================================================================
# 14. DESCRIPTION VARIANCE SUMMARY
# ============================================================================

description_variance_summary = (
    procedure_g
    .groupby("_CODE_KEY")
    .agg(
        Unique_Original_Descriptions=(
            "ORIGINAL_DESCRIPTION",
            "nunique"
        ),
        Unique_Canonical_Descriptions=(
            "CANONICAL_DESCRIPTION",
            "nunique"
        ),
        Procedure_Count=(
            "_CODE_KEY",
            "size"
        )
    )
    .reset_index()
    .rename(columns={"_CODE_KEY": "CODE"})
)

description_variance_summary = (
    description_variance_summary[
        description_variance_summary[
            "Unique_Original_Descriptions"
        ] > 1
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
)

print("\nDESCRIPTION VARIANCE SUMMARY")
print("-" * 70)
print(
    description_variance_summary.to_string(
        index=False
    )
)


# ============================================================================
# 15. BASE_COST = 431 VALIDATION
# ============================================================================

cost_431_count = (
    procedure_g["BASE_COST"].eq(431).sum()
)

non_431_count = (
    procedure_g["BASE_COST"].ne(431).sum()
)

total_cost = (
    procedure_g["BASE_COST"].sum()
)

cost_431 = (
    procedure_g.loc[
        procedure_g["BASE_COST"] == 431,
        "BASE_COST"
    ].sum()
)

non_431_cost = (
    procedure_g.loc[
        procedure_g["BASE_COST"] != 431,
        "BASE_COST"
    ].sum()
)

print("\nBASE_COST = 431 VALIDATION")
print("-" * 70)
print(f"431 records: {cost_431_count:,}")
print(f"Non-431 records: {non_431_count:,}")
print(
    f"431 record share: "
    f"{cost_431_count / len(procedure_g) * 100:.2f}%"
)
print(f"Total procedure cost: {total_cost:,.2f}")
print(f"431 cost: {cost_431:,.2f}")
print(f"Non-431 cost: {non_431_cost:,.2f}")
print(
    f"431 cost share: "
    f"{cost_431 / total_cost * 100:.2f}%"
)


# ============================================================================
# 16. PROCEDURE CODE SUMMARY
# ============================================================================

procedure_code_summary = (
    procedure_g
    .groupby(
        ["_CODE_KEY", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("_CODE_KEY", "size"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Total_Cost=("BASE_COST", "sum"),
        Average_Cost=("BASE_COST", "mean"),
        Median_Cost=("BASE_COST", "median"),
        Cost_431_Count=(
            "BASE_COST",
            lambda x: (x == 431).sum()
        ),
        Non_431_Count=(
            "BASE_COST",
            lambda x: (x != 431).sum()
        )
    )
    .reset_index()
    .rename(columns={"_CODE_KEY": "CODE"})
)

procedure_code_summary["Procedures_Per_Patient"] = (
    procedure_code_summary["Procedure_Count"]
    /
    procedure_code_summary["Unique_Patients"].replace(
        0, np.nan
    )
)

procedure_code_summary["Cost_Per_Patient"] = (
    procedure_code_summary["Total_Cost"]
    /
    procedure_code_summary["Unique_Patients"].replace(
        0, np.nan
    )
)

procedure_code_summary["Cost_431_Share"] = (
    procedure_code_summary["Cost_431_Count"]
    /
    procedure_code_summary["Procedure_Count"].replace(
        0, np.nan
    )
    * 100
)

procedure_code_summary["Non_431_Cost"] = (
    procedure_code_summary["Total_Cost"]
    -
    (
        procedure_code_summary["Cost_431_Count"]
        * 431
    )
)


# ============================================================================
# 17. CLEAN INTERNAL COLUMN
# ============================================================================

procedure_g = procedure_g.drop(
    columns=["_CODE_KEY"]
)


# ============================================================================
# 18. SUMMARY DATASET
# ============================================================================

procedure_g_summary = pd.DataFrame({
    "Metric": [
        "Total procedure records",
        "Unique procedure codes",
        "Unique original descriptions",
        "Unique canonical descriptions",
        "Codes with original description variance",
        "Codes with multiple canonical descriptions",
        "Records standardized",
        "Records with BASE_COST = 431",
        "Non-431 records",
        "Total procedure cost",
        "431 cost",
        "Non-431 cost",
        "431 record share (%)",
        "431 cost share (%)",
        "Missing patients",
        "Missing encounters"
    ],
    "Value": [
        len(procedure_g),
        procedure_g["CODE"].nunique(),
        procedure_g["ORIGINAL_DESCRIPTION"].nunique(),
        procedure_g["CANONICAL_DESCRIPTION"].nunique(),
        len(description_variance_summary),
        len(codes_with_multiple_canonical),
        standardized_records,
        cost_431_count,
        non_431_count,
        total_cost,
        cost_431,
        non_431_cost,
        cost_431_count / len(procedure_g) * 100,
        cost_431 / total_cost * 100,
        procedure_g["PATIENT"].isna().sum(),
        procedure_g["ENCOUNTER"].isna().sum()
    ]
})


# ============================================================================
# 19. FINAL VALIDATION
# ============================================================================

validation = pd.DataFrame({
    "Validation": [
        "Record count preserved",
        "Original CODE preserved",
        "Original DESCRIPTION preserved",
        "Canonical description available",
        "Exactly one canonical description per CODE",
        "431 flag available",
        "Non-431 cost available",
        "Procedure-level summary created",
        "No records deleted"
    ],
    "Result": [
        record_count_preserved,
        original_code_preserved,
        original_description_preserved,
        missing_canonical == 0 and blank_canonical == 0,
        len(codes_with_multiple_canonical) == 0,
        cost_431_count > 0,
        non_431_count > 0,
        len(procedure_code_summary) > 0,
        len(procedure_g) == len(source_df)
    ]
})

print("\nFINAL VALIDATION")
print("-" * 70)
print(validation.to_string(index=False))

if not validation["Result"].all():
    raise ValueError(
        "STEP 4.18G validation failed."
    )


# ============================================================================
# 20. EXPORT DATASETS TO GLOBAL KERNEL
# ============================================================================

canonical_mapping = pd.DataFrame(
    list(full_canonical_mapping.items()),
    columns=["CODE", "CANONICAL_DESCRIPTION"]
)

procedure_g_validation = validation.copy()


# ============================================================================
# 21. FINAL OUTPUT
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18G COMPLETE")
print("=" * 70)

print(
    f"Total procedure records: "
    f"{len(procedure_g):,}"
)

print(
    f"Unique procedure codes: "
    f"{procedure_g['CODE'].nunique():,}"
)

print(
    f"Unique original descriptions: "
    f"{procedure_g['ORIGINAL_DESCRIPTION'].nunique():,}"
)

print(
    f"Unique canonical descriptions: "
    f"{procedure_g['CANONICAL_DESCRIPTION'].nunique():,}"
)

print(
    "Codes with original description variance: "
    f"{len(description_variance_summary):,}"
)

print(
    "Codes with multiple canonical descriptions: "
    f"{len(codes_with_multiple_canonical):,}"
)

print(
    f"Records standardized: "
    f"{standardized_records:,}"
)

print(
    f"BASE_COST = 431 records: "
    f"{cost_431_count:,}"
)

print(
    f"431 record share: "
    f"{cost_431_count / len(procedure_g) * 100:.2f}%"
)

print(
    f"Total procedure cost: "
    f"{total_cost:,.2f}"
)

print(
    f"Non-431 procedure cost: "
    f"{non_431_cost:,.2f}"
)

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_g")
print("  - procedure_g_summary")
print("  - procedure_code_summary")
print("  - description_variance_summary")
print("  - canonical_mapping")
print("  - procedure_g_validation")

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18H — PROCEDURE EXECUTIVE ANALYTICS, COST CONCENTRATION &
#              UTILIZATION PRIORITIZATION
# ============================================================================
#
# SOURCE:
#   procedure_g
#
# PURPOSE:
#   Build the final procedure-level analytical and executive insight layer
#   after CODE → canonical description standardization.
#
# DESIGN PRINCIPLES:
#   - Preserve all 47,701 procedure records.
#   - Use procedure_g as the authoritative analytical source.
#   - Do not overwrite ORIGINAL_CODE or ORIGINAL_DESCRIPTION.
#   - Do not re-standardize canonical descriptions.
#   - Separate BASE_COST = 431 from economically meaningful non-431 cost.
#   - Produce cost, utilization, patient-burden and priority analytics.
#   - Create reusable datasets for downstream dashboards/reporting.
#
# ============================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18H — PROCEDURE EXECUTIVE ANALYTICS, COST CONCENTRATION &")
print("             UTILIZATION PRIORITIZATION")
print("=" * 70)


# ============================================================================
# 1. SOURCE DATA DISCOVERY
# ============================================================================

print("\nSOURCE DATA DISCOVERY")
print("-" * 70)

if "procedure_g" not in globals():
    raise NameError(
        "procedure_g is not available. "
        "Please run STEP 4.18G successfully before running STEP 4.18H."
    )

source_df = procedure_g.copy()

print("Using source dataset: procedure_g")
print(f"Source records: {len(source_df):,}")


# ============================================================================
# 2. REQUIRED COLUMN VALIDATION
# ============================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "ORIGINAL_CODE",
    "ORIGINAL_DESCRIPTION",
    "CANONICAL_DESCRIPTION",
    "BASE_COST",
    "PATIENT",
    "ENCOUNTER"
]

missing_columns = [
    col for col in required_columns
    if col not in source_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns in procedure_g: {missing_columns}"
    )

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print(f"Procedure records available: {len(source_df):,}")


# ============================================================================
# 3. CREATE ANALYTICAL COPY
# ============================================================================

procedure_h = source_df.copy()


# ============================================================================
# 4. BASIC DATA VALIDATION
# ============================================================================

print("\nANALYTICAL DATASET VALIDATION")
print("-" * 70)

if len(procedure_h) != 47701:
    print(
        f"WARNING: Expected 47,701 records but found "
        f"{len(procedure_h):,}."
    )

unique_codes = procedure_h["CODE"].nunique()
unique_canonical = procedure_h["CANONICAL_DESCRIPTION"].nunique()

print(f"Records: {len(procedure_h):,}")
print(f"Unique procedure codes: {unique_codes:,}")
print(f"Unique canonical descriptions: {unique_canonical:,}")


# ============================================================================
# 5. CANONICALIZATION INTEGRITY CHECK
# ============================================================================

canonical_check = (
    procedure_h
    .groupby("CODE")["CANONICAL_DESCRIPTION"]
    .nunique()
)

multi_canonical_codes = (
    canonical_check[canonical_check > 1]
)

missing_canonical = (
    procedure_h["CANONICAL_DESCRIPTION"].isna().sum()
)

blank_canonical = (
    procedure_h["CANONICAL_DESCRIPTION"]
    .astype("string")
    .str.strip()
    .eq("")
    .sum()
)

print("\nCANONICALIZATION INTEGRITY")
print("-" * 70)
print(f"Missing canonical descriptions: {missing_canonical:,}")
print(f"Blank canonical descriptions: {blank_canonical:,}")
print(
    "Codes with multiple canonical descriptions: "
    f"{len(multi_canonical_codes):,}"
)

if missing_canonical > 0:
    raise ValueError(
        "Missing canonical descriptions detected."
    )

if blank_canonical > 0:
    raise ValueError(
        "Blank canonical descriptions detected."
    )

if len(multi_canonical_codes) > 0:
    raise ValueError(
        "Some procedure codes still have multiple canonical descriptions."
    )

print("Canonicalization integrity: PASS")


# ============================================================================
# 6. BASE COST SEGMENTATION
# ============================================================================

procedure_h["IS_431_COST"] = (
    procedure_h["BASE_COST"] == 431
)

procedure_h["IS_NON_431_COST"] = (
    procedure_h["BASE_COST"] != 431
)

procedure_h["NON_431_COST"] = np.where(
    procedure_h["IS_NON_431_COST"],
    procedure_h["BASE_COST"],
    0
)

procedure_h["COST_431_COMPONENT"] = np.where(
    procedure_h["IS_431_COST"],
    procedure_h["BASE_COST"],
    0
)


# ============================================================================
# 7. TOTAL COST METRICS
# ============================================================================

total_records = len(procedure_h)

total_cost = procedure_h["BASE_COST"].sum()

cost_431 = procedure_h["COST_431_COMPONENT"].sum()

non_431_cost = procedure_h["NON_431_COST"].sum()

records_431 = procedure_h["IS_431_COST"].sum()

records_non_431 = procedure_h["IS_NON_431_COST"].sum()

record_431_share = (
    records_431 / total_records * 100
)

cost_431_share = (
    cost_431 / total_cost * 100
)

non_431_cost_share = (
    non_431_cost / total_cost * 100
)

print("\nBASE_COST SEGMENTATION")
print("-" * 70)
print(f"431 records: {records_431:,}")
print(f"Non-431 records: {records_non_431:,}")
print(f"431 record share: {record_431_share:.2f}%")
print(f"Total procedure cost: {total_cost:,.2f}")
print(f"431 cost: {cost_431:,.2f}")
print(f"Non-431 cost: {non_431_cost:,.2f}")
print(f"431 cost share: {cost_431_share:.2f}%")
print(f"Non-431 cost share: {non_431_cost_share:.2f}%")


# ============================================================================
# 8. PROCEDURE-LEVEL EXECUTIVE SUMMARY
# ============================================================================

procedure_summary = (
    procedure_h
    .groupby(
        ["CODE", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("CODE", "size"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),

        Total_Procedure_Cost=("BASE_COST", "sum"),

        Non_431_Count=("IS_NON_431_COST", "sum"),

        Non_431_Total_Cost=("NON_431_COST", "sum"),

        Average_Procedure_Cost=("BASE_COST", "mean"),

        Median_Procedure_Cost=("BASE_COST", "median"),

        Cost_431_Count=("IS_431_COST", "sum"),

        Maximum_Procedure_Cost=("BASE_COST", "max")
    )
    .reset_index()
)

procedure_summary["Procedures_Per_Patient"] = (
    procedure_summary["Procedure_Count"]
    /
    procedure_summary["Unique_Patients"].replace(0, np.nan)
)

procedure_summary["Procedures_Per_Encounter"] = (
    procedure_summary["Procedure_Count"]
    /
    procedure_summary["Unique_Encounters"].replace(0, np.nan)
)

procedure_summary["Cost_Per_Patient"] = (
    procedure_summary["Total_Procedure_Cost"]
    /
    procedure_summary["Unique_Patients"].replace(0, np.nan)
)

procedure_summary["Non_431_Cost_Per_Patient"] = (
    procedure_summary["Non_431_Total_Cost"]
    /
    procedure_summary["Unique_Patients"].replace(0, np.nan)
)

procedure_summary["Cost_431_Share"] = (
    procedure_summary["Cost_431_Count"]
    /
    procedure_summary["Procedure_Count"].replace(0, np.nan)
    * 100
)

procedure_summary["Non_431_Record_Share"] = (
    procedure_summary["Non_431_Count"]
    /
    procedure_summary["Procedure_Count"].replace(0, np.nan)
    * 100
)


# ============================================================================
# 9. COST RANKING
# ============================================================================

procedure_summary["Cost_Rank"] = (
    procedure_summary["Total_Procedure_Cost"]
    .rank(
        method="min",
        ascending=False
    )
)

procedure_summary["Non_431_Cost_Rank"] = (
    procedure_summary["Non_431_Total_Cost"]
    .rank(
        method="min",
        ascending=False
    )
)

procedure_summary["Utilization_Rank"] = (
    procedure_summary["Procedure_Count"]
    .rank(
        method="min",
        ascending=False
    )
)

procedure_summary["Cost_Per_Patient_Rank"] = (
    procedure_summary["Cost_Per_Patient"]
    .rank(
        method="min",
        ascending=False
    )
)


# ============================================================================
# 10. TOP PROCEDURES BY TOTAL COST
# ============================================================================

top_cost_procedures = (
    procedure_summary
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY TOTAL COST")
print("-" * 70)

print(
    top_cost_procedures[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Total_Procedure_Cost",
            "Unique_Patients",
            "Cost_Per_Patient",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 11. TOP PROCEDURES BY NON-431 COST
# ============================================================================

top_non_431_cost = (
    procedure_summary
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY NON-431 COST")
print("-" * 70)

print(
    top_non_431_cost[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Count",
            "Non_431_Total_Cost",
            "Average_Procedure_Cost",
            "Unique_Patients",
            "Cost_Per_Patient",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 12. TOP PROCEDURES BY UTILIZATION
# ============================================================================

top_utilization = (
    procedure_summary
    .sort_values(
        [
            "Procedure_Count",
            "Unique_Patients"
        ],
        ascending=[False, False]
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY UTILIZATION")
print("-" * 70)

print(
    top_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Procedures_Per_Patient",
            "Procedures_Per_Encounter",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 13. TOP PROCEDURES BY COST PER PATIENT
# ============================================================================

top_cost_per_patient = (
    procedure_summary[
        procedure_summary["Unique_Patients"] > 0
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY COST PER PATIENT")
print("-" * 70)

print(
    top_cost_per_patient[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 14. TOP REPEAT-UTILIZATION PROCEDURES
# ============================================================================

high_repeat_utilization = (
    procedure_summary[
        procedure_summary["Unique_Patients"] > 0
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .head(20)
    .copy()
)

print("\nTOP 20 PROCEDURES BY PROCEDURES PER PATIENT")
print("-" * 70)

print(
    high_repeat_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Procedures_Per_Patient",
            "Unique_Encounters",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].to_string(index=False)
)


# ============================================================================
# 15. COST CONCENTRATION
# ============================================================================

cost_ranked = (
    procedure_summary
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)

non_431_cost_ranked = (
    procedure_summary
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)

utilization_ranked = (
    procedure_summary
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


def concentration_share(df, column, n, denominator):
    if denominator == 0:
        return 0.0

    return (
        df[column]
        .head(n)
        .sum()
        /
        denominator
        * 100
    )


top_5_cost_share = concentration_share(
    cost_ranked,
    "Total_Procedure_Cost",
    5,
    total_cost
)

top_10_cost_share = concentration_share(
    cost_ranked,
    "Total_Procedure_Cost",
    10,
    total_cost
)

top_20_cost_share = concentration_share(
    cost_ranked,
    "Total_Procedure_Cost",
    20,
    total_cost
)

top_5_non_431_share = concentration_share(
    non_431_cost_ranked,
    "Non_431_Total_Cost",
    5,
    non_431_cost
)

top_10_non_431_share = concentration_share(
    non_431_cost_ranked,
    "Non_431_Total_Cost",
    10,
    non_431_cost
)

top_20_non_431_share = concentration_share(
    non_431_cost_ranked,
    "Non_431_Total_Cost",
    20,
    non_431_cost
)


# ============================================================================
# 16. UTILIZATION CONCENTRATION
# ============================================================================

top_5_volume_share = (
    utilization_ranked["Procedure_Count"]
    .head(5)
    .sum()
    /
    total_records
    * 100
)

top_10_volume_share = (
    utilization_ranked["Procedure_Count"]
    .head(10)
    .sum()
    /
    total_records
    * 100
)

top_20_volume_share = (
    utilization_ranked["Procedure_Count"]
    .head(20)
    .sum()
    /
    total_records
    * 100
)

print("\nPROCEDURE COST CONCENTRATION")
print("-" * 70)
print(f"Top 5 total-cost share: {top_5_cost_share:.2f}%")
print(f"Top 10 total-cost share: {top_10_cost_share:.2f}%")
print(f"Top 20 total-cost share: {top_20_cost_share:.2f}%")
print(f"Top 5 non-431 cost share: {top_5_non_431_share:.2f}%")
print(f"Top 10 non-431 cost share: {top_10_non_431_share:.2f}%")
print(f"Top 20 non-431 cost share: {top_20_non_431_share:.2f}%")

print("\nPROCEDURE UTILIZATION CONCENTRATION")
print("-" * 70)
print(f"Top 5 volume share: {top_5_volume_share:.2f}%")
print(f"Top 10 volume share: {top_10_volume_share:.2f}%")
print(f"Top 20 volume share: {top_20_volume_share:.2f}%")


# ============================================================================
# 17. ANALYTICAL THRESHOLDS
# ============================================================================
#
# Thresholds are derived from the procedure-level distribution rather than
# arbitrary absolute values.
# ============================================================================

cost_threshold = (
    procedure_summary["Non_431_Total_Cost"]
    .quantile(0.75)
)

utilization_threshold = (
    procedure_summary["Procedure_Count"]
    .quantile(0.75)
)

cost_per_patient_threshold = (
    procedure_summary["Cost_Per_Patient"]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
    .quantile(0.75)
)

repeat_utilization_threshold = (
    procedure_summary["Procedures_Per_Patient"]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
    .quantile(0.75)
)

print("\nANALYTICAL THRESHOLDS")
print("-" * 70)
print(
    f"High non-431 cost threshold: "
    f"{cost_threshold:,.2f}"
)
print(
    f"High utilization threshold: "
    f"{utilization_threshold:,.2f}"
)
print(
    f"High cost-per-patient threshold: "
    f"{cost_per_patient_threshold:,.2f}"
)
print(
    f"High repeat-utilization threshold: "
    f"{repeat_utilization_threshold:,.2f}"
)


# ============================================================================
# 18. COST / UTILIZATION SEGMENTATION
# ============================================================================

def assign_priority(row):

    high_cost = (
        row["Non_431_Total_Cost"]
        >= cost_threshold
    )

    high_utilization = (
        row["Procedure_Count"]
        >= utilization_threshold
    )

    cost_431_dominated = (
        row["Cost_431_Share"] >= 75
    )

    if cost_431_dominated:
        return "PRIORITY 4 — 431 COST DOMINATED"

    if high_cost and high_utilization:
        return "PRIORITY 1 — HIGH COST / HIGH UTILIZATION"

    if high_cost and not high_utilization:
        return "PRIORITY 2 — HIGH COST / LOWER UTILIZATION"

    if not high_cost and high_utilization:
        return "PRIORITY 3 — HIGH UTILIZATION / LOWER COST"

    return "STANDARD"


procedure_summary["Priority_Profile"] = (
    procedure_summary.apply(
        assign_priority,
        axis=1
    )
)


# ============================================================================
# 19. COST INTENSITY
# ============================================================================

def assign_cost_intensity(row):

    if row["Cost_431_Share"] >= 75:
        return "431_DOMINATED"

    if row["Non_431_Total_Cost"] >= (
        procedure_summary["Non_431_Total_Cost"]
        .quantile(0.99)
    ):
        return "EXTREME_NON_431_COST"

    if row["Non_431_Total_Cost"] >= (
        procedure_summary["Non_431_Total_Cost"]
        .quantile(0.90)
    ):
        return "HIGH_NON_431_COST"

    if row["Non_431_Total_Cost"] >= (
        procedure_summary["Non_431_Total_Cost"]
        .quantile(0.75)
    ):
        return "MODERATE_NON_431_COST"

    return "STANDARD_NON_431_COST"


procedure_summary["Cost_Intensity"] = (
    procedure_summary.apply(
        assign_cost_intensity,
        axis=1
    )
)


# ============================================================================
# 20. UTILIZATION INTENSITY
# ============================================================================

utilization_q75 = (
    procedure_summary["Procedure_Count"]
    .quantile(0.75)
)

utilization_q90 = (
    procedure_summary["Procedure_Count"]
    .quantile(0.90)
)

utilization_q99 = (
    procedure_summary["Procedure_Count"]
    .quantile(0.99)
)


def assign_utilization_intensity(row):

    value = row["Procedure_Count"]

    if value >= utilization_q99:
        return "EXTREME_UTILIZATION"

    if value >= utilization_q90:
        return "HIGH_UTILIZATION"

    if value >= utilization_q75:
        return "MODERATE_UTILIZATION"

    return "STANDARD_UTILIZATION"


procedure_summary["Utilization_Intensity"] = (
    procedure_summary.apply(
        assign_utilization_intensity,
        axis=1
    )
)


# ============================================================================
# 21. HIGH COST / HIGH UTILIZATION
# ============================================================================

high_cost_high_utilization = (
    procedure_summary[
        (
            procedure_summary["Non_431_Total_Cost"]
            >= cost_threshold
        )
        &
        (
            procedure_summary["Procedure_Count"]
            >= utilization_threshold
        )
        &
        (
            procedure_summary["Cost_431_Share"] < 75
        )
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .copy()
)

print("\nHIGH COST / HIGH UTILIZATION PROCEDURES")
print("-" * 70)

print(
    high_cost_high_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Total_Cost",
            "Unique_Patients",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 22. HIGH COST / LOW UTILIZATION
# ============================================================================

high_cost_low_utilization = (
    procedure_summary[
        (
            procedure_summary["Non_431_Total_Cost"]
            >= cost_threshold
        )
        &
        (
            procedure_summary["Procedure_Count"]
            < utilization_threshold
        )
        &
        (
            procedure_summary["Cost_431_Share"] < 75
        )
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .copy()
)

print("\nHIGH COST / LOWER UTILIZATION PROCEDURES")
print("-" * 70)

print(
    high_cost_low_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Non_431_Total_Cost",
            "Average_Procedure_Cost",
            "Cost_Per_Patient",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 23. HIGH UTILIZATION / LOWER COST
# ============================================================================

high_utilization_low_cost = (
    procedure_summary[
        (
            procedure_summary["Procedure_Count"]
            >= utilization_threshold
        )
        &
        (
            procedure_summary["Non_431_Total_Cost"]
            < cost_threshold
        )
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .copy()
)

print("\nHIGH UTILIZATION / LOWER NON-431 COST PROCEDURES")
print("-" * 70)

print(
    high_utilization_low_cost[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Procedures_Per_Patient",
            "Total_Procedure_Cost",
            "Non_431_Total_Cost",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 24. 431-DOMINATED PROCEDURES
# ============================================================================

procedures_431_dominated = (
    procedure_summary[
        procedure_summary["Cost_431_Share"] >= 75
    ]
    .sort_values(
        [
            "Procedure_Count",
            "Total_Procedure_Cost"
        ],
        ascending=[False, False]
    )
    .copy()
)

print("\n431-DOMINATED PROCEDURES")
print("-" * 70)

print(
    procedures_431_dominated[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 25. HIGH COST-PER-PATIENT PROCEDURES
# ============================================================================

high_cost_per_patient = (
    procedure_summary[
        (
            procedure_summary["Cost_Per_Patient"]
            >= cost_per_patient_threshold
        )
        &
        (
            procedure_summary["Unique_Patients"] > 0
        )
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .copy()
)

print("\nHIGH COST-PER-PATIENT PROCEDURES")
print("-" * 70)

print(
    high_cost_per_patient[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 26. HIGH REPEAT UTILIZATION
# ============================================================================

high_repeat_utilization = (
    procedure_summary[
        (
            procedure_summary["Procedures_Per_Patient"]
            >= repeat_utilization_threshold
        )
        &
        (
            procedure_summary["Unique_Patients"] > 0
        )
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .copy()
)

print("\nHIGH REPEAT UTILIZATION")
print("-" * 70)

print(
    high_repeat_utilization[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Procedures_Per_Patient",
            "Unique_Encounters",
            "Total_Procedure_Cost",
            "Cost_431_Share"
        ]
    ].head(20).to_string(index=False)
)


# ============================================================================
# 27. PRIORITY DISTRIBUTION
# ============================================================================

priority_distribution = (
    procedure_summary
    .groupby("Priority_Profile")
    .size()
    .reset_index(
        name="Procedure_Group_Count"
    )
    .sort_values(
        "Procedure_Group_Count",
        ascending=False
    )
)

priority_distribution["Group_Share_%"] = (
    priority_distribution["Procedure_Group_Count"]
    /
    len(procedure_summary)
    * 100
)

print("\nPROCEDURE PRIORITY DISTRIBUTION")
print("-" * 70)
print(
    priority_distribution.to_string(
        index=False
    )
)


# ============================================================================
# 28. COST INTENSITY DISTRIBUTION
# ============================================================================

cost_intensity_distribution = (
    procedure_summary
    .groupby("Cost_Intensity")
    .size()
    .reset_index(
        name="Procedure_Group_Count"
    )
    .sort_values(
        "Procedure_Group_Count",
        ascending=False
    )
)

cost_intensity_distribution["Group_Share_%"] = (
    cost_intensity_distribution["Procedure_Group_Count"]
    /
    len(procedure_summary)
    * 100
)

print("\nCOST INTENSITY DISTRIBUTION")
print("-" * 70)
print(
    cost_intensity_distribution.to_string(
        index=False
    )
)


# ============================================================================
# 29. UTILIZATION INTENSITY DISTRIBUTION
# ============================================================================

utilization_intensity_distribution = (
    procedure_summary
    .groupby("Utilization_Intensity")
    .size()
    .reset_index(
        name="Procedure_Group_Count"
    )
    .sort_values(
        "Procedure_Group_Count",
        ascending=False
    )
)

utilization_intensity_distribution["Group_Share_%"] = (
    utilization_intensity_distribution["Procedure_Group_Count"]
    /
    len(procedure_summary)
    * 100
)

print("\nUTILIZATION INTENSITY DISTRIBUTION")
print("-" * 70)
print(
    utilization_intensity_distribution.to_string(
        index=False
    )
)


# ============================================================================
# 30. EXECUTIVE INSIGHT METRICS
# ============================================================================

priority_1_count = (
    (
        procedure_summary["Priority_Profile"]
        ==
        "PRIORITY 1 — HIGH COST / HIGH UTILIZATION"
    )
    .sum()
)

priority_2_count = (
    (
        procedure_summary["Priority_Profile"]
        ==
        "PRIORITY 2 — HIGH COST / LOWER UTILIZATION"
    )
    .sum()
)

priority_3_count = (
    (
        procedure_summary["Priority_Profile"]
        ==
        "PRIORITY 3 — HIGH UTILIZATION / LOWER COST"
    )
    .sum()
)

priority_4_count = (
    (
        procedure_summary["Priority_Profile"]
        ==
        "PRIORITY 4 — 431 COST DOMINATED"
    )
    .sum()
)

standard_count = (
    (
        procedure_summary["Priority_Profile"]
        ==
        "STANDARD"
    )
    .sum()
)


# ============================================================================
# 31. EXECUTIVE INSIGHT TABLE
# ============================================================================

procedure_insight_metrics = pd.DataFrame({
    "Metric": [
        "Total procedure records",
        "Unique procedure codes",
        "Unique canonical descriptions",
        "Total observed procedure cost",
        "Total non-431 procedure cost",
        "431 record count",
        "Non-431 record count",
        "431 record share (%)",
        "431 cost share (%)",
        "Non-431 cost share (%)",

        "Top 5 total-cost concentration (%)",
        "Top 10 total-cost concentration (%)",
        "Top 20 total-cost concentration (%)",

        "Top 5 non-431 cost concentration (%)",
        "Top 10 non-431 cost concentration (%)",
        "Top 20 non-431 cost concentration (%)",

        "Top 5 utilization concentration (%)",
        "Top 10 utilization concentration (%)",
        "Top 20 utilization concentration (%)",

        "Priority 1 groups",
        "Priority 2 groups",
        "Priority 3 groups",
        "Priority 4 groups",
        "Standard groups",

        "High cost-per-patient groups",
        "High repeat-utilization groups"
    ],
    "Value": [
        total_records,
        unique_codes,
        unique_canonical,
        total_cost,
        non_431_cost,
        records_431,
        records_non_431,
        record_431_share,
        cost_431_share,
        non_431_cost_share,

        top_5_cost_share,
        top_10_cost_share,
        top_20_cost_share,

        top_5_non_431_share,
        top_10_non_431_share,
        top_20_non_431_share,

        top_5_volume_share,
        top_10_volume_share,
        top_20_volume_share,

        priority_1_count,
        priority_2_count,
        priority_3_count,
        priority_4_count,
        standard_count,

        len(high_cost_per_patient),
        len(high_repeat_utilization)
    ]
})


# ============================================================================
# 32. EXECUTIVE INSIGHT FLAGS
# ============================================================================

procedure_insight_summary = pd.DataFrame({
    "Insight_Category": [
        "Cost concentration",
        "Cost concentration",
        "Cost concentration",
        "Utilization concentration",
        "Utilization concentration",
        "Utilization concentration",
        "431 baseline",
        "431 baseline",
        "Priority",
        "Priority",
        "Priority",
        "Priority",
        "Repeat utilization"
    ],
    "Metric": [
        "Top 10 total-cost concentration",
        "Top 10 non-431 cost concentration",
        "Top 20 non-431 cost concentration",
        "Top 10 procedure volume concentration",
        "Top 20 procedure volume concentration",
        "Total procedure volume",
        "431 record share",
        "431 cost share",
        "Priority 1 groups",
        "Priority 2 groups",
        "Priority 3 groups",
        "431-dominated groups",
        "High repeat-utilization groups"
    ],
    "Value": [
        top_10_cost_share,
        top_10_non_431_share,
        top_20_non_431_share,
        top_10_volume_share,
        top_20_volume_share,
        total_records,
        record_431_share,
        cost_431_share,
        priority_1_count,
        priority_2_count,
        priority_3_count,
        priority_4_count,
        len(high_repeat_utilization)
    ]
})


# ============================================================================
# 33. PROCEDURE PRIORITY MATRIX
# ============================================================================

procedure_priority = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Total_Procedure_Cost",
            "Non_431_Total_Cost",
            "Average_Procedure_Cost",
            "Cost_Per_Patient",
            "Procedures_Per_Patient",
            "Cost_431_Share",
            "Cost_Intensity",
            "Utilization_Intensity",
            "Priority_Profile"
        ]
    ]
    .sort_values(
        [
            "Priority_Profile",
            "Non_431_Total_Cost",
            "Procedure_Count"
        ],
        ascending=[True, False, False]
    )
    .copy()
)


# ============================================================================
# 34. TOP NON-431 COST CONTRIBUTORS
# ============================================================================

high_cost_non431_procedures = (
    procedure_summary[
        procedure_summary["Non_431_Total_Cost"] > 0
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .head(50)
    .copy()
)


# ============================================================================
# 35. PATIENT BURDEN INSIGHT
# ============================================================================

patient_procedure_summary = (
    procedure_h
    .groupby("PATIENT")
    .agg(
        Procedure_Count=("CODE", "size"),
        Unique_Procedure_Codes=("CODE", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Non_431_Procedure_Cost=("NON_431_COST", "sum"),
        Cost_431_Count=("IS_431_COST", "sum")
    )
    .reset_index()
)

patient_procedure_summary["Procedures_Per_Encounter"] = (
    patient_procedure_summary["Procedure_Count"]
    /
    patient_procedure_summary["Unique_Encounters"].replace(
        0,
        np.nan
    )
)

patient_procedure_summary["Non_431_Cost_Share"] = (
    patient_procedure_summary["Non_431_Procedure_Cost"]
    /
    patient_procedure_summary["Total_Procedure_Cost"].replace(
        0,
        np.nan
    )
    * 100
)

high_procedure_burden_patients = (
    patient_procedure_summary
    .sort_values(
        [
            "Non_431_Procedure_Cost",
            "Procedure_Count"
        ],
        ascending=[False, False]
    )
    .head(50)
    .copy()
)


# ============================================================================
# 36. PROCEDURE COST BAND
# ============================================================================

procedure_summary["Cost_Band"] = np.select(
    [
        procedure_summary["Cost_431_Share"] >= 75,

        (
            procedure_summary["Non_431_Total_Cost"]
            >= procedure_summary["Non_431_Total_Cost"]
            .quantile(0.99)
        ),

        (
            procedure_summary["Non_431_Total_Cost"]
            >= procedure_summary["Non_431_Total_Cost"]
            .quantile(0.95)
        ),

        (
            procedure_summary["Non_431_Total_Cost"]
            > 0
        )
    ],
    [
        "431 Cost Dominated",
        "Extreme Non-431 Cost",
        "High Non-431 Cost",
        "Standard Non-431 Cost"
    ],
    default="No Non-431 Cost"
)


# ============================================================================
# 37. FINAL DATASET CLEANING
# ============================================================================

# Ensure numerical columns are finite where mathematically appropriate.

for df in [
    procedure_summary,
    procedure_priority,
    high_cost_high_utilization,
    high_cost_low_utilization,
    high_utilization_low_cost,
    high_cost_per_patient,
    high_repeat_utilization
]:

    numeric_columns = df.select_dtypes(
        include=[np.number]
    ).columns

    for col in numeric_columns:
        df[col] = df[col].replace(
            [np.inf, -np.inf],
            np.nan
        )


# ============================================================================
# 38. FINAL VALIDATION
# ============================================================================

record_count_preserved = (
    len(procedure_h) == len(source_df)
)

original_code_preserved = (
    procedure_h["ORIGINAL_CODE"].astype("string").tolist()
    ==
    source_df["ORIGINAL_CODE"].astype("string").tolist()
)

original_description_preserved = (
    procedure_h["ORIGINAL_DESCRIPTION"].astype("string").tolist()
    ==
    source_df["ORIGINAL_DESCRIPTION"].astype("string").tolist()
)

canonical_available = (
    procedure_h["CANONICAL_DESCRIPTION"].notna().all()
)

canonical_unique_per_code = (
    len(multi_canonical_codes) == 0
)

procedure_summary_created = (
    len(procedure_summary) == unique_codes
)

priority_profile_created = (
    procedure_summary["Priority_Profile"].notna().all()
)

no_records_deleted = (
    len(procedure_h) == len(source_df)
)

validation = pd.DataFrame({
    "Validation": [
        "Record count preserved",
        "Original CODE preserved",
        "Original DESCRIPTION preserved",
        "Canonical description available",
        "One canonical description per CODE",
        "431 cost segmentation available",
        "Non-431 cost available",
        "Procedure-level summary created",
        "Priority profile created",
        "Patient-level procedure summary created",
        "Cost concentration calculated",
        "Utilization concentration calculated",
        "No records deleted"
    ],
    "Result": [
        record_count_preserved,
        original_code_preserved,
        original_description_preserved,
        canonical_available,
        canonical_unique_per_code,
        "IS_431_COST" in procedure_h.columns,
        "NON_431_COST" in procedure_h.columns,
        procedure_summary_created,
        priority_profile_created,
        len(patient_procedure_summary) > 0,
        top_10_cost_share >= 0,
        top_10_volume_share >= 0,
        no_records_deleted
    ]
})

print("\nFINAL VALIDATION")
print("-" * 70)
print(
    validation.to_string(index=False)
)

if not validation["Result"].all():
    raise ValueError(
        "STEP 4.18H validation failed."
    )


# ============================================================================
# 39. FINAL EXECUTIVE SUMMARY
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18H COMPLETE")
print("=" * 70)

print(
    f"Total procedure records: "
    f"{total_records:,}"
)

print(
    f"Unique procedure codes: "
    f"{unique_codes:,}"
)

print(
    f"Unique canonical descriptions: "
    f"{unique_canonical:,}"
)

print(
    f"Total observed procedure cost: "
    f"{total_cost:,.2f}"
)

print(
    f"Total non-431 procedure cost: "
    f"{non_431_cost:,.2f}"
)

print(
    f"431 record share: "
    f"{record_431_share:.2f}%"
)

print(
    f"431 cost share: "
    f"{cost_431_share:.2f}%"
)

print(
    f"Top 10 total-cost concentration: "
    f"{top_10_cost_share:.2f}%"
)

print(
    f"Top 20 total-cost concentration: "
    f"{top_20_cost_share:.2f}%"
)

print(
    f"Top 10 non-431 cost concentration: "
    f"{top_10_non_431_share:.2f}%"
)

print(
    f"Top 20 non-431 cost concentration: "
    f"{top_20_non_431_share:.2f}%"
)

print(
    f"Top 10 procedure volume concentration: "
    f"{top_10_volume_share:.2f}%"
)

print(
    f"Top 20 procedure volume concentration: "
    f"{top_20_volume_share:.2f}%"
)

print(
    f"Priority 1 groups: "
    f"{priority_1_count}"
)

print(
    f"Priority 2 groups: "
    f"{priority_2_count}"
)

print(
    f"Priority 3 groups: "
    f"{priority_3_count}"
)

print(
    f"431-dominated groups: "
    f"{priority_4_count}"
)

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_h")
print("  - procedure_summary")
print("  - top_cost_procedures")
print("  - top_non_431_cost")
print("  - top_utilization")
print("  - top_cost_per_patient")
print("  - high_repeat_utilization")
print("  - high_cost_high_utilization")
print("  - high_cost_low_utilization")
print("  - high_utilization_low_cost")
print("  - procedures_431_dominated")
print("  - high_cost_per_patient")
print("  - high_cost_non431_procedures")
print("  - patient_procedure_summary")
print("  - high_procedure_burden_patients")
print("  - procedure_priority")
print("  - priority_distribution")
print("  - cost_intensity_distribution")
print("  - utilization_intensity_distribution")
print("  - procedure_insight_metrics")
print("  - procedure_insight_summary")
print("  - procedure_g_validation")

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================================
# STEP 4.18I — PROCEDURE EXECUTIVE / CLINICAL INSIGHT &
#              ACTION PRIORITIZATION
# ============================================================================
#
# PURPOSE
# -------
# Build the final procedure-level executive analytics layer on top of
# procedure_h.
#
# Covers:
#   1. Source validation
#   2. Procedure executive scorecard
#   3. Financial concentration
#   4. Volume concentration
#   5. Clinical/utilization burden
#   6. High-cost / high-utilization procedures
#   7. High-cost / lower-utilization procedures
#   8. High-utilization / lower-cost procedures
#   9. High patient-cost procedures
#  10. High repeat-utilization procedures
#  11. BASE_COST = 431 dependency
#  12. Patient-level procedure burden
#  13. Patient concentration
#  14. Executive priority scoring
#  15. Executive action matrix
#  16. Validation
#
# IMPORTANT
# ----------
# High cost, high utilization, or high repeat utilization DOES NOT mean
# clinically inappropriate care.
#
# These are analytical review signals only.
# ============================================================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 220)
pd.set_option("display.max_colwidth", 100)


# ============================================================================
# 1. SOURCE DATA DISCOVERY
# ============================================================================

print("=" * 70)
print("STEP 4.18I — PROCEDURE EXECUTIVE / CLINICAL INSIGHT &")
print("             ACTION PRIORITIZATION")
print("=" * 70)

# ---------------------------------------------------------------------------
# Prefer procedure_h because STEP 4.18H is the validated upstream dataset.
# ---------------------------------------------------------------------------

if "procedure_h" in globals():
    source_df = procedure_h.copy()
    source_name = "procedure_h"

elif "procedure_g" in globals():
    # Fallback only if procedure_h is not available.
    source_df = procedure_g.copy()
    source_name = "procedure_g"

elif "procedure_analysis" in globals():
    source_df = procedure_analysis.copy()
    source_name = "procedure_analysis"

else:
    raise NameError(
        "No validated procedure source dataset found. "
        "Expected procedure_h."
    )

print("\nSOURCE DATA DISCOVERY")
print("-" * 70)
print(f"Using source dataset: {source_name}")
print(f"Source records: {len(source_df):,}")


# ============================================================================
# 2. COLUMN RESOLUTION
# ============================================================================
#
# The upstream datasets may contain ORIGINAL_CODE / CODE,
# ORIGINAL_DESCRIPTION / DESCRIPTION / CANONICAL_DESCRIPTION, etc.
#
# We resolve columns without modifying the original dataset.
# ============================================================================

def resolve_column(df, candidates, required=True):
    """
    Return the first matching column from candidates.

    Matching is:
      1. Exact
      2. Case-insensitive
      3. Normalized
    """

    # Exact match
    for col in candidates:
        if col in df.columns:
            return col

    # Case-insensitive match
    lower_map = {str(c).lower(): c for c in df.columns}

    for col in candidates:
        if str(col).lower() in lower_map:
            return lower_map[str(col).lower()]

    # Normalized match
    def normalize_name(x):
        return (
            str(x)
            .strip()
            .upper()
            .replace(" ", "_")
            .replace("-", "_")
        )

    normalized_map = {
        normalize_name(c): c
        for c in df.columns
    }

    for col in candidates:
        key = normalize_name(col)

        if key in normalized_map:
            return normalized_map[key]

    if required:
        raise ValueError(
            f"Required column not found. "
            f"Expected one of: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


CODE_COL = resolve_column(
    source_df,
    [
        "CODE",
        "ORIGINAL_CODE",
        "PROCEDURE_CODE",
    ]
)

ORIGINAL_CODE_COL = resolve_column(
    source_df,
    [
        "ORIGINAL_CODE",
        "CODE",
        "PROCEDURE_CODE",
    ]
)

CANONICAL_DESC_COL = resolve_column(
    source_df,
    [
        "CANONICAL_DESCRIPTION",
        "CANONICAL_DESC",
        "DESCRIPTION_CANONICAL",
        "DESCRIPTION",
        "ORIGINAL_DESCRIPTION",
    ]
)

ORIGINAL_DESC_COL = resolve_column(
    source_df,
    [
        "ORIGINAL_DESCRIPTION",
        "DESCRIPTION",
        "CANONICAL_DESCRIPTION",
    ]
)

PATIENT_COL = resolve_column(
    source_df,
    [
        "PATIENT",
        "PATIENT_ID",
        "PATIENTID",
    ]
)

ENCOUNTER_COL = resolve_column(
    source_df,
    [
        "ENCOUNTER",
        "ENCOUNTER_ID",
        "ENCOUNTERID",
    ]
)

BASE_COST_COL = resolve_column(
    source_df,
    [
        "BASE_COST",
        "Base_Cost",
        "base_cost",
    ]
)

PROCEDURE_COST_COL = resolve_column(
    source_df,
    [
        "PROCEDURE_COST",
        "TOTAL_PROCEDURE_COST",
        "TOTAL_COST",
        "COST",
        "AMOUNT",
        "BASE_COST",
    ]
)


print("\nCOLUMN RESOLUTION")
print("-" * 70)
print(f"CODE:                    {CODE_COL}")
print(f"ORIGINAL_CODE:           {ORIGINAL_CODE_COL}")
print(f"CANONICAL_DESCRIPTION:   {CANONICAL_DESC_COL}")
print(f"ORIGINAL_DESCRIPTION:    {ORIGINAL_DESC_COL}")
print(f"PATIENT:                 {PATIENT_COL}")
print(f"ENCOUNTER:               {ENCOUNTER_COL}")
print(f"BASE_COST:               {BASE_COST_COL}")
print(f"PROCEDURE_COST:          {PROCEDURE_COST_COL}")


# ============================================================================
# 3. CREATE STANDARDIZED INTERNAL ANALYTICAL VIEW
# ============================================================================
#
# Do not overwrite source_df.
# ============================================================================

work = source_df.copy()

work["_CODE"] = work[CODE_COL]
work["_ORIGINAL_CODE"] = work[ORIGINAL_CODE_COL]
work["_CANONICAL_DESCRIPTION"] = work[CANONICAL_DESC_COL]
work["_ORIGINAL_DESCRIPTION"] = work[ORIGINAL_DESC_COL]
work["_PATIENT"] = work[PATIENT_COL]
work["_ENCOUNTER"] = work[ENCOUNTER_COL]

work["_BASE_COST"] = pd.to_numeric(
    work[BASE_COST_COL],
    errors="coerce"
)

work["_PROCEDURE_COST"] = pd.to_numeric(
    work[PROCEDURE_COST_COL],
    errors="coerce"
)

# If procedure cost is missing, fall back to BASE_COST.
work["_PROCEDURE_COST"] = work["_PROCEDURE_COST"].fillna(
    work["_BASE_COST"]
)

# ---------------------------------------------------------------------------
# Validate cost availability.
# ---------------------------------------------------------------------------

if work["_PROCEDURE_COST"].isna().all():
    raise ValueError(
        "No usable procedure cost values were found."
    )

# Missing costs are treated as zero only after reporting the issue.
missing_cost_count = work["_PROCEDURE_COST"].isna().sum()

if missing_cost_count > 0:
    print(
        f"\nWARNING: {missing_cost_count:,} procedure records "
        f"have missing procedure cost. They will be treated as 0 "
        f"for aggregation."
    )

work["_PROCEDURE_COST"] = work["_PROCEDURE_COST"].fillna(0)

# ---------------------------------------------------------------------------
# BASE_COST = 431 flag
# ---------------------------------------------------------------------------

work["_IS_431"] = np.isclose(
    work["_BASE_COST"].fillna(np.nan),
    431
)

work["_IS_NON_431"] = ~work["_IS_431"]


# ============================================================================
# 4. SOURCE DATA CHECK
# ============================================================================

SOURCE_RECORD_COUNT = len(source_df)

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS:", "PASS" if SOURCE_RECORD_COUNT > 0 else "FAIL")
print(f"Procedure records available: {SOURCE_RECORD_COUNT:,}")


# ============================================================================
# 5. CANONICALIZATION INTEGRITY
# ============================================================================

missing_canonical = work["_CANONICAL_DESCRIPTION"].isna().sum()

blank_canonical = (
    work["_CANONICAL_DESCRIPTION"]
    .astype("string")
    .str.strip()
    .eq("")
    .sum()
)

code_canonical_counts = (
    work.groupby("_CODE")["_CANONICAL_DESCRIPTION"]
    .nunique(dropna=True)
)

multiple_canonical_codes = (
    code_canonical_counts[
        code_canonical_counts > 1
    ]
)

unique_codes = work["_CODE"].nunique()
unique_canonical = work["_CANONICAL_DESCRIPTION"].nunique()

canonical_integrity_pass = (
    missing_canonical == 0
    and blank_canonical == 0
    and len(multiple_canonical_codes) == 0
)

print("\nANALYTICAL DATASET VALIDATION")
print("-" * 70)
print(f"Records: {len(work):,}")
print(f"Unique procedure codes: {unique_codes:,}")
print(f"Unique canonical descriptions: {unique_canonical:,}")

print("\nCANONICALIZATION INTEGRITY")
print("-" * 70)
print(f"Missing canonical descriptions: {missing_canonical:,}")
print(f"Blank canonical descriptions: {blank_canonical:,}")
print(
    f"Codes with multiple canonical descriptions: "
    f"{len(multiple_canonical_codes):,}"
)
print(
    "Canonicalization integrity:",
    "PASS" if canonical_integrity_pass else "FAIL"
)

if not canonical_integrity_pass:
    raise ValueError(
        "Canonicalization integrity failed."
    )


# ============================================================================
# 6. PROCEDURE-LEVEL EXECUTIVE SCORECARD
# ============================================================================

procedure_summary = (
    work.groupby(
        ["_CODE", "_CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("_CODE", "size"),
        Unique_Patients=("_PATIENT", "nunique"),
        Unique_Encounters=("_ENCOUNTER", "nunique"),
        Total_Procedure_Cost=("_PROCEDURE_COST", "sum"),
        Cost_431=(
            "_PROCEDURE_COST",
            lambda x: x[work.loc[x.index, "_IS_431"]].sum()
        ),
        Non_431_Total_Cost=(
            "_PROCEDURE_COST",
            lambda x: x[work.loc[x.index, "_IS_NON_431"]].sum()
        ),
        Cost_431_Count=(
            "_IS_431",
            "sum"
        ),
    )
    .reset_index()
)

procedure_summary.rename(
    columns={
        "_CODE": "CODE",
        "_CANONICAL_DESCRIPTION": "CANONICAL_DESCRIPTION",
    },
    inplace=True
)

procedure_summary["Average_Procedure_Cost"] = np.where(
    procedure_summary["Procedure_Count"] > 0,
    procedure_summary["Total_Procedure_Cost"]
    / procedure_summary["Procedure_Count"],
    0
)

procedure_summary["Cost_Per_Patient"] = np.where(
    procedure_summary["Unique_Patients"] > 0,
    procedure_summary["Total_Procedure_Cost"]
    / procedure_summary["Unique_Patients"],
    0
)

procedure_summary["Cost_Per_Encounter"] = np.where(
    procedure_summary["Unique_Encounters"] > 0,
    procedure_summary["Total_Procedure_Cost"]
    / procedure_summary["Unique_Encounters"],
    0
)

procedure_summary["Procedures_Per_Patient"] = np.where(
    procedure_summary["Unique_Patients"] > 0,
    procedure_summary["Procedure_Count"]
    / procedure_summary["Unique_Patients"],
    0
)

procedure_summary["Procedures_Per_Encounter"] = np.where(
    procedure_summary["Unique_Encounters"] > 0,
    procedure_summary["Procedure_Count"]
    / procedure_summary["Unique_Encounters"],
    0
)

procedure_summary["Cost_431_Share"] = np.where(
    procedure_summary["Total_Procedure_Cost"] > 0,
    procedure_summary["Cost_431"]
    / procedure_summary["Total_Procedure_Cost"]
    * 100,
    0
)

procedure_summary["Non_431_Cost_Share"] = np.where(
    procedure_summary["Total_Procedure_Cost"] > 0,
    procedure_summary["Non_431_Total_Cost"]
    / procedure_summary["Total_Procedure_Cost"]
    * 100,
    0
)


# ============================================================================
# 7. GLOBAL EXECUTIVE TOTALS
# ============================================================================

TOTAL_COST = work["_PROCEDURE_COST"].sum()
TOTAL_431_COST = work.loc[
    work["_IS_431"],
    "_PROCEDURE_COST"
].sum()

TOTAL_NON_431_COST = work.loc[
    work["_IS_NON_431"],
    "_PROCEDURE_COST"
].sum()

RECORDS_431 = int(work["_IS_431"].sum())
RECORDS_NON_431 = int(work["_IS_NON_431"].sum())

RECORD_SHARE_431 = (
    RECORDS_431 / len(work) * 100
    if len(work) > 0 else 0
)

COST_SHARE_431 = (
    TOTAL_431_COST / TOTAL_COST * 100
    if TOTAL_COST > 0 else 0
)

COST_SHARE_NON_431 = (
    TOTAL_NON_431_COST / TOTAL_COST * 100
    if TOTAL_COST > 0 else 0
)


print("\nPROCEDURE EXECUTIVE KPIs")
print("-" * 70)
print(f"Total procedure records: {len(work):,}")
print(f"Unique procedure codes: {unique_codes:,}")
print(f"Unique canonical descriptions: {unique_canonical:,}")
print(f"Unique patients: {work['_PATIENT'].nunique():,}")
print(f"Unique encounters: {work['_ENCOUNTER'].nunique():,}")

print(f"\nTotal procedure cost: {TOTAL_COST:,.2f}")
print(f"431 cost: {TOTAL_431_COST:,.2f}")
print(f"Non-431 cost: {TOTAL_NON_431_COST:,.2f}")

print(f"\n431 record share: {RECORD_SHARE_431:.2f}%")
print(f"431 cost share: {COST_SHARE_431:.2f}%")
print(f"Non-431 cost share: {COST_SHARE_NON_431:.2f}%")


# ============================================================================
# 8. COST AND VOLUME CONCENTRATION
# ============================================================================

procedure_summary = procedure_summary.sort_values(
    "Total_Procedure_Cost",
    ascending=False
).reset_index(drop=True)

procedure_summary["Cost_Concentration_Share"] = (
    procedure_summary["Total_Procedure_Cost"]
    / TOTAL_COST
    * 100
    if TOTAL_COST > 0 else 0
)

procedure_summary["Cumulative_Cost_Share"] = (
    procedure_summary["Cost_Concentration_Share"]
    .cumsum()
)

procedure_summary["Volume_Concentration_Share"] = (
    procedure_summary["Procedure_Count"]
    / len(work)
    * 100
)

procedure_summary["Cumulative_Volume_Share"] = (
    procedure_summary["Volume_Concentration_Share"]
    .cumsum()
)


def concentration_share(df, metric, n):
    if len(df) == 0:
        return 0

    return (
        df.nlargest(n, metric)[metric].sum()
        / df[metric].sum()
        * 100
        if df[metric].sum() > 0 else 0
    )


top1_cost_share = concentration_share(
    procedure_summary,
    "Total_Procedure_Cost",
    1
)

top5_cost_share = concentration_share(
    procedure_summary,
    "Total_Procedure_Cost",
    5
)

top10_cost_share = concentration_share(
    procedure_summary,
    "Total_Procedure_Cost",
    10
)

top20_cost_share = concentration_share(
    procedure_summary,
    "Total_Procedure_Cost",
    20
)

top5_volume_share = concentration_share(
    procedure_summary,
    "Procedure_Count",
    5
)

top10_volume_share = concentration_share(
    procedure_summary,
    "Procedure_Count",
    10
)

top20_volume_share = concentration_share(
    procedure_summary,
    "Procedure_Count",
    20
)

top5_non431_share = concentration_share(
    procedure_summary,
    "Non_431_Total_Cost",
    5
)

top10_non431_share = concentration_share(
    procedure_summary,
    "Non_431_Total_Cost",
    10
)

top20_non431_share = concentration_share(
    procedure_summary,
    "Non_431_Total_Cost",
    20
)

print("\nPROCEDURE COST CONCENTRATION")
print("-" * 70)
print(f"Top 1 total-cost share: {top1_cost_share:.2f}%")
print(f"Top 5 total-cost share: {top5_cost_share:.2f}%")
print(f"Top 10 total-cost share: {top10_cost_share:.2f}%")
print(f"Top 20 total-cost share: {top20_cost_share:.2f}%")

print(f"\nTop 5 non-431 cost share: {top5_non431_share:.2f}%")
print(f"Top 10 non-431 cost share: {top10_non431_share:.2f}%")
print(f"Top 20 non-431 cost share: {top20_non431_share:.2f}%")

print("\nPROCEDURE VOLUME CONCENTRATION")
print("-" * 70)
print(f"Top 5 volume share: {top5_volume_share:.2f}%")
print(f"Top 10 volume share: {top10_volume_share:.2f}%")
print(f"Top 20 volume share: {top20_volume_share:.2f}%")


# ============================================================================
# 9. RANKINGS
# ============================================================================

procedure_summary["Cost_Rank"] = (
    procedure_summary["Total_Procedure_Cost"]
    .rank(method="min", ascending=False)
    .astype(int)
)

procedure_summary["Non_431_Cost_Rank"] = (
    procedure_summary["Non_431_Total_Cost"]
    .rank(method="min", ascending=False)
    .astype(int)
)

procedure_summary["Utilization_Rank"] = (
    procedure_summary["Procedure_Count"]
    .rank(method="min", ascending=False)
    .astype(int)
)

procedure_summary["Cost_Per_Patient_Rank"] = (
    procedure_summary["Cost_Per_Patient"]
    .rank(method="min", ascending=False)
    .astype(int)
)

procedure_summary["Repeat_Utilization_Rank"] = (
    procedure_summary["Procedures_Per_Patient"]
    .rank(method="min", ascending=False)
    .astype(int)
)


# ============================================================================
# 10. BASE_COST DEPENDENCY CLASSIFICATION
# ============================================================================

def classify_431_dependency(row):

    share = row["Cost_431_Share"]

    if share >= 90:
        return "431_DOMINATED"

    elif share <= 10:
        return "NON_431_DOMINATED"

    else:
        return "MIXED_431_NON431"


procedure_summary["431_Dependency"] = (
    procedure_summary.apply(
        classify_431_dependency,
        axis=1
    )
)


# ============================================================================
# 11. ANALYTICAL THRESHOLDS
# ============================================================================
#
# Reuse the validated 4.18H methodology where possible.
#
# Quantile thresholds provide robust relative prioritization and avoid
# arbitrary absolute cutoffs.
# ============================================================================

high_non431_cost_threshold = (
    procedure_summary["Non_431_Total_Cost"]
    .quantile(0.75)
)

high_utilization_threshold = (
    procedure_summary["Procedure_Count"]
    .quantile(0.75)
)

high_cost_per_patient_threshold = (
    procedure_summary["Cost_Per_Patient"]
    .quantile(0.75)
)

high_repeat_threshold = (
    procedure_summary["Procedures_Per_Patient"]
    .quantile(0.75)
)

extreme_non431_cost_threshold = (
    procedure_summary["Non_431_Total_Cost"]
    .quantile(0.90)
)

extreme_utilization_threshold = (
    procedure_summary["Procedure_Count"]
    .quantile(0.90)
)

extreme_cost_per_patient_threshold = (
    procedure_summary["Cost_Per_Patient"]
    .quantile(0.90)
)

extreme_repeat_threshold = (
    procedure_summary["Procedures_Per_Patient"]
    .quantile(0.90)
)


print("\nANALYTICAL THRESHOLDS")
print("-" * 70)
print(
    f"High non-431 cost threshold: "
    f"{high_non431_cost_threshold:,.2f}"
)

print(
    f"High utilization threshold: "
    f"{high_utilization_threshold:,.2f}"
)

print(
    f"High cost-per-patient threshold: "
    f"{high_cost_per_patient_threshold:,.2f}"
)

print(
    f"High repeat-utilization threshold: "
    f"{high_repeat_threshold:,.2f}"
)

print(
    f"Extreme non-431 cost threshold: "
    f"{extreme_non431_cost_threshold:,.2f}"
)

print(
    f"Extreme utilization threshold: "
    f"{extreme_utilization_threshold:,.2f}"
)


# ============================================================================
# 12. EXECUTIVE PRIORITY SCORE
# ============================================================================
#
# Score components:
#
#   Financial exposure       0–30
#   Utilization burden       0–25
#   Patient cost exposure    0–20
#   Repeat utilization       0–15
#   Concentration contribution 0–10
#
# Maximum = 100
#
# BASE_COST dependency is deliberately NOT used as a penalty. It is
# reported separately so that high activity is not incorrectly interpreted
# as high financial exposure.
# ============================================================================

# Financial component
procedure_summary["Financial_Score"] = np.select(
    [
        procedure_summary["Non_431_Total_Cost"]
        >= extreme_non431_cost_threshold,

        procedure_summary["Non_431_Total_Cost"]
        >= high_non431_cost_threshold
    ],
    [
        30,
        20
    ],
    default=5
)

# Utilization component
procedure_summary["Utilization_Score"] = np.select(
    [
        procedure_summary["Procedure_Count"]
        >= extreme_utilization_threshold,

        procedure_summary["Procedure_Count"]
        >= high_utilization_threshold
    ],
    [
        25,
        17
    ],
    default=5
)

# Patient cost component
procedure_summary["Patient_Cost_Score"] = np.select(
    [
        procedure_summary["Cost_Per_Patient"]
        >= extreme_cost_per_patient_threshold,

        procedure_summary["Cost_Per_Patient"]
        >= high_cost_per_patient_threshold
    ],
    [
        20,
        13
    ],
    default=4
)

# Repeat utilization component
procedure_summary["Repeat_Score"] = np.select(
    [
        procedure_summary["Procedures_Per_Patient"]
        >= extreme_repeat_threshold,

        procedure_summary["Procedures_Per_Patient"]
        >= high_repeat_threshold
    ],
    [
        15,
        10
    ],
    default=3
)

# Concentration component
procedure_summary["Concentration_Score"] = np.select(
    [
        procedure_summary["Cost_Concentration_Share"]
        >= 5,

        procedure_summary["Cost_Concentration_Share"]
        >= 2
    ],
    [
        10,
        6
    ],
    default=2
)

procedure_summary["Executive_Score"] = (
    procedure_summary["Financial_Score"]
    + procedure_summary["Utilization_Score"]
    + procedure_summary["Patient_Cost_Score"]
    + procedure_summary["Repeat_Score"]
    + procedure_summary["Concentration_Score"]
)


# ============================================================================
# 13. EXECUTIVE PRIORITY PROFILE
# ============================================================================

def classify_priority(row):

    cost = row["Non_431_Total_Cost"]
    utilization = row["Procedure_Count"]
    patient_cost = row["Cost_Per_Patient"]
    repeat = row["Procedures_Per_Patient"]
    dependency = row["431_Dependency"]

    extreme_cost = cost >= extreme_non431_cost_threshold
    high_cost = cost >= high_non431_cost_threshold

    extreme_util = utilization >= extreme_utilization_threshold
    high_util = utilization >= high_utilization_threshold

    extreme_patient_cost = (
        patient_cost >= extreme_cost_per_patient_threshold
    )

    high_patient_cost = (
        patient_cost >= high_cost_per_patient_threshold
    )

    extreme_repeat = (
        repeat >= extreme_repeat_threshold
    )

    high_repeat = (
        repeat >= high_repeat_threshold
    )

    # Highest priority: extreme cost + extreme utilization
    if extreme_cost and extreme_util:
        return "CRITICAL_FINANCIAL_PRIORITY"

    # High cost + high utilization
    if high_cost and high_util:
        return "HIGH_FINANCIAL_PRIORITY"

    # Very high patient cost
    if extreme_patient_cost:
        return "HIGH_PATIENT_COST"

    # High repeat burden
    if extreme_repeat:
        return "HIGH_REPEAT_BURDEN"

    # High utilization without high financial exposure
    if high_util:
        return "HIGH_UTILIZATION_PRIORITY"

    # 431-dominated activity
    if dependency == "431_DOMINATED":
        return "431_DOMINATED"

    # Moderate financial attention
    if high_cost:
        return "HIGH_FINANCIAL_PRIORITY"

    return "STANDARD"


procedure_summary["Executive_Priority"] = (
    procedure_summary.apply(
        classify_priority,
        axis=1
    )
)


# ============================================================================
# 14. PROCEDURE EXECUTIVE SCORECARD
# ============================================================================

procedure_executive_scorecard = procedure_summary[
    [
        "CODE",
        "CANONICAL_DESCRIPTION",
        "Procedure_Count",
        "Unique_Patients",
        "Unique_Encounters",
        "Total_Procedure_Cost",
        "Non_431_Total_Cost",
        "Cost_431",
        "Cost_431_Share",
        "Non_431_Cost_Share",
        "Average_Procedure_Cost",
        "Cost_Per_Patient",
        "Cost_Per_Encounter",
        "Procedures_Per_Patient",
        "Procedures_Per_Encounter",
        "Cost_Concentration_Share",
        "Volume_Concentration_Share",
        "Cost_Rank",
        "Non_431_Cost_Rank",
        "Utilization_Rank",
        "Cost_Per_Patient_Rank",
        "Repeat_Utilization_Rank",
        "431_Dependency",
        "Executive_Score",
        "Executive_Priority"
    ]
].copy()

procedure_executive_scorecard = (
    procedure_executive_scorecard
    .sort_values(
        [
            "Executive_Score",
            "Non_431_Total_Cost"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 15. TOP PROCEDURE TABLES
# ============================================================================

top_cost_procedures = (
    procedure_summary
    .nlargest(20, "Total_Procedure_Cost")
    .copy()
)

top_non_431_cost = (
    procedure_summary
    .nlargest(20, "Non_431_Total_Cost")
    .copy()
)

top_utilization = (
    procedure_summary
    .nlargest(20, "Procedure_Count")
    .copy()
)

top_cost_per_patient = (
    procedure_summary
    .nlargest(20, "Cost_Per_Patient")
    .copy()
)

high_cost_per_patient = (
    procedure_summary[
        procedure_summary["Cost_Per_Patient"]
        >= high_cost_per_patient_threshold
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .copy()
)


# ============================================================================
# 16. HIGH COST / HIGH UTILIZATION
# ============================================================================

high_cost_high_utilization_i = procedure_summary[
    (
        procedure_summary["Non_431_Total_Cost"]
        >= high_non431_cost_threshold
    )
    &
    (
        procedure_summary["Procedure_Count"]
        >= high_utilization_threshold
    )
].copy()

high_cost_high_utilization_i = (
    high_cost_high_utilization_i
    .sort_values(
        [
            "Non_431_Total_Cost",
            "Procedure_Count"
        ],
        ascending=False
    )
)


# ============================================================================
# 17. HIGH COST / LOWER UTILIZATION
# ============================================================================

high_cost_low_utilization_i = procedure_summary[
    (
        procedure_summary["Non_431_Total_Cost"]
        >= high_non431_cost_threshold
    )
    &
    (
        procedure_summary["Procedure_Count"]
        < high_utilization_threshold
    )
].copy()

high_cost_low_utilization_i = (
    high_cost_low_utilization_i
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
)


# ============================================================================
# 18. HIGH UTILIZATION / LOWER NON-431 COST
# ============================================================================

high_utilization_low_cost_i = procedure_summary[
    (
        procedure_summary["Procedure_Count"]
        >= high_utilization_threshold
    )
    &
    (
        procedure_summary["Non_431_Total_Cost"]
        < high_non431_cost_threshold
    )
].copy()

high_utilization_low_cost_i = (
    high_utilization_low_cost_i
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
)


# ============================================================================
# 19. HIGH PATIENT-COST PROCEDURES
# ============================================================================

high_patient_cost_i = procedure_summary[
    procedure_summary["Cost_Per_Patient"]
    >= high_cost_per_patient_threshold
].copy()

high_patient_cost_i = (
    high_patient_cost_i
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
)


# ============================================================================
# 20. HIGH REPEAT-UTILIZATION PROCEDURES
# ============================================================================

high_repeat_procedures_i = procedure_summary[
    procedure_summary["Procedures_Per_Patient"]
    >= high_repeat_threshold
].copy()

high_repeat_procedures_i = (
    high_repeat_procedures_i
    .sort_values(
        [
            "Procedures_Per_Patient",
            "Procedure_Count"
        ],
        ascending=False
    )
)


# ============================================================================
# 21. 431 DEPENDENCY ANALYSIS
# ============================================================================

procedure_431_dependency = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_431",
            "Non_431_Total_Cost",
            "Cost_431_Share",
            "Non_431_Cost_Share",
            "431_Dependency",
            "Executive_Priority"
        ]
    ]
    .sort_values(
        [
            "431_Dependency",
            "Total_Procedure_Cost"
        ],
        ascending=[True, False]
    )
    .copy()
)


# ============================================================================
# 22. HIGH PROCEDURE BURDEN AT PATIENT LEVEL
# ============================================================================

patient_procedure_summary = (
    work.groupby(
        "_PATIENT",
        dropna=False
    )
    .agg(
        Procedure_Count=("_CODE", "size"),
        Unique_Procedure_Codes=("_CODE", "nunique"),
        Unique_Encounters=("_ENCOUNTER", "nunique"),
        Total_Procedure_Cost=("_PROCEDURE_COST", "sum"),
        Non_431_Cost=(
            "_PROCEDURE_COST",
            lambda x: x[work.loc[x.index, "_IS_NON_431"]].sum()
        ),
        Cost_431=(
            "_PROCEDURE_COST",
            lambda x: x[work.loc[x.index, "_IS_431"]].sum()
        ),
    )
    .reset_index()
)

patient_procedure_summary.rename(
    columns={
        "_PATIENT": "PATIENT"
    },
    inplace=True
)

patient_procedure_summary["Procedures_Per_Encounter"] = np.where(
    patient_procedure_summary["Unique_Encounters"] > 0,
    patient_procedure_summary["Procedure_Count"]
    / patient_procedure_summary["Unique_Encounters"],
    0
)

patient_procedure_summary["Procedures_Per_Code"] = np.where(
    patient_procedure_summary["Unique_Procedure_Codes"] > 0,
    patient_procedure_summary["Procedure_Count"]
    / patient_procedure_summary["Unique_Procedure_Codes"],
    0
)

patient_procedure_summary["Cost_431_Share"] = np.where(
    patient_procedure_summary["Total_Procedure_Cost"] > 0,
    patient_procedure_summary["Cost_431"]
    / patient_procedure_summary["Total_Procedure_Cost"]
    * 100,
    0
)

patient_procedure_summary["Non_431_Cost_Share"] = np.where(
    patient_procedure_summary["Total_Procedure_Cost"] > 0,
    patient_procedure_summary["Non_431_Cost"]
    / patient_procedure_summary["Total_Procedure_Cost"]
    * 100,
    0
)


# ============================================================================
# 23. IDENTIFY HIGHEST-COST PROCEDURE PER PATIENT
# ============================================================================

patient_procedure_lookup = (
    work.groupby(
        [
            "_PATIENT",
            "_CODE",
            "_CANONICAL_DESCRIPTION"
        ],
        dropna=False
    )
    .agg(
        Procedure_Count=("_CODE", "size"),
        Procedure_Cost=("_PROCEDURE_COST", "sum")
    )
    .reset_index()
)

idx_highest_patient_procedure = (
    patient_procedure_lookup
    .groupby("_PATIENT")["Procedure_Cost"]
    .idxmax()
)

highest_patient_procedure = (
    patient_procedure_lookup
    .loc[idx_highest_patient_procedure]
    .copy()
)

highest_patient_procedure.rename(
    columns={
        "_PATIENT": "PATIENT",
        "_CODE": "Highest_Cost_Procedure_Code",
        "_CANONICAL_DESCRIPTION":
            "Highest_Cost_Procedure",
        "Procedure_Cost":
            "Highest_Procedure_Cost"
    },
    inplace=True
)

patient_procedure_summary = patient_procedure_summary.merge(
    highest_patient_procedure[
        [
            "PATIENT",
            "Highest_Cost_Procedure_Code",
            "Highest_Cost_Procedure",
            "Highest_Procedure_Cost"
        ]
    ],
    on="PATIENT",
    how="left"
)


# ============================================================================
# 24. PATIENT PROCEDURE BURDEN THRESHOLDS
# ============================================================================

patient_count_threshold = (
    patient_procedure_summary["Procedure_Count"]
    .quantile(0.90)
)

patient_cost_threshold = (
    patient_procedure_summary["Total_Procedure_Cost"]
    .quantile(0.90)
)

patient_encounter_burden_threshold = (
    patient_procedure_summary["Procedures_Per_Encounter"]
    .quantile(0.90)
)

high_procedure_burden_patients = (
    patient_procedure_summary[
        (
            patient_procedure_summary["Procedure_Count"]
            >= patient_count_threshold
        )
        |
        (
            patient_procedure_summary["Total_Procedure_Cost"]
            >= patient_cost_threshold
        )
    ]
    .sort_values(
        [
            "Total_Procedure_Cost",
            "Procedure_Count"
        ],
        ascending=False
    )
    .copy()
)


# ============================================================================
# 25. PATIENT COST CONCENTRATION
# ============================================================================

def patient_concentration_share(df, n):

    total = df["Total_Procedure_Cost"].sum()

    if total <= 0:
        return 0

    return (
        df.nlargest(
            min(n, len(df)),
            "Total_Procedure_Cost"
        )["Total_Procedure_Cost"].sum()
        / total
        * 100
    )


patient_top1_cost_share = patient_concentration_share(
    patient_procedure_summary,
    1
)

patient_top5_cost_share = patient_concentration_share(
    patient_procedure_summary,
    5
)

patient_top10_cost_share = patient_concentration_share(
    patient_procedure_summary,
    10
)

patient_top20_cost_share = patient_concentration_share(
    patient_procedure_summary,
    20
)

patient_count = len(patient_procedure_summary)

top_1_percent_n = max(
    1,
    int(np.ceil(patient_count * 0.01))
)

top_5_percent_n = max(
    1,
    int(np.ceil(patient_count * 0.05))
)

top_10_percent_n = max(
    1,
    int(np.ceil(patient_count * 0.10))
)

top_20_percent_n = max(
    1,
    int(np.ceil(patient_count * 0.20))
)

patient_top1_percent_share = (
    patient_concentration_share(
        patient_procedure_summary,
        top_1_percent_n
    )
)

patient_top5_percent_share = (
    patient_concentration_share(
        patient_procedure_summary,
        top_5_percent_n
    )
)

patient_top10_percent_share = (
    patient_concentration_share(
        patient_procedure_summary,
        top_10_percent_n
    )
)

patient_top20_percent_share = (
    patient_concentration_share(
        patient_procedure_summary,
        top_20_percent_n
    )
)


# ============================================================================
# 26. EXECUTIVE ACTION MATRIX
# ============================================================================

def determine_primary_signal(row):

    if row["Executive_Priority"] == "CRITICAL_FINANCIAL_PRIORITY":
        return "CRITICAL_COST_AND_UTILIZATION"

    if row["Executive_Priority"] == "HIGH_FINANCIAL_PRIORITY":
        return "HIGH_FINANCIAL_EXPOSURE"

    if row["Executive_Priority"] == "HIGH_PATIENT_COST":
        return "HIGH_PATIENT_COST"

    if row["Executive_Priority"] == "HIGH_REPEAT_BURDEN":
        return "HIGH_REPEAT_UTILIZATION"

    if row["Executive_Priority"] == "HIGH_UTILIZATION_PRIORITY":
        return "HIGH_UTILIZATION"

    if row["Executive_Priority"] == "431_DOMINATED":
        return "431_COST_DEPENDENCY"

    if row["431_Dependency"] == "NON_431_DOMINATED":
        return "NON_431_FINANCIAL_EXPOSURE"

    return "STANDARD"


def determine_action(row):

    signal = row["Primary_Signal"]

    if signal == "CRITICAL_COST_AND_UTILIZATION":
        return (
            "Immediate clinical and financial review; "
            "assess utilization pattern, cost drivers and "
            "patient complexity."
        )

    if signal == "HIGH_FINANCIAL_EXPOSURE":
        return (
            "Prioritize financial and clinical pathway review."
        )

    if signal == "HIGH_PATIENT_COST":
        return (
            "Review patient-level cost concentration and "
            "clinical context."
        )

    if signal == "HIGH_REPEAT_UTILIZATION":
        return (
            "Review repeat utilization patterns and "
            "clinical pathway context."
        )

    if signal == "HIGH_UTILIZATION":
        return (
            "Review operational volume, workflow and "
            "utilization patterns."
        )

    if signal == "431_COST_DEPENDENCY":
        return (
            "Monitor activity and distinguish service volume "
            "from direct financial exposure."
        )

    if signal == "NON_431_FINANCIAL_EXPOSURE":
        return (
            "Monitor non-431 financial exposure and "
            "cost drivers."
        )

    return (
        "Routine monitoring."
    )


procedure_action_matrix = procedure_executive_scorecard[
    [
        "CODE",
        "CANONICAL_DESCRIPTION",
        "Procedure_Count",
        "Unique_Patients",
        "Unique_Encounters",
        "Total_Procedure_Cost",
        "Non_431_Total_Cost",
        "Cost_Per_Patient",
        "Procedures_Per_Patient",
        "Cost_431_Share",
        "431_Dependency",
        "Executive_Score",
        "Executive_Priority"
    ]
].copy()

procedure_action_matrix["Primary_Signal"] = (
    procedure_action_matrix.apply(
        determine_primary_signal,
        axis=1
    )
)

procedure_action_matrix["Financial_Exposure"] = np.where(
    procedure_action_matrix["Non_431_Total_Cost"]
    >= high_non431_cost_threshold,
    "HIGH",
    "STANDARD"
)

procedure_action_matrix["Utilization_Burden"] = np.where(
    procedure_action_matrix["Procedure_Count"]
    >= high_utilization_threshold,
    "HIGH",
    "STANDARD"
)

procedure_action_matrix["Patient_Burden"] = np.where(
    procedure_action_matrix["Cost_Per_Patient"]
    >= high_cost_per_patient_threshold,
    "HIGH",
    "STANDARD"
)

procedure_action_matrix["Recommended_Review"] = (
    procedure_action_matrix.apply(
        determine_action,
        axis=1
    )
)

procedure_action_matrix = (
    procedure_action_matrix
    .sort_values(
        [
            "Executive_Score",
            "Non_431_Total_Cost"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================================
# 27. PRIORITY DISTRIBUTION
# ============================================================================

priority_distribution = (
    procedure_summary["Executive_Priority"]
    .value_counts()
    .rename_axis("Executive_Priority")
    .reset_index(name="Procedure_Group_Count")
)

priority_distribution["Group_Share_%"] = (
    priority_distribution["Procedure_Group_Count"]
    / len(procedure_summary)
    * 100
)

priority_distribution = priority_distribution.sort_values(
    "Procedure_Group_Count",
    ascending=False
).reset_index(drop=True)


# ============================================================================
# 28. PROCEDURE INSIGHT KPIs
# ============================================================================

executive_procedure_kpis = pd.DataFrame(
    {
        "Metric": [
            "Total Procedure Records",
            "Unique Procedure Codes",
            "Unique Canonical Descriptions",
            "Unique Patients",
            "Unique Encounters",
            "Total Procedure Cost",
            "431 Cost",
            "Non-431 Cost",
            "431 Record Share %",
            "431 Cost Share %",
            "Non-431 Cost Share %",
            "Top 1 Total Cost Share %",
            "Top 5 Total Cost Share %",
            "Top 10 Total Cost Share %",
            "Top 20 Total Cost Share %",
            "Top 5 Non-431 Cost Share %",
            "Top 10 Non-431 Cost Share %",
            "Top 20 Non-431 Cost Share %",
            "Top 5 Volume Share %",
            "Top 10 Volume Share %",
            "Top 20 Volume Share %",
            "High Procedure Burden Patients",
            "Top 1% Patient Procedure Cost Share %",
            "Top 5% Patient Procedure Cost Share %",
            "Top 10% Patient Procedure Cost Share %",
            "Top 20% Patient Procedure Cost Share %",
            "431-Dominated Procedure Groups",
            "Mixed 431/Non-431 Procedure Groups",
            "Non-431-Dominated Procedure Groups",
        ],
        "Value": [
            len(work),
            unique_codes,
            unique_canonical,
            work["_PATIENT"].nunique(),
            work["_ENCOUNTER"].nunique(),
            TOTAL_COST,
            TOTAL_431_COST,
            TOTAL_NON_431_COST,
            RECORD_SHARE_431,
            COST_SHARE_431,
            COST_SHARE_NON_431,
            top1_cost_share,
            top5_cost_share,
            top10_cost_share,
            top20_cost_share,
            top5_non431_share,
            top10_non431_share,
            top20_non431_share,
            top5_volume_share,
            top10_volume_share,
            top20_volume_share,
            len(high_procedure_burden_patients),
            patient_top1_percent_share,
            patient_top5_percent_share,
            patient_top10_percent_share,
            patient_top20_percent_share,
            (
                procedure_summary["431_Dependency"]
                .eq("431_DOMINATED")
                .sum()
            ),
            (
                procedure_summary["431_Dependency"]
                .eq("MIXED_431_NON431")
                .sum()
            ),
            (
                procedure_summary["431_Dependency"]
                .eq("NON_431_DOMINATED")
                .sum()
            ),
        ]
    }
)


# ============================================================================
# 29. HUMAN-READABLE EXECUTIVE INSIGHT SUMMARY
# ============================================================================

procedure_insight_summary = pd.DataFrame(
    {
        "Insight_Area": [
            "Cost Concentration",
            "Non-431 Cost Concentration",
            "Volume Concentration",
            "431 Dependency",
            "High-Cost/High-Utilization",
            "High-Cost/Low-Utilization",
            "High-Utilization/Lower-Cost",
            "High Patient Cost",
            "High Repeat Utilization",
            "Patient Procedure Burden",
        ],
        "Finding": [
            (
                f"Top 10 procedures account for "
                f"{top10_cost_share:.2f}% of total observed procedure cost."
            ),
            (
                f"Top 10 procedures account for "
                f"{top10_non431_share:.2f}% of non-431 procedure cost."
            ),
            (
                f"Top 10 procedures account for "
                f"{top10_volume_share:.2f}% of procedure volume."
            ),
            (
                f"{RECORD_SHARE_431:.2f}% of records are BASE_COST=431, "
                f"while 431 represents only {COST_SHARE_431:.2f}% "
                f"of total procedure cost."
            ),
            (
                f"{len(high_cost_high_utilization_i)} procedure groups "
                f"meet both high-cost and high-utilization criteria."
            ),
            (
                f"{len(high_cost_low_utilization_i)} procedure groups "
                f"meet high-cost/lower-utilization criteria."
            ),
            (
                f"{len(high_utilization_low_cost_i)} procedure groups "
                f"meet high-utilization/lower-cost criteria."
            ),
            (
                f"{len(high_patient_cost_i)} procedure groups exceed "
                f"the high cost-per-patient threshold."
            ),
            (
                f"{len(high_repeat_procedures_i)} procedure groups exceed "
                f"the high repeat-utilization threshold."
            ),
            (
                f"{len(high_procedure_burden_patients):,} patients "
                f"meet the high procedure-burden criteria."
            ),
        ]
    }
)


# ============================================================================
# 30. TOP EXECUTIVE PRIORITY PROCEDURES
# ============================================================================

critical_procedure_priority = (
    procedure_executive_scorecard[
        procedure_executive_scorecard["Executive_Priority"]
        == "CRITICAL_FINANCIAL_PRIORITY"
    ]
    .copy()
)

high_financial_priority = (
    procedure_executive_scorecard[
        procedure_executive_scorecard["Executive_Priority"]
        == "HIGH_FINANCIAL_PRIORITY"
    ]
    .copy()
)


# ============================================================================
# 31. PRINT EXECUTIVE TABLES
# ============================================================================

display_columns = [
    "CODE",
    "CANONICAL_DESCRIPTION",
    "Procedure_Count",
    "Unique_Patients",
    "Total_Procedure_Cost",
    "Non_431_Total_Cost",
    "Cost_Per_Patient",
    "Procedures_Per_Patient",
    "Cost_431_Share",
    "Executive_Priority"
]

print("\nTOP EXECUTIVE PRIORITY PROCEDURES")
print("-" * 70)

print(
    procedure_executive_scorecard[
        display_columns
    ].head(20).to_string(index=False)
)


print("\nHIGH COST / HIGH UTILIZATION")
print("-" * 70)

print(
    high_cost_high_utilization_i[
        display_columns
    ].to_string(index=False)
)


print("\nHIGH COST / LOWER UTILIZATION")
print("-" * 70)

print(
    high_cost_low_utilization_i[
        display_columns
    ].head(20).to_string(index=False)
)


print("\nHIGH UTILIZATION / LOWER NON-431 COST")
print("-" * 70)

print(
    high_utilization_low_cost_i[
        display_columns
    ].head(20).to_string(index=False)
)


print("\nHIGH PATIENT-COST PROCEDURES")
print("-" * 70)

print(
    high_patient_cost_i[
        display_columns
    ].head(20).to_string(index=False)
)


print("\nHIGH REPEAT-UTILIZATION PROCEDURES")
print("-" * 70)

print(
    high_repeat_procedures_i[
        display_columns
    ].head(20).to_string(index=False)
)


# ============================================================================
# 32. 431 DEPENDENCY SUMMARY
# ============================================================================

dependency_counts = (
    procedure_summary["431_Dependency"]
    .value_counts()
)

print("\n431 COST DEPENDENCY")
print("-" * 70)

print(
    f"431-dominated procedures: "
    f"{dependency_counts.get('431_DOMINATED', 0):,}"
)

print(
    f"Mixed 431/non-431 procedures: "
    f"{dependency_counts.get('MIXED_431_NON431', 0):,}"
)

print(
    f"Non-431-dominated procedures: "
    f"{dependency_counts.get('NON_431_DOMINATED', 0):,}"
)


# ============================================================================
# 33. PATIENT BURDEN SUMMARY
# ============================================================================

print("\nPATIENT PROCEDURE BURDEN")
print("-" * 70)

print(
    f"Patients analyzed: "
    f"{len(patient_procedure_summary):,}"
)

print(
    f"High procedure-burden patients: "
    f"{len(high_procedure_burden_patients):,}"
)

print(
    f"Top 1% patient procedure-cost share: "
    f"{patient_top1_percent_share:.2f}%"
)

print(
    f"Top 5% patient procedure-cost share: "
    f"{patient_top5_percent_share:.2f}%"
)

print(
    f"Top 10% patient procedure-cost share: "
    f"{patient_top10_percent_share:.2f}%"
)

print(
    f"Top 20% patient procedure-cost share: "
    f"{patient_top20_percent_share:.2f}%"
)


# ============================================================================
# 34. KEY EXECUTIVE INSIGHTS
# ============================================================================

print("\nKEY EXECUTIVE INSIGHTS")
print("-" * 70)

print(
    f"1. Procedure expenditure is concentrated: "
    f"the top 10 procedures represent "
    f"{top10_cost_share:.2f}% of total observed procedure cost."
)

print(
    f"2. Non-431 procedures are the principal direct cost driver: "
    f"they represent {COST_SHARE_NON_431:.2f}% of total procedure cost."
)

print(
    f"3. A relatively small set of procedures combine material "
    f"financial exposure and utilization: "
    f"{len(high_cost_high_utilization_i)} groups meet both criteria."
)

print(
    f"4. {len(high_repeat_procedures_i)} procedure groups demonstrate "
    f"high repeat utilization based on the analytical threshold."
)

print(
    f"5. {len(high_patient_cost_i)} procedure groups demonstrate "
    f"high cost per patient."
)

print(
    f"6. BASE_COST=431 represents {RECORD_SHARE_431:.2f}% of procedure "
    f"records but only {COST_SHARE_431:.2f}% of procedure cost."
)

print(
    f"7. The top 10 procedures represent "
    f"{top10_volume_share:.2f}% of procedure volume."
)

print(
    f"8. Patient-level procedure costs are concentrated in the "
    f"highest-cost patient groups; this should be interpreted "
    f"alongside patient complexity and clinical context."
)


# ============================================================================
# 35. CLINICAL INTERPRETATION SAFEGUARD
# ============================================================================

print("\nCLINICAL INTERPRETATION SAFEGUARD")
print("-" * 70)

print(
    "No procedure has been classified as clinically inappropriate "
    "solely from cost or utilization data."
)

print(
    "High utilization does not equal inappropriate utilization."
)

print(
    "High cost does not equal inappropriate care."
)

print(
    "High repeat utilization does not equal unnecessary care."
)

print(
    "Findings represent analytical review priorities requiring "
    "clinical, operational and contextual validation."
)


# ============================================================================
# 36. FINAL VALIDATION
# ============================================================================

validation_results = {
    "Record count preserved":
        len(work) == SOURCE_RECORD_COUNT,

    "Original CODE preserved":
        ORIGINAL_CODE_COL in source_df.columns,

    "Original DESCRIPTION preserved":
        ORIGINAL_DESC_COL in source_df.columns,

    "Canonical description available":
        missing_canonical == 0
        and blank_canonical == 0,

    "One canonical description per CODE":
        len(multiple_canonical_codes) == 0,

    "431 cost segmentation available":
        "_IS_431" in work.columns,

    "Non-431 cost available":
        "Non_431_Total_Cost" in procedure_summary.columns,

    "Procedure executive scorecard created":
        isinstance(
            procedure_executive_scorecard,
            pd.DataFrame
        ),

    "Procedure priority created":
        "Executive_Priority"
        in procedure_executive_scorecard.columns,

    "Patient procedure summary created":
        isinstance(
            patient_procedure_summary,
            pd.DataFrame
        ),

    "Cost concentration calculated":
        "Cost_Concentration_Share"
        in procedure_summary.columns,

    "Volume concentration calculated":
        "Volume_Concentration_Share"
        in procedure_summary.columns,

    "Repeat utilization analysis created":
        isinstance(
            high_repeat_procedures_i,
            pd.DataFrame
        ),

    "Patient burden analysis created":
        isinstance(
            high_procedure_burden_patients,
            pd.DataFrame
        ),

    "Executive action matrix created":
        isinstance(
            procedure_action_matrix,
            pd.DataFrame
        ),

    "No records deleted":
        len(work) == len(source_df)
}

procedure_i_validation = pd.DataFrame(
    {
        "Validation": list(validation_results.keys()),
        "Result": list(validation_results.values())
    }
)

print("\nFINAL VALIDATION")
print("-" * 70)
print(
    procedure_i_validation.to_string(index=False)
)

validation_pass = all(
    validation_results.values()
)

if not validation_pass:
    raise ValueError(
        "STEP 4.18I validation FAILED. "
        "Review the validation table above."
    )


# ============================================================================
# 37. FINAL DATASET REGISTRY
# ============================================================================
#
# These are the principal analytical outputs of STEP 4.18I.
# ============================================================================

# Main dataset
procedure_i = work.copy()

# Executive datasets
procedure_executive_priority = (
    procedure_executive_scorecard.copy()
)

procedure_financial_concentration = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Total_Procedure_Cost",
            "Non_431_Total_Cost",
            "Cost_Concentration_Share",
            "Cumulative_Cost_Share",
            "Cost_Rank"
        ]
    ]
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_volume_concentration = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Volume_Concentration_Share",
            "Cumulative_Volume_Share",
            "Utilization_Rank"
        ]
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_cost_patient_analysis = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Total_Procedure_Cost",
            "Cost_Per_Patient",
            "Cost_Per_Encounter",
            "Procedures_Per_Patient"
        ]
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_utilization_burden = (
    procedure_summary[
        [
            "CODE",
            "CANONICAL_DESCRIPTION",
            "Procedure_Count",
            "Unique_Patients",
            "Unique_Encounters",
            "Procedures_Per_Patient",
            "Procedures_Per_Encounter",
            "Total_Procedure_Cost"
        ]
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_repeat_analysis = high_repeat_procedures_i.copy()


# ============================================================================
# 38. FINAL OUTPUT REPORT
# ============================================================================

print("\n" + "=" * 70)
print("STEP 4.18I COMPLETE")
print("=" * 70)

print(f"Total procedure records: {len(work):,}")
print(f"Unique procedure codes: {unique_codes:,}")
print(f"Unique canonical descriptions: {unique_canonical:,}")

print(f"Total observed procedure cost: {TOTAL_COST:,.2f}")
print(f"Total non-431 procedure cost: {TOTAL_NON_431_COST:,.2f}")

print(f"431 record share: {RECORD_SHARE_431:.2f}%")
print(f"431 cost share: {COST_SHARE_431:.2f}%")

print(
    f"Top 10 total-cost concentration: "
    f"{top10_cost_share:.2f}%"
)

print(
    f"Top 20 total-cost concentration: "
    f"{top20_cost_share:.2f}%"
)

print(
    f"Top 10 non-431 cost concentration: "
    f"{top10_non431_share:.2f}%"
)

print(
    f"Top 20 non-431 cost concentration: "
    f"{top20_non431_share:.2f}%"
)

print(
    f"Top 10 procedure volume concentration: "
    f"{top10_volume_share:.2f}%"
)

print(
    f"Top 20 procedure volume concentration: "
    f"{top20_volume_share:.2f}%"
)

print(
    f"Critical financial priority groups: "
    f"{len(critical_procedure_priority):,}"
)

print(
    f"High financial priority groups: "
    f"{len(high_financial_priority):,}"
)

print(
    f"High patient-cost groups: "
    f"{len(high_patient_cost_i):,}"
)

print(
    f"High repeat-utilization groups: "
    f"{len(high_repeat_procedures_i):,}"
)

print(
    f"431-dominated groups: "
    f"{dependency_counts.get('431_DOMINATED', 0):,}"
)

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_i")
print("  - procedure_executive_scorecard")
print("  - procedure_executive_priority")
print("  - procedure_financial_concentration")
print("  - procedure_volume_concentration")
print("  - procedure_cost_patient_analysis")
print("  - procedure_utilization_burden")
print("  - procedure_repeat_analysis")
print("  - procedure_431_dependency")
print("  - procedure_action_matrix")
print("  - patient_procedure_summary")
print("  - high_procedure_burden_patients")
print("  - critical_procedure_priority")
print("  - high_financial_priority")
print("  - high_cost_high_utilization_i")
print("  - high_cost_low_utilization_i")
print("  - high_utilization_low_cost_i")
print("  - high_patient_cost_i")
print("  - high_repeat_procedures_i")
print("  - high_cost_per_patient")
print("  - procedure_431_dependency")
print("  - priority_distribution")
print("  - executive_procedure_kpis")
print("  - procedure_insight_summary")
print("  - procedure_i_validation")

print("\nSTATUS: PASS");
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18I — PROCEDURE-LEVEL EXECUTIVE / CLINICAL INSIGHT ENGINE
#              COST • UTILIZATION • PATIENT BURDEN • REPEAT USE
#              OUTLIERS • CONCENTRATION • 431/NON-431 • PRIORITIZATION
# ======================================================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.width", 240)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")


# ======================================================================
# 1. SOURCE DATA DISCOVERY
# ======================================================================

print("=" * 70)
print("STEP 4.18I — PROCEDURE-LEVEL EXECUTIVE / CLINICAL INSIGHT ENGINE")
print("=" * 70)

# Use existing Step 4.18H analytical dataset
if "procedure_h" in globals():
    source_df = procedure_h.copy()
    source_name = "procedure_h"

elif "procedure_g" in globals():
    source_df = procedure_g.copy()
    source_name = "procedure_g"

else:
    raise ValueError(
        "Neither 'procedure_h' nor 'procedure_g' exists. "
        "Run Step 4.18H or Step 4.18G first."
    )

print("\nSOURCE DATA DISCOVERY")
print("-" * 70)
print(f"Using source dataset: {source_name}")
print(f"Source records: {len(source_df):,}")


# ======================================================================
# 2. SOURCE COLUMN VALIDATION
# ======================================================================

required_columns = [
    "CODE",
    "DESCRIPTION",
    "CANONICAL_DESCRIPTION",
    "BASE_COST",
    "PATIENT",
    "ENCOUNTER"
]

missing_columns = [
    col for col in required_columns
    if col not in source_df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("\nSOURCE DATA CHECK")
print("-" * 70)
print("STATUS: PASS")
print(f"Procedure records available: {len(source_df):,}")


# ======================================================================
# 3. RECORD PRESERVATION
# ======================================================================

original_record_count = len(source_df)

procedure_i = source_df.copy()

if len(procedure_i) != original_record_count:
    raise ValueError("Record count changed during source preparation.")

print("\nRECORD PRESERVATION")
print("-" * 70)
print(f"Source records: {original_record_count:,}")
print(f"Analytical records: {len(procedure_i):,}")
print(f"Record count preserved: {len(procedure_i) == original_record_count}")


# ======================================================================
# 4. ORIGINAL VALUE PRESERVATION
# ======================================================================

procedure_i["ORIGINAL_CODE"] = procedure_i["CODE"]
procedure_i["ORIGINAL_DESCRIPTION"] = procedure_i["DESCRIPTION"]

print("\nORIGINAL VALUE PRESERVATION")
print("-" * 70)
print(
    f"Original CODE preserved: "
    f"{procedure_i['ORIGINAL_CODE'].equals(procedure_i['CODE'])}"
)

print(
    f"Original DESCRIPTION preserved: "
    f"{procedure_i['ORIGINAL_DESCRIPTION'].equals(procedure_i['DESCRIPTION'])}"
)


# ======================================================================
# 5. DATA TYPE STANDARDIZATION
# ======================================================================

procedure_i["CODE"] = procedure_i["CODE"].astype(str).str.strip()

procedure_i["CANONICAL_DESCRIPTION"] = (
    procedure_i["CANONICAL_DESCRIPTION"]
    .astype(str)
    .str.strip()
)

procedure_i["PATIENT"] = procedure_i["PATIENT"].astype(str).str.strip()
procedure_i["ENCOUNTER"] = procedure_i["ENCOUNTER"].astype(str).str.strip()

procedure_i["BASE_COST"] = pd.to_numeric(
    procedure_i["BASE_COST"],
    errors="coerce"
)

if procedure_i["BASE_COST"].isna().any():
    raise ValueError(
        "BASE_COST contains missing or non-numeric values."
    )


# ======================================================================
# 6. BASE COST SEGMENTATION
# ======================================================================

procedure_i["IS_431"] = (
    procedure_i["BASE_COST"] == 431
)

procedure_i["IS_NON_431"] = (
    procedure_i["BASE_COST"] != 431
)

procedure_i["NON_431_COST"] = np.where(
    procedure_i["IS_NON_431"],
    procedure_i["BASE_COST"],
    0
)

total_records = len(procedure_i)

records_431 = int(procedure_i["IS_431"].sum())
records_non_431 = int(procedure_i["IS_NON_431"].sum())

total_cost = procedure_i["BASE_COST"].sum()
cost_431 = procedure_i.loc[
    procedure_i["IS_431"], "BASE_COST"
].sum()

non_431_cost = procedure_i["NON_431_COST"].sum()

print("\nBASE_COST SEGMENTATION")
print("-" * 70)
print(f"431 records: {records_431:,}")
print(f"Non-431 records: {records_non_431:,}")
print(f"431 record share: {(records_431 / total_records) * 100:.2f}%")
print(f"Total procedure cost: {total_cost:,.2f}")
print(f"431 cost: {cost_431:,.2f}")
print(f"Non-431 cost: {non_431_cost:,.2f}")
print(f"431 cost share: {(cost_431 / total_cost) * 100:.2f}%")
print(f"Non-431 cost share: {(non_431_cost / total_cost) * 100:.2f}%")


# ======================================================================
# 7. PROCEDURE-LEVEL EXECUTIVE SUMMARY
# ======================================================================

procedure_summary = (
    procedure_i
    .groupby(
        ["CODE", "CANONICAL_DESCRIPTION"],
        dropna=False
    )
    .agg(
        Procedure_Count=("CODE", "size"),
        Unique_Patients=("PATIENT", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Non_431_Total_Cost=("NON_431_COST", "sum"),
        Cost_431=("IS_431", "sum")
    )
    .reset_index()
)

procedure_summary["Average_Procedure_Cost"] = (
    procedure_summary["Total_Procedure_Cost"]
    / procedure_summary["Procedure_Count"]
)

procedure_summary["Cost_Per_Patient"] = np.where(
    procedure_summary["Unique_Patients"] > 0,
    procedure_summary["Total_Procedure_Cost"]
    / procedure_summary["Unique_Patients"],
    0
)

procedure_summary["Non_431_Cost_Per_Patient"] = np.where(
    procedure_summary["Unique_Patients"] > 0,
    procedure_summary["Non_431_Total_Cost"]
    / procedure_summary["Unique_Patients"],
    0
)

procedure_summary["Procedures_Per_Patient"] = np.where(
    procedure_summary["Unique_Patients"] > 0,
    procedure_summary["Procedure_Count"]
    / procedure_summary["Unique_Patients"],
    0
)

procedure_summary["Procedures_Per_Encounter"] = np.where(
    procedure_summary["Unique_Encounters"] > 0,
    procedure_summary["Procedure_Count"]
    / procedure_summary["Unique_Encounters"],
    0
)

procedure_summary["Cost_431_Share"] = np.where(
    procedure_summary["Total_Procedure_Cost"] > 0,
    (
        procedure_summary["Cost_431"]
        * 431
        / procedure_summary["Total_Procedure_Cost"]
    ) * 100,
    0
)

procedure_summary["Non_431_Cost_Share"] = np.where(
    procedure_summary["Total_Procedure_Cost"] > 0,
    (
        procedure_summary["Non_431_Total_Cost"]
        / procedure_summary["Total_Procedure_Cost"]
    ) * 100,
    0
)


# ======================================================================
# 8. COST CONCENTRATION
# ======================================================================

cost_ranked = procedure_summary.sort_values(
    "Total_Procedure_Cost",
    ascending=False
).reset_index(drop=True)

cost_ranked["Cumulative_Cost"] = (
    cost_ranked["Total_Procedure_Cost"].cumsum()
)

cost_ranked["Cumulative_Cost_Share"] = (
    cost_ranked["Cumulative_Cost"]
    / total_cost
) * 100

top_5_cost_share = (
    cost_ranked.head(5)["Total_Procedure_Cost"].sum()
    / total_cost
) * 100

top_10_cost_share = (
    cost_ranked.head(10)["Total_Procedure_Cost"].sum()
    / total_cost
) * 100

top_20_cost_share = (
    cost_ranked.head(20)["Total_Procedure_Cost"].sum()
    / total_cost
) * 100


# ======================================================================
# 9. NON-431 COST CONCENTRATION
# ======================================================================

non431_ranked = procedure_summary.sort_values(
    "Non_431_Total_Cost",
    ascending=False
).reset_index(drop=True)

non431_ranked["Cumulative_Non_431_Cost"] = (
    non431_ranked["Non_431_Total_Cost"].cumsum()
)

non431_ranked["Cumulative_Non_431_Cost_Share"] = np.where(
    non_431_cost > 0,
    (
        non431_ranked["Cumulative_Non_431_Cost"]
        / non_431_cost
    ) * 100,
    0
)

top_5_non431_share = (
    non431_ranked.head(5)["Non_431_Total_Cost"].sum()
    / non_431_cost
) * 100

top_10_non431_share = (
    non431_ranked.head(10)["Non_431_Total_Cost"].sum()
    / non_431_cost
) * 100

top_20_non431_share = (
    non431_ranked.head(20)["Non_431_Total_Cost"].sum()
    / non_431_cost
) * 100


# ======================================================================
# 10. UTILIZATION CONCENTRATION
# ======================================================================

util_ranked = procedure_summary.sort_values(
    "Procedure_Count",
    ascending=False
).reset_index(drop=True)

util_ranked["Cumulative_Volume"] = (
    util_ranked["Procedure_Count"].cumsum()
)

util_ranked["Cumulative_Volume_Share"] = (
    util_ranked["Cumulative_Volume"]
    / total_records
) * 100

top_5_volume_share = (
    util_ranked.head(5)["Procedure_Count"].sum()
    / total_records
) * 100

top_10_volume_share = (
    util_ranked.head(10)["Procedure_Count"].sum()
    / total_records
) * 100

top_20_volume_share = (
    util_ranked.head(20)["Procedure_Count"].sum()
    / total_records
) * 100


# ======================================================================
# 11. PATIENT-LEVEL PROCEDURE BURDEN
# ======================================================================

patient_procedure_summary = (
    procedure_i
    .groupby("PATIENT")
    .agg(
        Total_Procedures=("CODE", "size"),
        Unique_Procedure_Types=("CODE", "nunique"),
        Unique_Encounters=("ENCOUNTER", "nunique"),
        Total_Procedure_Cost=("BASE_COST", "sum"),
        Non_431_Procedure_Cost=("NON_431_COST", "sum"),
        Procedure_431_Count=("IS_431", "sum")
    )
    .reset_index()
)

patient_procedure_summary["Procedures_Per_Encounter"] = np.where(
    patient_procedure_summary["Unique_Encounters"] > 0,
    patient_procedure_summary["Total_Procedures"]
    / patient_procedure_summary["Unique_Encounters"],
    0
)

patient_procedure_summary["Non_431_Cost_Share"] = np.where(
    patient_procedure_summary["Total_Procedure_Cost"] > 0,
    (
        patient_procedure_summary["Non_431_Procedure_Cost"]
        / patient_procedure_summary["Total_Procedure_Cost"]
    ) * 100,
    0
)


# ======================================================================
# 12. PATIENT BURDEN THRESHOLDS
# ======================================================================

burden_procedure_threshold = (
    patient_procedure_summary["Total_Procedures"].quantile(0.95)
)

burden_cost_threshold = (
    patient_procedure_summary["Total_Procedure_Cost"].quantile(0.95)
)

high_procedure_burden_patients = (
    patient_procedure_summary[
        (
            patient_procedure_summary["Total_Procedures"]
            >= burden_procedure_threshold
        )
        |
        (
            patient_procedure_summary["Total_Procedure_Cost"]
            >= burden_cost_threshold
        )
    ]
    .sort_values(
        ["Total_Procedure_Cost", "Total_Procedures"],
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 13. REPEAT UTILIZATION
# ======================================================================

procedure_summary["Repeat_Utilization_Flag"] = (
    procedure_summary["Procedures_Per_Patient"]
    > 1
)

repeat_ranked = procedure_summary.sort_values(
    "Procedures_Per_Patient",
    ascending=False
).reset_index(drop=True)

high_repeat_threshold = (
    procedure_summary["Procedures_Per_Patient"]
    .quantile(0.90)
)

high_repeat_utilization = (
    procedure_summary[
        procedure_summary["Procedures_Per_Patient"]
        >= high_repeat_threshold
    ]
    .sort_values(
        "Procedures_Per_Patient",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 14. COST-PER-PATIENT OUTLIERS
# ======================================================================

cost_per_patient_threshold = (
    procedure_summary["Cost_Per_Patient"]
    .quantile(0.90)
)

high_cost_per_patient = (
    procedure_summary[
        procedure_summary["Cost_Per_Patient"]
        >= cost_per_patient_threshold
    ]
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 15. NON-431 COST OUTLIERS
# ======================================================================

non431_cost_threshold = (
    procedure_summary["Non_431_Total_Cost"]
    .quantile(0.90)
)

high_cost_non431_procedures = (
    procedure_summary[
        procedure_summary["Non_431_Total_Cost"]
        >= non431_cost_threshold
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 16. 431-DOMINATED PROCEDURES
# ======================================================================

procedures_431_dominated = (
    procedure_summary[
        procedure_summary["Cost_431_Share"] >= 90
    ]
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 17. HIGH VOLUME / LOW COST
# ======================================================================

utilization_threshold = (
    procedure_summary["Procedure_Count"]
    .quantile(0.90)
)

high_utilization_low_cost = (
    procedure_summary[
        (
            procedure_summary["Procedure_Count"]
            >= utilization_threshold
        )
        &
        (
            procedure_summary["Non_431_Total_Cost"]
            < non431_cost_threshold
        )
    ]
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 18. HIGH COST / LOW UTILIZATION
# ======================================================================

high_cost_low_utilization = (
    procedure_summary[
        (
            procedure_summary["Non_431_Total_Cost"]
            >= non431_cost_threshold
        )
        &
        (
            procedure_summary["Procedure_Count"]
            < utilization_threshold
        )
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 19. HIGH COST / HIGH UTILIZATION
# ======================================================================

high_cost_high_utilization = (
    procedure_summary[
        (
            procedure_summary["Non_431_Total_Cost"]
            >= non431_cost_threshold
        )
        &
        (
            procedure_summary["Procedure_Count"]
            >= utilization_threshold
        )
    ]
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)


# ======================================================================
# 20. PROCEDURE BURDEN / COST EFFICIENCY PROFILE
# ======================================================================

procedure_summary["Cost_Burden_Index"] = np.where(
    total_cost > 0,
    (
        procedure_summary["Total_Procedure_Cost"]
        / total_cost
    ) * 100,
    0
)

procedure_summary["Volume_Burden_Index"] = (
    procedure_summary["Procedure_Count"]
    / total_records
) * 100

procedure_summary["Cost_vs_Volume_Ratio"] = np.where(
    procedure_summary["Volume_Burden_Index"] > 0,
    procedure_summary["Cost_Burden_Index"]
    / procedure_summary["Volume_Burden_Index"],
    0
)


# ======================================================================
# 21. COST/VOLUME IMBALANCE FLAGS
# ======================================================================

procedure_summary["Cost_Volume_Profile"] = np.select(
    [
        (
            procedure_summary["Cost_Burden_Index"]
            >= procedure_summary["Cost_Burden_Index"].quantile(0.75)
        )
        &
        (
            procedure_summary["Volume_Burden_Index"]
            >= procedure_summary["Volume_Burden_Index"].quantile(0.75)
        ),

        (
            procedure_summary["Cost_Burden_Index"]
            >= procedure_summary["Cost_Burden_Index"].quantile(0.75)
        )
        &
        (
            procedure_summary["Volume_Burden_Index"]
            < procedure_summary["Volume_Burden_Index"].quantile(0.75)
        ),

        (
            procedure_summary["Volume_Burden_Index"]
            >= procedure_summary["Volume_Burden_Index"].quantile(0.75)
        )
        &
        (
            procedure_summary["Cost_Burden_Index"]
            < procedure_summary["Cost_Burden_Index"].quantile(0.75)
        )
    ],
    [
        "HIGH_COST_HIGH_VOLUME",
        "HIGH_COST_LOW_VOLUME",
        "HIGH_VOLUME_LOW_COST"
    ],
    default="STANDARD"
)


# ======================================================================
# 22. EXECUTIVE PRIORITY SCORE
# ======================================================================

def percentile_score(series):
    return series.rank(pct=True) * 100


procedure_summary["Cost_Priority_Score"] = percentile_score(
    procedure_summary["Non_431_Total_Cost"]
)

procedure_summary["Volume_Priority_Score"] = percentile_score(
    procedure_summary["Procedure_Count"]
)

procedure_summary["Patient_Cost_Score"] = percentile_score(
    procedure_summary["Cost_Per_Patient"]
)

procedure_summary["Repeat_Use_Score"] = percentile_score(
    procedure_summary["Procedures_Per_Patient"]
)

procedure_summary["Priority_Score"] = (
    procedure_summary["Cost_Priority_Score"] * 0.35
    +
    procedure_summary["Volume_Priority_Score"] * 0.25
    +
    procedure_summary["Patient_Cost_Score"] * 0.20
    +
    procedure_summary["Repeat_Use_Score"] * 0.20
)


# ======================================================================
# 23. EXECUTIVE PRIORITY BANDS
# ======================================================================

procedure_summary["Executive_Priority"] = pd.cut(
    procedure_summary["Priority_Score"],
    bins=[-np.inf, 40, 60, 80, np.inf],
    labels=[
        "STANDARD",
        "MONITOR",
        "HIGH PRIORITY",
        "CRITICAL PRIORITY"
    ]
)


# ======================================================================
# 24. CLINICAL / OPERATIONAL INSIGHT FLAGS
# ======================================================================

procedure_summary["Insight_Flags"] = ""

def add_flag(df, condition, flag):
    df.loc[
        condition,
        "Insight_Flags"
    ] = np.where(
        df.loc[condition, "Insight_Flags"].eq(""),
        flag,
        df.loc[condition, "Insight_Flags"] + " | " + flag
    )


add_flag(
    procedure_summary,
    procedure_summary["Cost_Burden_Index"]
    >= procedure_summary["Cost_Burden_Index"].quantile(0.90),
    "TOP_COST_DRIVER"
)

add_flag(
    procedure_summary,
    procedure_summary["Volume_Burden_Index"]
    >= procedure_summary["Volume_Burden_Index"].quantile(0.90),
    "TOP_VOLUME_DRIVER"
)

add_flag(
    procedure_summary,
    procedure_summary["Cost_Per_Patient"]
    >= cost_per_patient_threshold,
    "HIGH_PATIENT_COST"
)

add_flag(
    procedure_summary,
    procedure_summary["Procedures_Per_Patient"]
    >= high_repeat_threshold,
    "HIGH_REPEAT_USE"
)

add_flag(
    procedure_summary,
    procedure_summary["Cost_431_Share"] >= 90,
    "431_DOMINATED"
)

add_flag(
    procedure_summary,
    (
        procedure_summary["Non_431_Total_Cost"]
        >= non431_cost_threshold
    )
    &
    (
        procedure_summary["Procedure_Count"]
        < utilization_threshold
    ),
    "HIGH_COST_LOW_VOLUME"
)

add_flag(
    procedure_summary,
    (
        procedure_summary["Procedure_Count"]
        >= utilization_threshold
    )
    &
    (
        procedure_summary["Non_431_Total_Cost"]
        < non431_cost_threshold
    ),
    "HIGH_VOLUME_LOW_COST"
)


# ======================================================================
# 25. PROCEDURE PRIORITY DATASET
# ======================================================================

procedure_priority = (
    procedure_summary
    .sort_values(
        [
            "Priority_Score",
            "Non_431_Total_Cost",
            "Procedure_Count"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_priority["Priority_Rank"] = (
    procedure_priority.index + 1
)


# ======================================================================
# 26. TOP EXECUTIVE PROCEDURES
# ======================================================================

top_executive_procedures = (
    procedure_priority
    .head(25)
    .copy()
)


# ======================================================================
# 27. EXECUTIVE INSIGHT METRICS
# ======================================================================

priority_counts = (
    procedure_summary["Executive_Priority"]
    .value_counts()
)

insight_metrics = pd.DataFrame({
    "Metric": [
        "Total Procedure Records",
        "Unique Procedure Codes",
        "Unique Canonical Procedures",
        "Total Procedure Cost",
        "431 Record Share",
        "431 Cost Share",
        "Non-431 Cost Share",
        "Top 5 Total Cost Concentration",
        "Top 10 Total Cost Concentration",
        "Top 20 Total Cost Concentration",
        "Top 5 Non-431 Cost Concentration",
        "Top 10 Non-431 Cost Concentration",
        "Top 20 Non-431 Cost Concentration",
        "Top 5 Procedure Volume Concentration",
        "Top 10 Procedure Volume Concentration",
        "Top 20 Procedure Volume Concentration",
        "High Repeat Utilization Procedures",
        "High Cost Per Patient Procedures",
        "431-Dominated Procedures",
        "High Cost / High Utilization Procedures",
        "High Cost / Low Utilization Procedures",
        "High Utilization / Lower Cost Procedures",
        "High Procedure-Burden Patients",
        "Critical Priority Procedures",
        "High Priority Procedures"
    ],
    "Value": [
        total_records,
        procedure_summary["CODE"].nunique(),
        procedure_summary["CANONICAL_DESCRIPTION"].nunique(),
        total_cost,
        (records_431 / total_records) * 100,
        (cost_431 / total_cost) * 100,
        (non_431_cost / total_cost) * 100,
        top_5_cost_share,
        top_10_cost_share,
        top_20_cost_share,
        top_5_non431_share,
        top_10_non431_share,
        top_20_non431_share,
        top_5_volume_share,
        top_10_volume_share,
        top_20_volume_share,
        len(high_repeat_utilization),
        len(high_cost_per_patient),
        len(procedures_431_dominated),
        len(high_cost_high_utilization),
        len(high_cost_low_utilization),
        len(high_utilization_low_cost),
        len(high_procedure_burden_patients),
        int(priority_counts.get("CRITICAL PRIORITY", 0)),
        int(priority_counts.get("HIGH PRIORITY", 0))
    ]
})


# ======================================================================
# 28. AUTOMATED EXECUTIVE INSIGHT SUMMARY
# ======================================================================

executive_insights = []

# Cost concentration
if top_10_non431_share >= 70:
    executive_insights.append(
        f"Non-431 procedure expenditure is highly concentrated: "
        f"the top 10 procedures account for {top_10_non431_share:.2f}% "
        f"of non-431 procedure cost."
    )
else:
    executive_insights.append(
        f"Non-431 procedure expenditure is moderately distributed: "
        f"the top 10 procedures account for {top_10_non431_share:.2f}% "
        f"of non-431 procedure cost."
    )

# Volume concentration
if top_10_volume_share >= 60:
    executive_insights.append(
        f"Procedure utilization is concentrated: the top 10 procedure "
        f"types account for {top_10_volume_share:.2f}% of all procedure records."
    )

# 431
if (cost_431 / total_cost) < (records_431 / total_records):
    executive_insights.append(
        "BASE_COST = 431 procedures represent a larger share of records "
        "than of total cost, indicating that this segment is relatively "
        "low-cost per observed procedure."
    )

# High cost/high utilization
if len(high_cost_high_utilization) > 0:
    executive_insights.append(
        f"{len(high_cost_high_utilization)} procedure groups meet both "
        "high-cost and high-utilization criteria and should receive "
        "priority operational and clinical review."
    )

# Repeat utilization
if len(high_repeat_utilization) > 0:
    executive_insights.append(
        f"{len(high_repeat_utilization)} procedure groups show unusually "
        "high repeat utilization relative to the procedure population."
    )

# Patient burden
if len(high_procedure_burden_patients) > 0:
    executive_insights.append(
        f"{len(high_procedure_burden_patients)} patients fall into the "
        "high procedure-burden segment based on the 95th-percentile "
        "procedure-volume or cost threshold."
    )

# 431 dominated
if len(procedures_431_dominated) > 0:
    executive_insights.append(
        f"{len(procedures_431_dominated)} procedure groups are "
        "431-cost dominated and should be interpreted separately from "
        "non-431 expenditure drivers."
    )

procedure_insight_summary = pd.DataFrame({
    "Insight_ID": range(1, len(executive_insights) + 1),
    "Executive_Insight": executive_insights
})


# ======================================================================
# 29. CLINICAL / OPERATIONAL WATCHLIST
# ======================================================================

procedure_watchlist = (
    procedure_summary[
        procedure_summary["Insight_Flags"] != ""
    ]
    .sort_values(
        [
            "Priority_Score",
            "Non_431_Total_Cost",
            "Procedure_Count"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

procedure_watchlist["Watchlist_Rank"] = (
    procedure_watchlist.index + 1
)


# ======================================================================
# 30. TOP COST PROCEDURES
# ======================================================================

top_cost_procedures = (
    procedure_summary
    .sort_values(
        "Total_Procedure_Cost",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


# ======================================================================
# 31. TOP NON-431 COST
# ======================================================================

top_non_431_cost = (
    procedure_summary
    .sort_values(
        "Non_431_Total_Cost",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


# ======================================================================
# 32. TOP UTILIZATION
# ======================================================================

top_utilization = (
    procedure_summary
    .sort_values(
        "Procedure_Count",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


# ======================================================================
# 33. COST PER PATIENT
# ======================================================================

top_cost_per_patient = (
    procedure_summary
    .sort_values(
        "Cost_Per_Patient",
        ascending=False
    )
    .head(20)
    .reset_index(drop=True)
)


# ======================================================================
# 34. COST INTENSITY DISTRIBUTION
# ======================================================================

procedure_summary["Cost_Intensity"] = pd.cut(
    procedure_summary["Non_431_Total_Cost"],
    bins=[
        -np.inf,
        0,
        procedure_summary["Non_431_Total_Cost"].quantile(0.50),
        procedure_summary["Non_431_Total_Cost"].quantile(0.75),
        procedure_summary["Non_431_Total_Cost"].quantile(0.90),
        np.inf
    ],
    labels=[
        "431_DOMINATED",
        "STANDARD_NON_431_COST",
        "MODERATE_NON_431_COST",
        "HIGH_NON_431_COST",
        "EXTREME_NON_431_COST"
    ],
    include_lowest=True
)

cost_intensity_distribution = (
    procedure_summary["Cost_Intensity"]
    .value_counts(dropna=False)
    .rename_axis("Cost_Intensity")
    .reset_index(name="Procedure_Group_Count")
)

cost_intensity_distribution["Group_Share_%"] = (
    cost_intensity_distribution["Procedure_Group_Count"]
    / len(procedure_summary)
) * 100


# ======================================================================
# 35. UTILIZATION INTENSITY
# ======================================================================

procedure_summary["Utilization_Intensity"] = pd.cut(
    procedure_summary["Procedure_Count"],
    bins=[
        -np.inf,
        procedure_summary["Procedure_Count"].quantile(0.50),
        procedure_summary["Procedure_Count"].quantile(0.75),
        procedure_summary["Procedure_Count"].quantile(0.90),
        np.inf
    ],
    labels=[
        "STANDARD_UTILIZATION",
        "MODERATE_UTILIZATION",
        "HIGH_UTILIZATION",
        "EXTREME_UTILIZATION"
    ],
    include_lowest=True
)

utilization_intensity_distribution = (
    procedure_summary["Utilization_Intensity"]
    .value_counts(dropna=False)
    .rename_axis("Utilization_Intensity")
    .reset_index(name="Procedure_Group_Count")
)

utilization_intensity_distribution["Group_Share_%"] = (
    utilization_intensity_distribution["Procedure_Group_Count"]
    / len(procedure_summary)
) * 100


# ======================================================================
# 36. EXECUTIVE PRIORITY DISTRIBUTION
# ======================================================================

priority_distribution = (
    procedure_summary["Executive_Priority"]
    .value_counts()
    .rename_axis("Executive_Priority")
    .reset_index(name="Procedure_Group_Count")
)

priority_distribution["Group_Share_%"] = (
    priority_distribution["Procedure_Group_Count"]
    / len(procedure_summary)
) * 100


# ======================================================================
# 37. CONCENTRATION DATASET
# ======================================================================

procedure_concentration = procedure_summary[
    [
        "CODE",
        "CANONICAL_DESCRIPTION",
        "Procedure_Count",
        "Total_Procedure_Cost",
        "Non_431_Total_Cost"
    ]
].copy()

procedure_concentration = procedure_concentration.sort_values(
    "Total_Procedure_Cost",
    ascending=False
).reset_index(drop=True)

procedure_concentration["Cost_Rank"] = (
    procedure_concentration.index + 1
)

procedure_concentration["Cumulative_Cost"] = (
    procedure_concentration["Total_Procedure_Cost"].cumsum()
)

procedure_concentration["Cumulative_Cost_Share"] = (
    procedure_concentration["Cumulative_Cost"]
    / total_cost
) * 100

procedure_concentration["Cumulative_Volume"] = (
    procedure_concentration["Procedure_Count"].cumsum()
)

procedure_concentration["Cumulative_Volume_Share"] = (
    procedure_concentration["Cumulative_Volume"]
    / total_records
) * 100


# ======================================================================
# 38. FINAL VALIDATION
# ======================================================================

validation_results = {
    "Record count preserved":
        len(procedure_i) == original_record_count,

    "Original CODE preserved":
        procedure_i["ORIGINAL_CODE"].equals(
            source_df["CODE"]
        ),

    "Original DESCRIPTION preserved":
        procedure_i["ORIGINAL_DESCRIPTION"].equals(
            source_df["DESCRIPTION"]
        ),

    "Canonical descriptions available":
        procedure_i["CANONICAL_DESCRIPTION"]
        .notna()
        .all(),

    "No blank canonical descriptions":
        ~procedure_i["CANONICAL_DESCRIPTION"]
        .astype(str)
        .str.strip()
        .eq("")
        .any(),

    "One canonical description per CODE":
        procedure_i.groupby("CODE")[
            "CANONICAL_DESCRIPTION"
        ].nunique().max() == 1,

    "431 segmentation available":
        procedure_i["IS_431"].notna().all(),

    "Non-431 cost available":
        procedure_i["NON_431_COST"].notna().all(),

    "Procedure summary created":
        len(procedure_summary) > 0,

    "Patient procedure summary created":
        len(patient_procedure_summary) > 0,

    "Cost concentration calculated":
        len(procedure_concentration) > 0,

    "Priority score created":
        procedure_summary["Priority_Score"].notna().all(),

    "Executive priority created":
        procedure_summary["Executive_Priority"].notna().all(),

    "No records deleted":
        len(procedure_i) == original_record_count
}

procedure_i_validation = pd.DataFrame(
    list(validation_results.items()),
    columns=["Validation", "Result"]
)

print("\nFINAL VALIDATION")
print("-" * 70)
print(procedure_i_validation.to_string(index=False))


if not procedure_i_validation["Result"].all():
    raise ValueError(
        "STEP 4.18I validation failed. "
        "Review procedure_i_validation."
    )


# ======================================================================
# 39. FINAL EXECUTIVE OUTPUT
# ======================================================================

print("\n")
print("=" * 70)
print("STEP 4.18I COMPLETE")
print("=" * 70)

print(f"Total procedure records: {len(procedure_i):,}")
print(
    f"Unique procedure codes: "
    f"{procedure_summary['CODE'].nunique():,}"
)
print(
    f"Unique canonical descriptions: "
    f"{procedure_summary['CANONICAL_DESCRIPTION'].nunique():,}"
)

print(f"Total observed procedure cost: {total_cost:,.2f}")
print(f"Total non-431 procedure cost: {non_431_cost:,.2f}")

print(
    f"431 record share: "
    f"{(records_431 / total_records) * 100:.2f}%"
)

print(
    f"431 cost share: "
    f"{(cost_431 / total_cost) * 100:.2f}%"
)

print(
    f"Top 10 total-cost concentration: "
    f"{top_10_cost_share:.2f}%"
)

print(
    f"Top 20 total-cost concentration: "
    f"{top_20_cost_share:.2f}%"
)

print(
    f"Top 10 non-431 cost concentration: "
    f"{top_10_non431_share:.2f}%"
)

print(
    f"Top 20 non-431 cost concentration: "
    f"{top_20_non431_share:.2f}%"
)

print(
    f"Top 10 procedure-volume concentration: "
    f"{top_10_volume_share:.2f}%"
)

print(
    f"Top 20 procedure-volume concentration: "
    f"{top_20_volume_share:.2f}%"
)

print(
    f"High repeat-utilization procedures: "
    f"{len(high_repeat_utilization):,}"
)

print(
    f"High cost-per-patient procedures: "
    f"{len(high_cost_per_patient):,}"
)

print(
    f"High cost/high utilization procedures: "
    f"{len(high_cost_high_utilization):,}"
)

print(
    f"High cost/lower utilization procedures: "
    f"{len(high_cost_low_utilization):,}"
)

print(
    f"High utilization/lower cost procedures: "
    f"{len(high_utilization_low_cost):,}"
)

print(
    f"431-dominated procedures: "
    f"{len(procedures_431_dominated):,}"
)

print(
    f"High procedure-burden patients: "
    f"{len(high_procedure_burden_patients):,}"
)

print(
    f"Critical priority procedures: "
    f"{int(priority_counts.get('CRITICAL PRIORITY', 0)):,}"
)

print(
    f"High priority procedures: "
    f"{int(priority_counts.get('HIGH PRIORITY', 0)):,}"
)


# ======================================================================
# 40. DATASETS CREATED
# ======================================================================

print("\nANALYTICAL DATASETS CREATED:")
print("  - procedure_i")
print("  - procedure_summary")
print("  - top_cost_procedures")
print("  - top_non_431_cost")
print("  - top_utilization")
print("  - top_cost_per_patient")
print("  - high_repeat_utilization")
print("  - high_cost_per_patient")
print("  - high_cost_non431_procedures")
print("  - high_cost_high_utilization")
print("  - high_cost_low_utilization")
print("  - high_utilization_low_cost")
print("  - procedures_431_dominated")
print("  - patient_procedure_summary")
print("  - high_procedure_burden_patients")
print("  - procedure_concentration")
print("  - procedure_priority")
print("  - procedure_watchlist")
print("  - top_executive_procedures")
print("  - priority_distribution")
print("  - cost_intensity_distribution")
print("  - utilization_intensity_distribution")
print("  - insight_metrics")
print("  - procedure_insight_summary")
print("  - procedure_i_validation")

print("\nSTATUS: PASS")
print("=" * 70)
```

```python
# ============================================================
# STEP 4.5 — ENCOUNTER DURATION OUTLIER VALIDATION
# ============================================================

duration_outliers = (
    encounters_clean[
        encounters_clean["DURATION_DAYS"] > 7
    ]
    [
        [
            "Id",
            "PATIENT",
            "ENCOUNTERCLASS",
            "DESCRIPTION",
            "START",
            "STOP",
            "DURATION_DAYS",
            "TOTAL_CLAIM_COST",
            "PAYER_COVERAGE",
            "PATIENT_RESPONSIBILITY"
        ]
    ]
    .sort_values(
        "DURATION_DAYS",
        ascending=False
    )
)

print("Long encounters (>7 days):", len(duration_outliers))

duration_outliers.head(20)
```

```python
# ============================================================
# LONG ENCOUNTERS BY ENCOUNTER TYPE
# ============================================================

long_duration_by_type = (
    duration_outliers
    .groupby("ENCOUNTERCLASS")
    .agg(
        Long_Encounter_Count=("Id", "count"),
        Average_Duration_Days=("DURATION_DAYS", "mean"),
        Maximum_Duration_Days=("DURATION_DAYS", "max"),
        Total_Claim_Cost=("TOTAL_CLAIM_COST", "sum")
    )
    .sort_values(
        "Long_Encounter_Count",
        ascending=False
    )
)

long_duration_by_type
```

```python
# ============================================================
# DURATION OUTLIER SEVERITY
# ============================================================

duration_outliers["DURATION_CATEGORY"] = pd.cut(
    duration_outliers["DURATION_DAYS"],
    bins=[0, 1, 7, 30, 90, 365, float("inf")],
    labels=[
        "≤1 day",
        "2–7 days",
        "8–30 days",
        "31–90 days",
        "91–365 days",
        ">365 days"
    ],
    include_lowest=True
)

duration_outlier_summary = (
    duration_outliers["DURATION_CATEGORY"]
    .value_counts()
    .sort_index()
)

duration_outlier_summary
```

```python
# ============================================================
# STEP 4.6 — ENCOUNTER DURATION QUALITY FLAG
# ============================================================

encounters_clean["DURATION_FLAG"] = pd.cut(
    encounters_clean["DURATION_DAYS"],
    bins=[
        -float("inf"),
        1,
        7,
        30,
        90,
        365,
        float("inf")
    ],
    labels=[
        "Normal: ≤1 day",
        "Extended: 2–7 days",
        "Long: 8–30 days",
        "Very Long: 31–90 days",
        "Extreme: 91–365 days",
        "Extreme: >365 days"
    ]
)

duration_flag_summary = (
    encounters_clean["DURATION_FLAG"]
    .value_counts()
    .sort_index()
)

duration_flag_summary
```

```python
# ============================================================
# DURATION FLAG PERCENTAGES
# ============================================================

duration_flag_percentages = (
    encounters_clean["DURATION_FLAG"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .sort_index()
)

duration_flag_percentages
```

```python
# ============================================================
# STEP 4.7 — DURATION PROFILE BY ENCOUNTER TYPE
# ============================================================

duration_profile = (
    encounters_clean
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Average_Duration_Days=("DURATION_DAYS", "mean"),
        Median_Duration_Days=("DURATION_DAYS", "median"),
        P95_Duration_Days=(
            "DURATION_DAYS",
            lambda x: x.quantile(0.95)
        ),
        Maximum_Duration_Days=("DURATION_DAYS", "max"),
        Total_Claim_Cost=("TOTAL_CLAIM_COST", "sum")
    )
    .sort_values(
        "Median_Duration_Days",
        ascending=False
    )
)

duration_profile.round(2)
```

```python
# ============================================================
# STEP 4.13 — PATIENT-LEVEL CLINICAL UTILIZATION
# ============================================================

patient_utilization = (
    encounters_clean
    .groupby("PATIENT")
    .agg(
        Encounter_Count=("Id", "count"),
        Total_Encounter_Cost=("TOTAL_CLAIM_COST", "sum"),
        Total_Payer_Coverage=("PAYER_COVERAGE", "sum"),
        Total_Patient_Responsibility=("PATIENT_RESPONSIBILITY", "sum"),
        Inpatient_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "inpatient").sum()
        ),
        Emergency_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "emergency").sum()
        ),
        Outpatient_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "outpatient").sum()
        ),
        Ambulatory_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "ambulatory").sum()
        ),
        UrgentCare_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "urgentcare").sum()
        ),
        Wellness_Encounters=(
            "ENCOUNTERCLASS",
            lambda x: (x == "wellness").sum()
        )
    )
    .reset_index()
)

patient_utilization.head()
```

```python
# ============================================================
# STEP 4.13.1 — HIGHEST-UTILIZATION PATIENTS
# ============================================================

high_utilization_patients = (
    patient_utilization[
        [
            "PATIENT",
            "Encounter_Count",
            "Inpatient_Encounters",
            "Emergency_Encounters",
            "Outpatient_Encounters",
            "Ambulatory_Encounters",
            "UrgentCare_Encounters",
            "Wellness_Encounters",
            "Total_Encounter_Cost",
            "Total_Patient_Responsibility"
        ]
    ]
    .sort_values(
        "Encounter_Count",
        ascending=False
    )
)

high_utilization_patients.head(20)
```

```python
# ============================================================
# STEP 4.13.2 — UTILIZATION CONCENTRATION
# ============================================================

total_encounters = patient_utilization["Encounter_Count"].sum()

patient_utilization_ranked = (
    patient_utilization
    .sort_values("Encounter_Count", ascending=False)
    .reset_index(drop=True)
)

concentration_results = []

for pct in [0.01, 0.05, 0.10, 0.25]:
    
    n_patients = max(1, round(len(patient_utilization_ranked) * pct))
    
    top_encounters = (
        patient_utilization_ranked
        .head(n_patients)["Encounter_Count"]
        .sum()
    )
    
    encounter_share = (
        top_encounters / total_encounters
    ) * 100
    
    concentration_results.append({
        "Patient_Segment": f"Top {pct:.0%}",
        "Patients": n_patients,
        "Encounters": top_encounters,
        "Encounter_Share_Percentage": round(encounter_share, 2)
    })

utilization_concentration = pd.DataFrame(
    concentration_results
)

utilization_concentration
```

```python
# ============================================================
# STEP 4.13.3 — CLINICAL UTILIZATION INTENSITY
# ============================================================

patient_utilization["High_Acuity_Encounters"] = (
    patient_utilization["Inpatient_Encounters"]
    + patient_utilization["Emergency_Encounters"]
)

patient_utilization["Emergency_Rate"] = (
    patient_utilization["Emergency_Encounters"]
    / patient_utilization["Encounter_Count"]
    * 100
).round(2)

patient_utilization["Inpatient_Rate"] = (
    patient_utilization["Inpatient_Encounters"]
    / patient_utilization["Encounter_Count"]
    * 100
).round(2)

patient_utilization["High_Acuity_Rate"] = (
    patient_utilization["High_Acuity_Encounters"]
    / patient_utilization["Encounter_Count"]
    * 100
).round(2)

patient_utilization[
    [
        "PATIENT",
        "Encounter_Count",
        "Inpatient_Encounters",
        "Emergency_Encounters",
        "High_Acuity_Encounters",
        "Emergency_Rate",
        "Inpatient_Rate",
        "High_Acuity_Rate"
    ]
].sort_values(
    "High_Acuity_Rate",
    ascending=False
).head(20)
```

```python
# Cost composition of high-utilization + high-cost patients

high_risk_patient_ids = high_risk_high_cost["PATIENT"]

high_risk_encounters = encounters_clean[
    encounters_clean["PATIENT"].isin(high_risk_patient_ids)
].copy()

cost_by_encounter_type = (
    high_risk_encounters
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Total_Cost=("TOTAL_CLAIM_COST", "sum"),
        Patient_Responsibility=("PATIENT_RESPONSIBILITY", "sum")
    )
    .sort_values("Total_Cost", ascending=False)
)

cost_by_encounter_type["Cost_Share_Percentage"] = (
    cost_by_encounter_type["Total_Cost"]
    / cost_by_encounter_type["Total_Cost"].sum()
    * 100
)

cost_by_encounter_type
```

```python
# Calculate average cost and patient responsibility per encounter type

cost_intensity = (
    high_risk_encounters
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Total_Cost=("TOTAL_CLAIM_COST", "sum"),
        Total_Patient_Responsibility=("PATIENT_RESPONSIBILITY", "sum")
    )
)

cost_intensity["Average_Cost_Per_Encounter"] = (
    cost_intensity["Total_Cost"]
    / cost_intensity["Encounter_Count"]
)

cost_intensity["Average_Patient_Responsibility_Per_Encounter"] = (
    cost_intensity["Total_Patient_Responsibility"]
    / cost_intensity["Encounter_Count"]
)

cost_intensity = cost_intensity.sort_values(
    "Average_Cost_Per_Encounter",
    ascending=False
)

cost_intensity
```

```python
# Rank high-utilization + high-cost patients by financial burden

financial_burden_high_risk = (
    high_risk_high_cost[
        [
            "PATIENT",
            "Encounter_Count",
            "Total_Encounter_Cost",
            "Total_Payer_Coverage",
            "Total_Patient_Responsibility"
        ]
    ]
    .copy()
)

financial_burden_high_risk["Patient_Responsibility_Rate"] = (
    financial_burden_high_risk["Total_Patient_Responsibility"]
    / financial_burden_high_risk["Total_Encounter_Cost"]
    * 100
)

financial_burden_high_risk = (
    financial_burden_high_risk
    .sort_values(
        "Total_Patient_Responsibility",
        ascending=False
    )
)

financial_burden_high_risk.head(20)
```

```python
# Segment patients by utilization and cost

patient_risk_segments = patient_utilization.copy()

patient_risk_segments["Risk_Segment"] = np.select(
    [
        (patient_risk_segments["Encounter_Count"] > high_utilization_threshold) &
        (patient_risk_segments["Total_Encounter_Cost"] > high_cost_threshold),

        (patient_risk_segments["Encounter_Count"] > high_utilization_threshold) &
        (patient_risk_segments["Total_Encounter_Cost"] <= high_cost_threshold),

        (patient_risk_segments["Encounter_Count"] <= high_utilization_threshold) &
        (patient_risk_segments["Total_Encounter_Cost"] > high_cost_threshold),

        (patient_risk_segments["Encounter_Count"] <= high_utilization_threshold) &
        (patient_risk_segments["Total_Encounter_Cost"] <= high_cost_threshold)
    ],
    [
        "High Utilization + High Cost",
        "High Utilization + Lower Cost",
        "Lower Utilization + High Cost",
        "Lower Utilization + Lower Cost"
    ],
    default="Unclassified"
)

risk_segment_summary = (
    patient_risk_segments
    .groupby("Risk_Segment")
    .agg(
        Patients=("PATIENT", "count"),
        Total_Encounters=("Encounter_Count", "sum"),
        Total_Cost=("Total_Encounter_Cost", "sum"),
        Total_Patient_Responsibility=("Total_Patient_Responsibility", "sum")
    )
    .sort_values("Total_Cost", ascending=False)
)

risk_segment_summary["Patient_Share_Percentage"] = (
    risk_segment_summary["Patients"] /
    len(patient_risk_segments) * 100
).round(2)

risk_segment_summary["Cost_Share_Percentage"] = (
    risk_segment_summary["Total_Cost"] /
    risk_segment_summary["Total_Cost"].sum() * 100
).round(2)

risk_segment_summary
```

```python
# Profile the high-utilization + high-cost patient segment

high_risk_profile = patient_utilization[
    (patient_utilization["Encounter_Count"] > high_utilization_threshold) &
    (patient_utilization["Total_Encounter_Cost"] > high_cost_threshold)
].copy()

high_risk_profile_summary = pd.DataFrame({
    "Metric": [
        "Patients",
        "Average Encounters per Patient",
        "Median Encounters per Patient",
        "Average Total Cost per Patient",
        "Median Total Cost per Patient",
        "Average Patient Responsibility",
        "Median Patient Responsibility",
        "Average High-Acuity Rate"
    ],
    "Value": [
        len(high_risk_profile),
        high_risk_profile["Encounter_Count"].mean(),
        high_risk_profile["Encounter_Count"].median(),
        high_risk_profile["Total_Encounter_Cost"].mean(),
        high_risk_profile["Total_Encounter_Cost"].median(),
        high_risk_profile["Total_Patient_Responsibility"].mean(),
        high_risk_profile["Total_Patient_Responsibility"].median(),
        high_risk_profile["High_Acuity_Rate"].mean()
    ]
})

high_risk_profile_summary
```

Patient utilization and cost were highly concentrated. Only 3.70% of patients were classified as high-utilization/high-cost, yet they accounted for 44.45% of total healthcare expenditure. A further 6.37% of patients were lower-utilization but high-cost, contributing another 26.04% of total costs. Combined, approximately 10% of patients generated over 70% of total healthcare costs. Meanwhile, high-utilization/low-cost patients represented 6.37% of the population and approximately 30% of encounters but only 4.68% of total costs. This demonstrates that healthcare utilization volume alone is insufficient for identifying financially high-impact patients and supports a segmented approach to care management and resource allocation.

```python
# STEP 6.4 — Overall financial burden and high-risk comparison

overall_financial_summary = pd.DataFrame({
    "Metric": [
        "Total Healthcare Cost",
        "Total Payer Coverage",
        "Total Patient Responsibility",
        "Overall Patient Responsibility Rate",
        "High-Risk Healthcare Cost",
        "High-Risk Payer Coverage",
        "High-Risk Patient Responsibility",
        "High-Risk Patient Responsibility Rate"
    ],
    "Value": [
        patient_utilization["Total_Encounter_Cost"].sum(),
        patient_utilization["Total_Payer_Coverage"].sum(),
        patient_utilization["Total_Patient_Responsibility"].sum(),
        (
            patient_utilization["Total_Patient_Responsibility"].sum()
            / patient_utilization["Total_Encounter_Cost"].sum()
            * 100
        ),
        high_util_high_cost["Total_Encounter_Cost"].sum(),
        high_util_high_cost["Total_Payer_Coverage"].sum(),
        high_util_high_cost["Total_Patient_Responsibility"].sum(),
        (
            high_util_high_cost["Total_Patient_Responsibility"].sum()
            / high_util_high_cost["Total_Encounter_Cost"].sum()
            * 100
        )
    ]
})

overall_financial_summary
```

```python
# STEP 6.5 — Cost concentration among high-cost patients

high_cost_segments = patient_risk_segments[
    patient_risk_segments["Risk_Segment"].isin([
        "High Utilization + High Cost",
        "Lower Utilization + High Cost"
    ])
].copy()

high_cost_concentration = pd.DataFrame({
    "Metric": [
        "Total Patients",
        "Patient Share (%)",
        "Total Healthcare Cost",
        "Cost Share (%)"
    ],
    "Value": [
        high_cost_segments["PATIENT"].nunique(),
        (
            high_cost_segments["PATIENT"].nunique()
            / patient_risk_segments["PATIENT"].nunique()
            * 100
        ),
        high_cost_segments["Total_Encounter_Cost"].sum(),
        (
            high_cost_segments["Total_Encounter_Cost"].sum()
            / patient_risk_segments["Total_Encounter_Cost"].sum()
            * 100
        )
    ]
})

high_cost_concentration
```

```python
# STEP 6.6 — Profile the four patient risk segments

risk_segment_profile = (
    patient_risk_segments
    .groupby("Risk_Segment")
    .agg(
        Patients=("PATIENT", "nunique"),
        Avg_Encounters_Per_Patient=("Encounter_Count", "mean"),
        Median_Encounters_Per_Patient=("Encounter_Count", "median"),
        Avg_Healthcare_Cost=("Total_Encounter_Cost", "mean"),
        Median_Healthcare_Cost=("Total_Encounter_Cost", "median"),
        Avg_Patient_Responsibility=("Total_Patient_Responsibility", "mean"),
        Median_Patient_Responsibility=("Total_Patient_Responsibility", "median"),
        Avg_High_Acuity_Rate=("High_Acuity_Rate", "mean")
    )
    .reset_index()
)

risk_segment_profile
```

```python
# STEP 6.2 — High-Risk Patient Characteristics

# Identify the 36 high-utilization + high-cost patients
high_risk_patients = patient_utilization[
    (patient_utilization["Encounter_Count"] > high_utilization_threshold) &
    (patient_utilization["Total_Encounter_Cost"] > high_cost_threshold)
].copy()

# Create a profile of the high-risk group
high_risk_profile = pd.DataFrame({
    "Metric": [
        "Patients",
        "Average Encounters",
        "Median Encounters",
        "Average Healthcare Cost",
        "Median Healthcare Cost",
        "Average Patient Responsibility",
        "Average Emergency Rate",
        "Average Inpatient Rate",
        "Average High-Acuity Rate"
    ],
    "Value": [
        len(high_risk_patients),
        high_risk_patients["Encounter_Count"].mean(),
        high_risk_patients["Encounter_Count"].median(),
        high_risk_patients["Total_Encounter_Cost"].mean(),
        high_risk_patients["Total_Encounter_Cost"].median(),
        high_risk_patients["Total_Patient_Responsibility"].mean(),
        high_risk_patients["Emergency_Rate"].mean(),
        high_risk_patients["Inpatient_Rate"].mean(),
        high_risk_patients["High_Acuity_Rate"].mean()
    ]
})

high_risk_profile
```

```python
# Create comparison group
non_high_risk = patient_utilization[
    ~patient_utilization["PATIENT"].isin(high_risk_patients["PATIENT"])
].copy()

comparison = pd.DataFrame({
    "Metric": [
        "Average Encounters",
        "Average Healthcare Cost",
        "Average Patient Responsibility",
        "Average Emergency Rate",
        "Average Inpatient Rate",
        "Average High-Acuity Rate"
    ],
    "High_Risk": [
        high_risk_patients["Encounter_Count"].mean(),
        high_risk_patients["Total_Encounter_Cost"].mean(),
        high_risk_patients["Total_Patient_Responsibility"].mean(),
        high_risk_patients["Emergency_Rate"].mean(),
        high_risk_patients["Inpatient_Rate"].mean(),
        high_risk_patients["High_Acuity_Rate"].mean()
    ],
    "Other_Patients": [
        non_high_risk["Encounter_Count"].mean(),
        non_high_risk["Total_Encounter_Cost"].mean(),
        non_high_risk["Total_Patient_Responsibility"].mean(),
        non_high_risk["Emergency_Rate"].mean(),
        non_high_risk["Inpatient_Rate"].mean(),
        non_high_risk["High_Acuity_Rate"].mean()
    ]
})

comparison
```

```python
# STEP 6.3 — High-Risk Encounter Mix and Cost Drivers

# Get all encounters belonging to high-risk patients
high_risk_ids = high_risk_patients["PATIENT"]

high_risk_encounters = encounters_clean[
    encounters_clean["PATIENT"].isin(high_risk_ids)
].copy()

# Summarize encounter types
high_risk_encounter_mix = (
    high_risk_encounters
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("Id", "count"),
        Total_Cost=("TOTAL_CLAIM_COST", "sum"),
        Patient_Responsibility=("PATIENT_RESPONSIBILITY", "sum")
    )
    .reset_index()
)

# Calculate percentages
high_risk_encounter_mix["Encounter_Share_Percentage"] = (
    high_risk_encounter_mix["Encounter_Count"]
    / high_risk_encounter_mix["Encounter_Count"].sum()
    * 100
)

high_risk_encounter_mix["Cost_Share_Percentage"] = (
    high_risk_encounter_mix["Total_Cost"]
    / high_risk_encounter_mix["Total_Cost"].sum()
    * 100
)

# Average cost per encounter
high_risk_encounter_mix["Average_Cost_Per_Encounter"] = (
    high_risk_encounter_mix["Total_Cost"]
    / high_risk_encounter_mix["Encounter_Count"]
)

high_risk_encounter_mix.sort_values(
    "Total_Cost",
    ascending=False
)
```

```python
# STEP 6.4 — Highest-Burden High-Risk Patients

high_risk_patient_profile = high_risk_patients[
    [
        "PATIENT",
        "Encounter_Count",
        "Total_Encounter_Cost",
        "Total_Payer_Coverage",
        "Total_Patient_Responsibility",
        "Inpatient_Encounters",
        "Emergency_Encounters",
        "Outpatient_Encounters",
        "Ambulatory_Encounters",
        "UrgentCare_Encounters",
        "Wellness_Encounters",
        "High_Acuity_Encounters",
        "Emergency_Rate",
        "Inpatient_Rate",
        "High_Acuity_Rate"
    ]
].copy()

# Rank by total healthcare cost
high_risk_patient_profile = (
    high_risk_patient_profile
    .sort_values(
        "Total_Encounter_Cost",
        ascending=False
    )
    .reset_index(drop=True)
)

high_risk_patient_profile.head(10)
```

```python
# Top 10 high-risk patient cost concentration

top_10_cost = high_risk_patient_profile.head(10)["Total_Encounter_Cost"].sum()

high_risk_total_cost = high_risk_patient_profile[
    "Total_Encounter_Cost"
].sum()

top_10_cost_share = (
    top_10_cost / high_risk_total_cost
) * 100

print(f"Top 10 high-risk patient cost: ${top_10_cost:,.2f}")
print(f"High-risk total cost: ${high_risk_total_cost:,.2f}")
print(f"Top 10 cost share: {top_10_cost_share:.2f}%")
```

```python
# STEP 6.5 — High-Risk Utilization Summary

high_risk_summary = pd.DataFrame({
    "Metric": [
        "High-Risk Patients",
        "High-Risk Patient Share",
        "High-Risk Total Cost",
        "High-Risk Cost Share",
        "High-Risk Patient Responsibility",
        "High-Risk Patient Responsibility Rate",
        "Top 10 High-Risk Cost",
        "Top 10 Cost Share of High-Risk Group",
        "Average Encounters per High-Risk Patient",
        "Average Healthcare Cost per High-Risk Patient",
        "Average Patient Responsibility per High-Risk Patient",
        "Average High-Acuity Rate"
    ],
    "Value": [
        len(high_risk_patients),
        len(high_risk_patients) / len(patient_utilization) * 100,
        high_risk_patients["Total_Encounter_Cost"].sum(),
        high_risk_patients["Total_Encounter_Cost"].sum()
        / patient_utilization["Total_Encounter_Cost"].sum() * 100,
        high_risk_patients["Total_Patient_Responsibility"].sum(),
        high_risk_patients["Total_Patient_Responsibility"].sum()
        / high_risk_patients["Total_Encounter_Cost"].sum() * 100,
        high_risk_patient_profile.head(10)["Total_Encounter_Cost"].sum(),
        high_risk_patient_profile.head(10)["Total_Encounter_Cost"].sum()
        / high_risk_patients["Total_Encounter_Cost"].sum() * 100,
        high_risk_patients["Encounter_Count"].mean(),
        high_risk_patients["Total_Encounter_Cost"].mean(),
        high_risk_patients["Total_Patient_Responsibility"].mean(),
        high_risk_patients["High_Acuity_Rate"].mean()
    ]
})

high_risk_summary
```

```python
# STEP 7.1 — High-risk encounter-type cost analysis

# Get encounter IDs belonging to high-risk patients
high_risk_patient_ids = high_util_high_cost["PATIENT"]

# Filter encounter-level data
high_risk_encounters = encounters_clean[
    encounters_clean["PATIENT"].isin(high_risk_patient_ids)
].copy()

# Analyze encounter class
high_risk_encounter_class = (
    high_risk_encounters
    .groupby("ENCOUNTERCLASS")
    .agg(
        Encounter_Count=("ENCOUNTERCLASS", "size"),
        Total_Cost=("TOTAL_CLAIM_COST", "sum"),
        Patient_Responsibility=("TOTAL_CLAIM_COST", "sum")
    )
    .sort_values("Total_Cost", ascending=False)
)

# Calculate percentages
high_risk_encounter_class["Encounter_Share_Percentage"] = (
    high_risk_encounter_class["Encounter_Count"]
    / high_risk_encounter_class["Encounter_Count"].sum()
    * 100
)

high_risk_encounter_class["Cost_Share_Percentage"] = (
    high_risk_encounter_class["Total_Cost"]
    / high_risk_encounter_class["Total_Cost"].sum()
    * 100
)

high_risk_encounter_class["Average_Cost_Per_Encounter"] = (
    high_risk_encounter_class["Total_Cost"]
    / high_risk_encounter_class["Encounter_Count"]
)

high_risk_encounter_class
```

```python
# STEP 7.2 — High-risk utilization profile

high_risk_utilization_profile = high_util_high_cost[
    [
        "PATIENT",
        "Encounter_Count",
        "Total_Encounter_Cost",
        "Total_Patient_Responsibility",
        "Inpatient_Encounters",
        "Emergency_Encounters",
        "Outpatient_Encounters",
        "Ambulatory_Encounters",
        "UrgentCare_Encounters",
        "Wellness_Encounters",
        "High_Acuity_Encounters",
        "Emergency_Rate",
        "Inpatient_Rate",
        "High_Acuity_Rate"
    ]
].copy()

high_risk_utilization_profile.head(10)
```

```python
# Aggregate encounter types across high-risk patients

high_risk_utilization_by_type = pd.DataFrame({
    "Encounter_Type": [
        "Inpatient",
        "Emergency",
        "Outpatient",
        "Ambulatory",
        "Urgent Care",
        "Wellness"
    ],
    "Encounter_Count": [
        high_util_high_cost["Inpatient_Encounters"].sum(),
        high_util_high_cost["Emergency_Encounters"].sum(),
        high_util_high_cost["Outpatient_Encounters"].sum(),
        high_util_high_cost["Ambulatory_Encounters"].sum(),
        high_util_high_cost["UrgentCare_Encounters"].sum(),
        high_util_high_cost["Wellness_Encounters"].sum()
    ]
})

high_risk_utilization_by_type["Utilization_Share_Percentage"] = (
    high_risk_utilization_by_type["Encounter_Count"]
    / high_risk_utilization_by_type["Encounter_Count"].sum()
    * 100
)

high_risk_utilization_by_type = (
    high_risk_utilization_by_type
    .sort_values("Encounter_Count", ascending=False)
    .reset_index(drop=True)
)

high_risk_utilization_by_type
```

```python
# Check that high-risk encounter counts reconcile
print(
    "Total high-risk encounters:",
    high_risk_utilization_by_type["Encounter_Count"].sum()
)

print(
    "Expected high-risk encounters:",
    high_util_high_cost["Encounter_Count"].sum()
)
```
